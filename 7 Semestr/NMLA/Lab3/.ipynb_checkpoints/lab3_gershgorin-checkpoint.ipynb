# Лаба: Метод Данилевского для симметрической матрицы + график по кольцам Гершгорина

В этой версии оставлены только необходимые части:
- Генерация симметричной матрицы.
- Метод Данилевского → матрица Фробениуса и коэффициенты характеристического многочлена.
- Функции для значений/векторов собственных и ортогональности (для проверки корректности).
- Интервалы Гершгорина.
- **НОВОЕ:** Функция построения графика характеристического многочлена на объединённом отрезке по кольцам Гершгорина с отметкой самих отрезков.

Удалено как лишнее в данной интерпретации:
- `euclidean_norm` (используем `LinearAlgebra.norm`).
- `identity_matrix` (используем стандартную `I`).
- `union_intervals` и текстовая проверка `check_gershorin` (вместо неё — графическая визуализация по Гершгорину).
- Демонстрационные ячейки с «n = 4», «n = 10» и печатью промежуточных матриц.
- Проверка Виета (`check_vieta`) — опционально, можно добавить обратно при необходимости, но для данной версии считаем избыточной.


```

using Random
using LinearAlgebra
using Polynomials
using Plots

# Генерация симметричной матрицы
function generate_symmetric_matrix(l::Float64, r::Float64, n::Int)
    A = rand(n, n) .* (r - l) .+ l
    return (A + A') / 2
end

# Метод Данилевского. Возвращает:
#   Bs — произведение преобразований
#   F  — приведённая к форме Фробениуса матрица
function danilevsky_algo(A::Matrix{Float64})
    n = size(A, 1)
    Bs = Matrix(I, n, n)
    D  = copy(A)

    for i in n:-1:2
        B = Matrix(I, n, n)
        d = D[i, i - 1]
        # Строка i-1 заполняется
        B[i-1, :] .= -D[i, :] ./ d
        B[i-1, i-1] = 1 / d

        D = inv(B) * D * B
        Bs = Bs * B
    end
    return Bs, D
end

# Коэффициенты характеристического многочлена из формы Фробениуса
# Если F — матрица Фробениуса, то её первая строка задаёт коэффициенты с обратными знаками.
function characteristic_coeffs_from_frobenius(F::Matrix{Float64})
    coeffs = -copy(F[1, :])              # [-a1, -a2, ..., -an]
    coeffs = vcat([1.0], coeffs)         # [1, -a1, -a2, ..., -an]
    return coeffs
end

# Значение многочлена в точке
function poly_eval(coeffs::Vector{Float64}, x::Float64)
    # coeffs предполагаются в порядке степеней по убыванию: [a0, a1, ..., an], где p(λ)=a0*λ^n + ... + an
    # Мы собрали [1, c2, ..., cn+1] как для Polynomials: Polynomial(coeffs) ожидает по возрастанию степеней.
    # Поэтому сделаем явную оценку по схеме Горнера:
    val = 0.0
    for c in coeffs
        val = val * x + c
    end
    return val
end

# Поиск собственных векторов по найденным значениям (через Bs из Данилевского)
function find_eigen_vectors(vals::Vector{Float64}, Bs::Matrix{Float64})
    n = size(Bs, 1)
    m = length(vals)
    # Базис вида [λ^{n-1}, ..., λ, 1] для каждого λ
    Y = zeros(m, n)
    for i in 1:m
        λ = vals[i]
        for j in 0:n-1
            Y[i, j+1] = λ^(n-1-j)
        end
    end
    X = Y * Bs'  # преобразуем базисные столбцы
    # Нормировка строк как векторов
    for i in 1:m
        X[i, :] ./= norm(X[i, :])
    end
    return X
end

# Проверка ортогональности собственных векторов
function check_orthogonal(vects::Matrix{Float64}; tol=1e-6)
    m = size(vects, 1)
    for i in 1:m-1, j in i+1:m
        s = dot(vects[i, :], vects[j, :])
        if abs(s) > tol
            println("Векторы i=$i и j=$j НЕ ортогональны (|dot| = ", abs(s), ")")
            return false
        end
    end
    println("Собственные векторы ортогональны (в пределах допуска).")
    return true
end

# Интервалы Гершгорина (строчные)
function gershgorin_intervals(A::Matrix{Float64})
    n = size(A, 1)
    intervals = Vector{Tuple{Float64,Float64}}()
    for i in 1:n
        r = sum(abs.(A[i, :])) - abs(A[i, i])
        c = A[i, i]
        push!(intervals, (c - r, c + r))
    end
    return intervals
end

# Построение графика характеристического многочлена на объединённом отрезке по кольцам Гершгорина
function plot_gershgorin_characteristic(A::Matrix{Float64}; num_points::Int=2000, margin::Float64=0.1)
    # Данилевский → коэффициенты многочлена
    Bs, F = danilevsky_algo(A)
    coeffs = characteristic_coeffs_from_frobenius(F)

    # Интервалы Гершгорина и общий отрезок для графика
    ints = gershgorin_intervals(A)
    xmin = minimum(first.(ints))
    xmax = maximum(last.(ints))
    L = xmax - xmin
    xmin -= margin*L
    xmax += margin*L

    xs = range(xmin, xmax, length=num_points)
    ys = [poly_eval(coeffs, x) for x in xs]

    plt = plot(xs, ys, xlabel="λ", ylabel="p(λ)=det(λI - A)",
               legend=false, title="Характеристический многочлен и интервалы Гершгорина")
    hline!([0.0])

    # Отметим интервалы Гершгорина на оси x (тонкими вертикальными рисками по y≈0)
    for (l, r) in ints
        # Рисуем небольшой отрезок на уровне y=0
        plot!([l, r], [0.0, 0.0], linewidth=4)
    end

    display(plt)
    return plt
end
```

## Мини-демо

Ниже — пример генерации симметричной матрицы и построения графика по кольцам Гершгорина.


```

# Пример
Random.seed!(42)
n = 5
A = generate_symmetric_matrix(-2.0, 2.0, n)

# Построить график
plot_gershgorin_characteristic(A)

# Проверка ортогональности на найденных собственных векторах (через библиотеку для стабильности)
λs = eigvals(Symmetric(A))
Bs, F = danilevsky_algo(A)
vecs = find_eigen_vectors(real(λs), Bs) # в учебных целях, без упорядочивания
check_orthogonal(vecs)
```
