```julia
#Сфера
using PyPlot

u = range(-π, π; length=400)
v = range(-π/2, π/2; length=200)

X = cos.(u') .* cos.(v)
Y = sin.(u') .* cos.(v)
Z = sin.(v) .* ones(length(u))'

surf(X, Y, Z; alpha=0.9, cmap="inferno")
title("Сфера")

PyPlot.display_figs()
```

```julia
# Поверхность Дини
using PyPlot

u = range(0, 4π; length=400)
v = range(0.001, 2; length=200)

U = u'
V = v
X = cos.(U) .* sin.(V)
Y = sin.(U) .* sin.(V)
Z = cos.(V) .+ log.(tan.(V ./ 2)) .+ 0.2 .* U .- 4

surf(X, Y, Z; alpha=0.9, cmap="viridis")
ax = gca()
ax.view_init(elev=20, azim=20)

title("Поверхность Дини")

PyPlot.display_figs()
```

```julia
#Трилистник
using PyPlot

u = range(-2π, 2π; length=400)
v = range(-π, π; length=200)

U = u'
V = v

X = cos.(U) .* cos.(V) .+ 3 .* cos.(U) .* (1.5 .+ sin.(1.5 .* U))
Y = sin.(U) .* cos.(V) .+ 3 .* sin.(U) .* (1.5 .+ sin.(1.5 .* U))
Z = sin.(V) .+ 2 .* cos.(1.5 .* U)

surf(X, Y, Z; alpha=0.9, cmap="viridis")
title("Трилистник")

ax = gca()
ax.view_init(elev=20, azim=45)

PyPlot.display_figs()
```
