```julia
using Random 
using LinearAlgebra
using PyPlot
```

```julia
function gen_matr(n)
    return rand(n,n)
end
```

```julia
A = gen_matr(3)
B = gen_matr(3)
```

```julia
function simple_multiply(A,B)  
    n1,m1 = size(A)
    n2,m2 = size(B)
    C = zeros(n1,m2)
    for i in 1:n1
        for j in 1:m2
            for k in 1:m1
                C[i,j] += A[i,k] * B[k,j]
            end
        end
    end
    return C
end
```

```julia
simple_multiply(A,B)
```

```julia
function strass_rec(A, B, n_min)
    n = size(A, 1)

    if n <= n_min
        return A * B
    end
    
    half = div(n, 2)
    
    A11 = A[1:half, 1:half]
    A12 = A[1:half, half+1:end]
    A21 = A[half+1:end, 1:half]
    A22 = A[half+1:end, half+1:end]
    B11 = B[1:half, 1:half]
    B12 = B[1:half, half+1:end]
    B21 = B[half+1:end, 1:half]
    B22 = B[half+1:end, half+1:end]
    
    P1 = strass_rec(A11 + A22, B11 + B22, n_min)
    P2 = strass_rec(A21 + A22, B11, n_min)
    P3 = strass_rec(A11, B12 - B22, n_min)
    P4 = strass_rec(A22, B21 - B11, n_min)
    P5 = strass_rec(A11 + A12, B22, n_min)
    P6 = strass_rec(A21 - A11, B11 + B12, n_min)
    P7 = strass_rec(A12 - A22, B21 + B22, n_min)
    
    C11 = P1 + P4 - P5 + P7
    C12 = P3 + P5
    C21 = P2 + P4
    C22 = P1 - P2 + P3 + P6
    
    C = zeros(n, n)
    C[1:half, 1:half] = C11
    C[1:half, half+1:end] = C12
    C[half+1:end, 1:half] = C21
    C[half+1:end, half+1:end] = C22

    return C
end
```

```julia
function strassen(A, B)
    m, n = size(A)
    n2, p = size(B)

    max_size = nextpow(2, max(m, n, p))
    A_padded = zeros(max_size, max_size)
    B_padded = zeros(max_size, max_size)
    A_padded[1:m, 1:n] = A
    B_padded[1:n, 1:p] = B
    
    C_padded = strass_rec(A_padded, B_padded, 64)
    return C_padded[1:m, 1:p]
end
```

```julia
strassen(A, B)
```

```julia
function vinograd(A, B)
    m, n = size(A)
    n, p = size(B)
    
    one_minus = zeros(m)
    for i in 1:m
        for k in 1:2:div(n-1, 2)*2
            one_minus[i] += A[i, k] * A[i, k+1]
        end
    end
    
    two_minus = zeros(p)
    for j in 1:p
        for k in 1:2:div(n-1, 2)*2
            two_minus[j] += B[k, j] * B[k+1, j]
        end
    end
    
    C = zeros(m, p)
    for i in 1:m
        for j in 1:p
            C[i, j] = - one_minus[i] - two_minus[j]
            for k in 1:2:div(n-1, 2)*2
                C[i, j] += (A[i, k] + B[k+1, j]) * (A[i, k+1] + B[k, j])
            end
        end
    end
    
    if isodd(n)
        for i in 1:m
            for j in 1:p
                C[i, j] += A[i, n] * B[n, j]
            end
        end
    end
    
    return C
end
```

```julia
vinograd(A, B)
```

```julia
function gen_matr(n)
    return 1.0 .+ 9.0 * rand(n, n)
end
```

```julia
n_values = collect(2:100:300)
times = Float64[]
timesV = Float64[]
timesStr= Float64[]

for n in n_values
    Random.seed!(0)
    A = gen_matr(n)
    B = gen_matr(n)
    
    t_standard =  @elapsed simple_multiply(A, B)
    t_vinograd =  @elapsed vinograd(A, B)
    t_strassen =  @elapsed strassen(A, B)
    
    push!(times, t_standard)
    push!(timesV, t_vinograd)
    push!(timesStr, t_strassen)
end
```

```julia
PyPlot.figure(figsize=(6, 5))
PyPlot.xlabel("n")
PyPlot.ylabel("time")
PyPlot.plot(n_values, times, label="Ordinary", linewidth=2)
PyPlot.plot(n_values, timesStr, label="Strassen", linewidth=2)
PyPlot.plot(n_values, timesV, label="Vinograd", linewidth=2)
PyPlot.grid(true)
PyPlot.legend()
PyPlot.tight_layout()
PyPlot.savefig("strassen_vinograd_ord.png", dpi=150)
PyPlot.show()
```
