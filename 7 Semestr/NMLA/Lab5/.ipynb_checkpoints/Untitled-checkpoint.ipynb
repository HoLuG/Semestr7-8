```julia
# 15 октября 18:52 - на паре 3 баллаs
using LinearAlgebra
using Printf
using Random

function generate_diag_dominant_matrix(n; rng=Random.default_rng())
    A = rand(rng, n, n) .* 10 .- 5
    for i in 1:n
        A[i,i] = sum(abs, A[i,j] for j in 1:n if j != i) + rand(rng)*5 + 1
    end
    return A
end

generate_vector(n; rng=Random.default_rng()) = rand(rng, n) .* 10 .- 5

norm_inf(v::AbstractVector) = maximum(abs.(v))

function is_strictly_diag_dom(A::AbstractMatrix)
    n, m = size(A)
    for i in 1:n
        if abs(A[i,i]) <= sum(abs, A[i,j] for j in 1:n if j != i)
            return false
        end
    end
    return true
end

function split_ADU(A::AbstractMatrix)
    n, m = size(A)
    D  = zeros(eltype(A), n, n)
    Am = zeros(eltype(A), n, n)
    Ap = zeros(eltype(A), n, n)
    for i in 1:n, j in 1:n
        if i == j
            D[i,i] = A[i,i]
        elseif i > j
            Am[i,j] = A[i,j]
        else
            Ap[i,j] = A[i,j]
        end
    end
    return Am, D, Ap
end

#  P = -(D^{-1})(A_- + A_+),  g = D^{-1} f.
function jacobi(A::AbstractMatrix, f::AbstractVector;
                x0=nothing,
                tol=1e-8, maxiter=1000, verbose=false)
    n, m = size(A)
    Am, D, Ap = split_ADU(A)

    P = -(Am + Ap)
    for i in 1:n
        P[i, :] ./= D[i,i]
    end
    g = f ./ diag(D)

    x_prev = copy(x0)
    history = Float64[]

    for k in 1:maxiter
        x = P * x_prev + g
        δ = norm_inf(x - x_prev)
        push!(history, δ)

        if verbose
            r = A * x - f
            @printf("Jacobi %5d:  ||Δx||∞=%.3e   ||Ax-f||₂=%.3e\n", k, δ, norm(r))
        end

        if δ ≤ tol
            return x, k, history
        end

        x_prev = x
    end
    return x_prev, maxiter, history
end

# P = -(D + A_-)^{-1} A_+,  g = (D + A_-)^{-1} f.
function seidel(A::AbstractMatrix, f::AbstractVector;
                               x0=nothing,
                               tol=1e-8, maxiter=1000, verbose=false)

    n, m = size(A)
    Am, D, Ap = split_ADU(A)
    M = D + Am
    invM = inv(M)

    P = -invM * Ap
    g =  invM * f

    x_prev = copy(x0)
    history = Float64[]
    for k in 1:maxiter
        x = P*x_prev + g
        δ = norm_inf(x - x_prev)
        push!(history, δ)
        if verbose
            r = A*x - f
            @printf("Seidel %3d:  ||Δx||∞=%g   ||Ax-f||₂=%g\n", k, δ, norm(r))
        end
        if δ ≤ tol
            return x, k, history, P, g
        end
        x_prev = x
    end
    return x_prev, maxiter, history, P, g
end
```

```julia
n   = 4
tol = 1e-6
x0  = zeros(n)
rng = MersenneTwister(42)

A = generate_diag_dominant_matrix(n; rng=rng)
f = generate_vector(n; rng=rng)

println("Матрица A =")
println(A)
println("\nВектор f =")
println(f)

println("\nПроверка диагонального преобладания: ", is_strictly_diag_dom(A))

tJ = @elapsed begin
    xJ, kJ, histJ = jacobi(A, f; x0=x0, tol=tol, verbose=true)
end
δJ = histJ[end]
println("\nМетод Якоби: k=$kJ,  x = ", xJ)
println("Сходимость на шаге $kJ, δ = $δJ,  δ < $tol")
println("Время работы: ", tJ * 1000, " мс")

tS = @elapsed begin
    xS, kS, histS = seidel(A, f; x0=x0, tol=tol, verbose=true)
end
δS = histS[end]
println("\nМетод Зейделя: k=$kS,  x = ", xS)
println("Сходимость на шаге $kS, δ = $δS,  δ < $tol")
println("Время работы: ", tS * 1000, " мс")

x_exact = A \ f
println("\nТочное решение: ", x_exact)

println("\n------------")
println("По итерациям: метод Зейделя в ", kJ / kS, " раз быстрее")
println("По времени:   метод Зейделя в ", tJ / tS, " раз быстрее")
```
