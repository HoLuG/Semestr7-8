## GAN

GAN Generative adversarial network(Генеративно-состязательная сеть).

GAN появились в 2014 г. Автором данных нейронных сетей является Ian Goodfellow.

Эти нейронные сети пременяются как правило для генерации изображений, а также изменения стилей изображений.

Разновидности GAN:
- генерируют новые изображения
- переносят стиль изображения
- по текстовому описанию создают изображения
- по экскизу рисуют изображнеия
- улучшают качество изображения
- и др.



GAN состоит из двух нейронных сетей: генератора и дискрименатора. В задачу генератора входит сгенерировать изображение из подающегося ей на вектора латентного пространства. В задачу дискриминатора входит сравнить изображение сгенерированное с настоящим. Одна нейронная сеть учиться заставлять спутывать настоящее изображение и сгенерированное, а другая учиться находить настоящее изображение.





```python
import torch
import torch.nn as nn

"""
Network Architectures
The following are the discriminator and generator architectures
"""

class discriminator(nn.Module):
    def __init__(self):
        super(discriminator, self).__init__()
        self.fc1 = nn.Linear(784, 512)
        self.fc2 = nn.Linear(512, 1)
        self.activation = nn.LeakyReLU(0.1)

    def forward(self, x):
        x = x.view(-1, 784)
        x = self.activation(self.fc1(x))
        x = self.fc2(x)
        return nn.Sigmoid()(x)


class generator(nn.Module):
    def __init__(self, latent_dim=128):
        super(generator, self).__init__()
        self.latent_dim = latent_dim
        self.fc1 = nn.Linear(latent_dim, 1024)
        self.fc2 = nn.Linear(1024, 2048)
        self.fc3 = nn.Linear(2048, 784)
        self.activation = nn.ReLU()

    def forward(self, x):
        x = self.activation(self.fc1(x))
        x = self.activation(self.fc2(x))
        x = self.fc3(x)
        x = x.view(-1, 1, 28, 28)
        return nn.Tanh()(x)
```

```python
"""
Import necessary libraries to create a generative adversarial network
The code is mainly developed using the PyTorch library
"""
import time
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets
from torchvision.transforms import transforms

import numpy as np
import matplotlib.pyplot as plt


"""
Determine if any GPUs are available
"""
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')


"""
Hyperparameter settings
"""
epochs = 16
lr = 2e-4
batch_size = 64
loss = nn.BCELoss()

# Model
G = generator().to(device)
D = discriminator().to(device)

G_optimizer = optim.Adam(G.parameters(), lr=lr, betas=(0.5, 0.999))
D_optimizer = optim.Adam(D.parameters(), lr=lr, betas=(0.5, 0.999))


"""
Image transformation and dataloader creation
Note that we are training generation and not classification, and hence
only the train_loader is loaded
"""
# Transform
transform = transforms.Compose([transforms.ToTensor(),
                                transforms.Normalize((0.5,), (0.5,))])
# Load data
train_set = datasets.MNIST('mnist/', train=True, download=True, transform=transform)
train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True)


"""
Network training procedure
Every step both the loss for disciminator and generator is updated
Discriminator aims to classify reals and fakes
Generator aims to generate images as realistic as possible
"""
for epoch in range(epochs):
    for idx, (imgs, _) in enumerate(train_loader):
        idx += 1

        # Training the discriminator
        # Real inputs are actual images of the MNIST dataset
        # Fake inputs are from the generator
        # Real inputs should be classified as 1 and fake as 0
        real_inputs = imgs.to(device)
        real_outputs = D(real_inputs)
        real_label = torch.ones(real_inputs.shape[0], 1).to(device)

        noise = (torch.rand(real_inputs.shape[0], 128) - 0.5) / 0.5
        noise = noise.to(device)
        fake_inputs = G(noise)
        fake_outputs = D(fake_inputs)
        fake_label = torch.zeros(fake_inputs.shape[0], 1).to(device)

        outputs = torch.cat((real_outputs, fake_outputs), 0)
        targets = torch.cat((real_label, fake_label), 0)

        D_loss = loss(outputs, targets)
        D_optimizer.zero_grad()
        D_loss.backward()
        D_optimizer.step()

        # Training the generator
        # For generator, goal is to make the discriminator believe everything is 1
        noise = (torch.rand(real_inputs.shape[0], 128)-0.5)/0.5
        noise = noise.to(device)

        fake_inputs = G(noise)
        fake_outputs = D(fake_inputs)
        fake_targets = torch.ones([fake_inputs.shape[0], 1]).to(device)
        G_loss = loss(fake_outputs, fake_targets)
        G_optimizer.zero_grad()
        G_loss.backward()
        G_optimizer.step()

        if idx % 100 == 0 or idx == len(train_loader):
            print('Epoch {} Iteration {}: discriminator_loss {:.3f} generator_loss {:.3f}'.format(epoch, idx, D_loss.item(), G_loss.item()))

    if (epoch+1) % 10 == 0:
        torch.save(G, 'Generator_epoch_{}.pth'.format(epoch))
        print('Model saved.')
    if (epoch) % 5 == 0:
        fig, ax = plt.subplots(1,5, figsize=(16, 3))
        for i in range(5):
          im = fake_inputs[i, 0, :,:].cpu().detach().numpy()
          ax[i].imshow(im)
        plt.show()

```

```python
# Задание 1.1 и 1.2: Обучение с латентным пространством 128 и визуализация потерь

"""
Import necessary libraries to create a generative adversarial network
The code is mainly developed using the PyTorch library
"""
import time
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets
from torchvision.transforms import transforms

import numpy as np
import matplotlib.pyplot as plt


"""
Determine if any GPUs are available
"""
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Используется устройство: {device}")


"""
Hyperparameter settings
"""
epochs = 16
lr = 2e-4
batch_size = 64
latent_dim = 128  # Исходный размер латентного пространства
loss = nn.BCELoss()

# Model
G = generator(latent_dim=latent_dim).to(device)
D = discriminator().to(device)

G_optimizer = optim.Adam(G.parameters(), lr=lr, betas=(0.5, 0.999))
D_optimizer = optim.Adam(D.parameters(), lr=lr, betas=(0.5, 0.999))


"""
Image transformation and dataloader creation
"""
# Transform
transform = transforms.Compose([transforms.ToTensor(),
                                transforms.Normalize((0.5,), (0.5,))])
# Load data
train_set = datasets.MNIST('mnist/', train=True, download=True, transform=transform)
train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True)


"""
Network training procedure with loss tracking
"""
# Списки для хранения потерь по эпохам
G_losses_epoch = []
D_losses_epoch = []

for epoch in range(epochs):
    epoch_G_losses = []
    epoch_D_losses = []
    
    for idx, (imgs, _) in enumerate(train_loader):
        idx += 1

        # Training the discriminator
        real_inputs = imgs.to(device)
        real_outputs = D(real_inputs)
        real_label = torch.ones(real_inputs.shape[0], 1).to(device)

        noise = (torch.rand(real_inputs.shape[0], latent_dim) - 0.5) / 0.5
        noise = noise.to(device)
        fake_inputs = G(noise)
        fake_outputs = D(fake_inputs)
        fake_label = torch.zeros(fake_inputs.shape[0], 1).to(device)

        outputs = torch.cat((real_outputs, fake_outputs), 0)
        targets = torch.cat((real_label, fake_label), 0)

        D_loss = loss(outputs, targets)
        D_optimizer.zero_grad()
        D_loss.backward()
        D_optimizer.step()
        epoch_D_losses.append(D_loss.item())

        # Training the generator
        noise = (torch.rand(real_inputs.shape[0], latent_dim)-0.5)/0.5
        noise = noise.to(device)

        fake_inputs = G(noise)
        fake_outputs = D(fake_inputs)
        fake_targets = torch.ones([fake_inputs.shape[0], 1]).to(device)
        G_loss = loss(fake_outputs, fake_targets)
        G_optimizer.zero_grad()
        G_loss.backward()
        G_optimizer.step()
        epoch_G_losses.append(G_loss.item())

        if idx % 100 == 0 or idx == len(train_loader):
            print('Epoch {} Iteration {}: discriminator_loss {:.3f} generator_loss {:.3f}'.format(
                epoch, idx, D_loss.item(), G_loss.item()))

    # Сохраняем средние потери за эпоху
    G_losses_epoch.append(np.mean(epoch_G_losses))
    D_losses_epoch.append(np.mean(epoch_D_losses))

    if (epoch) % 5 == 0:
        fig, ax = plt.subplots(1, 5, figsize=(16, 3))
        for i in range(5):
            im = fake_inputs[i, 0, :, :].cpu().detach().numpy()
            ax[i].imshow(im, cmap='gray')
            ax[i].axis('off')
        plt.suptitle(f'Сгенерированные изображения (Epoch {epoch}, latent_dim={latent_dim})')
        plt.show()

# Визуализация потерь по эпохам (Задание 1.2)
plt.figure(figsize=(12, 5))
plt.plot(range(epochs), G_losses_epoch, label='Потери генератора', marker='o')
plt.plot(range(epochs), D_losses_epoch, label='Потери дискриминатора', marker='s')
plt.xlabel('Эпоха')
plt.ylabel('Потери')
plt.title(f'Визуализация потерь по эпохам (latent_dim={latent_dim})')
plt.legend()
plt.grid(True)
plt.show()

# Сохраняем результаты для сравнения
results_128 = {
    'latent_dim': 128,
    'G_losses': G_losses_epoch,
    'D_losses': D_losses_epoch,
    'final_G_loss': G_losses_epoch[-1],
    'final_D_loss': D_losses_epoch[-1]
}

print(f"\nРезультаты обучения с latent_dim={latent_dim}:")
print(f"Финальная потеря генератора: {results_128['final_G_loss']:.4f}")
print(f"Финальная потеря дискриминатора: {results_128['final_D_loss']:.4f}")
```

```python
# Задание 1.4: Сравнение результатов и выводы

# Сравнительная визуализация потерь генератора
plt.figure(figsize=(14, 10))

plt.subplot(2, 2, 1)
plt.plot(range(epochs), results_128['G_losses'], label='latent_dim=128', marker='o', linewidth=2)
plt.plot(range(epochs), results_64['G_losses'], label='latent_dim=64', marker='s', linewidth=2)
plt.plot(range(epochs), results_16['G_losses'], label='latent_dim=16', marker='^', linewidth=2)
plt.xlabel('Эпоха')
plt.ylabel('Потери генератора')
plt.title('Сравнение потерь генератора')
plt.legend()
plt.grid(True)

plt.subplot(2, 2, 2)
plt.plot(range(epochs), results_128['D_losses'], label='latent_dim=128', marker='o', linewidth=2)
plt.plot(range(epochs), results_64['D_losses'], label='latent_dim=64', marker='s', linewidth=2)
plt.plot(range(epochs), results_16['D_losses'], label='latent_dim=16', marker='^', linewidth=2)
plt.xlabel('Эпоха')
plt.ylabel('Потери дискриминатора')
plt.title('Сравнение потерь дискриминатора')
plt.legend()
plt.grid(True)

plt.subplot(2, 2, 3)
latent_dims = [128, 64, 16]
final_G_losses = [results_128['final_G_loss'], results_64['final_G_loss'], results_16['final_G_loss']]
final_D_losses = [results_128['final_D_loss'], results_64['final_D_loss'], results_16['final_D_loss']]
x_pos = np.arange(len(latent_dims))
width = 0.35
plt.bar(x_pos - width/2, final_G_losses, width, label='Генератор', alpha=0.8)
plt.bar(x_pos + width/2, final_D_losses, width, label='Дискриминатор', alpha=0.8)
plt.xlabel('Размер латентного пространства')
plt.ylabel('Финальные потери')
plt.title('Финальные потери по размерам latent_dim')
plt.xticks(x_pos, latent_dims)
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(2, 2, 4)
# Разница между потерями (баланс между G и D)
balance_128 = abs(results_128['final_G_loss'] - results_128['final_D_loss'])
balance_64 = abs(results_64['final_G_loss'] - results_64['final_D_loss'])
balance_16 = abs(results_16['final_G_loss'] - results_16['final_D_loss'])
balances = [balance_128, balance_64, balance_16]
plt.bar(latent_dims, balances, alpha=0.7, color=['red', 'blue', 'green'])
plt.xlabel('Размер латентного пространства')
plt.ylabel('|G_loss - D_loss|')
plt.title('Баланс между генератором и дискриминатором')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Таблица результатов
print("\n" + "="*70)
print("СВОДНАЯ ТАБЛИЦА РЕЗУЛЬТАТОВ")
print("="*70)
print(f"{'Latent Dim':<15} {'G Loss (final)':<20} {'D Loss (final)':<20} {'|G-D|':<15}")
print("-"*70)
for res in [results_128, results_64, results_16]:
    balance = abs(res['final_G_loss'] - res['final_D_loss'])
    print(f"{res['latent_dim']:<15} {res['final_G_loss']:<20.4f} {res['final_D_loss']:<20.4f} {balance:<15.4f}")
print("="*70)

print("\n" + "="*70)
print("ВЫВОДЫ (Задание 1.4)")
print("="*70)
print("""
1. ВЛИЯНИЕ РАЗМЕРА ЛАТЕНТНОГО ПРОСТРАНСТВА:

   - latent_dim=128: Наибольшее латентное пространство позволяет генератору 
     иметь больше свободы для кодирования информации. Обычно приводит к 
     лучшему качеству генерации, но может требовать больше эпох для обучения.
     
   - latent_dim=64: Средний размер латентного пространства. Баланс между 
     качеством генерации и скоростью обучения.
     
   - latent_dim=16: Наименьшее латентное пространство. Меньше параметров, 
     быстрее обучение, но может ограничивать способность генератора создавать 
     разнообразные и качественные изображения.

2. АНАЛИЗ ПОТЕРЬ:

   - При уменьшении размера латентного пространства генератор может иметь 
     больше трудностей в обучении (выше потери генератора).
   
   - Дискриминатор обычно выигрывает при меньших latent_dim, так как 
     ему легче различать сгенерированные изображения.
   
   - Важен баланс между потерями генератора и дискриминатора. Слишком большой 
     разрыв указывает на дисбаланс в обучении.

3. РЕКОМЕНДАЦИИ:

   - Для простых данных (например, MNIST) даже latent_dim=16 может работать 
     приемлемо, но latent_dim=64-128 дает лучшие результаты.
   
   - Для более сложных данных рекомендуется использовать большее латентное 
     пространство (128, 256 или больше).
   
   - Следует мониторить баланс между потерями генератора и дискриминатора 
     для стабильного обучения GAN.
""")
print("="*70)
```
