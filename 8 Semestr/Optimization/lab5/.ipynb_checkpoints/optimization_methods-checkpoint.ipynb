```julia
function gradient_descent(f, grad_f, x0, alpha, max_iter=1000, tol=1e-6; run_label="", verbose=true)
    x = copy(x0)
    history = [(x=copy(x), f=f(x), iter=0)]
    converged = false
    
    for i in 1:max_iter
        grad = grad_f(x)
        
        if norm(grad) < tol
            converged = true
            if verbose
                prefix = isempty(run_label) ? "" : "[" * run_label * "] "
                println(prefix * "Сходимость достигнута на итерации " * string(i))
            end
            break
        end
        
        x = x - alpha * grad

        push!(history, (x=copy(x), f=f(x), iter=i))
    end
    
    if verbose && !converged
        prefix = isempty(run_label) ? "" : "[" * run_label * "] "
        println(prefix * "Сходимость не достигнута за " * string(max_iter) * " итераций; ||grad|| = " * string(norm(grad_f(x))))
    end

    return x, history
end
```

```julia
function numerical_gradient(f, x; h=1e-6)
    g = zeros(length(x))
    for i in eachindex(x)
        x_plus = copy(x)
        x_minus = copy(x)
        x_plus[i] += h
        x_minus[i] -= h
        g[i] = (f(x_plus) - f(x_minus)) / (2.0 * h)
    end
    return g
end

function line_search(f, grad_f, x, direction;
                     alpha_init=1.0,
                     alpha_max=100.0,
                     tol_ls=1e-6,
                     max_iter_ls=80)
    if norm(direction) < 1e-14
        return 0.0
    end

    g(alpha) = f(x + alpha * direction)

    left = 0.0
    right = alpha_init
    g_left = g(left)
    g_right = g(right)

    while g_right < g_left && right < alpha_max
        left = right
        g_left = g_right
        right = min(2.0 * right, alpha_max)
        g_right = g(right)
    end

    if right == alpha_max && g_right < g_left
        return right
    end

    a = 0.0
    b = right
    tau = (sqrt(5.0) - 1.0) / 2.0

    c1 = b - tau * (b - a)
    c2 = a + tau * (b - a)
    g1 = g(c1)
    g2 = g(c2)

    for _ in 1:max_iter_ls
        if abs(b - a) < tol_ls
            break
        end

        if g1 > g2
            a = c1
            c1 = c2
            g1 = g2
            c2 = a + tau * (b - a)
            g2 = g(c2)
        else
            b = c2
            c2 = c1
            g2 = g1
            c1 = b - tau * (b - a)
            g1 = g(c1)
        end
    end

    return (a + b) / 2.0
end

function steepest_descent(f, grad_f, x0, max_iter=1000, tol=1e-6;
                          run_label="",
                          verbose=true,
                          ortho_tol=1e-3)
    x = copy(x0)
    g0 = grad_f(x)
    history = [(x=copy(x), f=f(x), iter=0, alpha=0.0, grad_norm=norm(g0), ortho_cos=NaN)]
    converged = false

    max_abs_cos = 0.0
    ortho_ok = true

    for i in 1:max_iter
        grad = grad_f(x)

        if norm(grad) < tol
            converged = true
            if verbose
                prefix = isempty(run_label) ? "" : "[" * run_label * "] "
                println(prefix * "Сходимость достигнута на итерации " * string(i))
            end
            break
        end

        direction = -grad
        alpha = line_search(f, grad_f, x, direction)

        x_new = x + alpha * direction
        grad_new = grad_f(x_new)

        denom = max(norm(grad) * norm(grad_new), 1e-14)
        ortho_cos = abs(dot(grad, grad_new)) / denom
        max_abs_cos = max(max_abs_cos, ortho_cos)
        if ortho_cos > ortho_tol
            ortho_ok = false
        end

        x = x_new

        push!(history, (
            x=copy(x),
            f=f(x),
            iter=i,
            alpha=alpha,
            grad_norm=norm(grad_new),
            ortho_cos=ortho_cos
        ))
    end

    if verbose
        prefix = isempty(run_label) ? "" : "[" * run_label * "] "
        if length(history) > 1
            status = ortho_ok ? "пройдена" : "не пройдена"
            println(prefix * "Проверка ортогональности соседних градиентов: " * status * "; max |cos| = " * string(max_abs_cos))
        else
            println(prefix * "Проверка ортогональности соседних градиентов: недостаточно итераций")
        end
    end

    if verbose && !converged
        prefix = isempty(run_label) ? "" : "[" * run_label * "] "
        println(prefix * "Сходимость не достигнута за " * string(max_iter) * " итераций; ||grad|| = " * string(norm(grad_f(x))))
    end

    return x, history
end
```

```julia
function cg_beta(grad_new, grad, direction_prev; rule=:polak_ribiere, eps=1e-14)
    y = grad_new - grad

    if rule == :hestenes_stiefel
        denom_raw = dot(direction_prev, y)
        denom = abs(denom_raw) < eps ? sign(denom_raw + eps) * eps : denom_raw
        return dot(grad_new, y) / denom
    elseif rule == :dixon
        denom_raw = dot(direction_prev, grad)
        denom = abs(denom_raw) < eps ? sign(denom_raw + eps) * eps : denom_raw
        return -dot(grad_new, grad_new) / denom
    elseif rule == :dai_yuan
        denom_raw = dot(direction_prev, y)
        denom = abs(denom_raw) < eps ? sign(denom_raw + eps) * eps : denom_raw
        return -dot(grad_new, y) / denom
    elseif rule == :polak_ribiere
        denom = max(dot(grad, grad), eps)
        return dot(grad_new, y) / denom
    else
        error("Неизвестная формула beta: $(rule)")
    end
end

function conjugate_gradient(f, grad_f, x0, max_iter=1000, tol=1e-6;
                            beta_rule=:polak_ribiere,
                            run_label="",
                            verbose=true)
    x = copy(x0)
    grad = grad_f(x)
    direction = -grad
    history = [(x=copy(x), f=f(x), iter=0, alpha=0.0, beta=0.0)]
    converged = false

    for i in 1:max_iter
        if norm(grad) < tol
            converged = true
            if verbose
                prefix = isempty(run_label) ? "" : "[" * run_label * "] "
                println(prefix * "Сходимость достигнута на итерации " * string(i))
            end
            break
        end

        alpha = line_search(f, grad_f, x, direction)
        x_new = x + alpha * direction
        grad_new = grad_f(x_new)

        beta = cg_beta(grad_new, grad, direction; rule=beta_rule)
        direction_new = -grad_new + beta * direction

        x = x_new
        grad = grad_new
        direction = direction_new

        push!(history, (x=copy(x), f=f(x), iter=i, alpha=alpha, beta=beta))
    end

    if verbose && !converged
        prefix = isempty(run_label) ? "" : "[" * run_label * "] "
        println(prefix * "Сходимость не достигнута за " * string(max_iter) * " итераций; ||grad|| = " * string(norm(grad)))
    end

    return x, history
end

function cg_quadratic_n_step_check(a, b, x0; tol=1e-8, verbose=true)
    A = Diagonal([2.0 * a, 2.0 * b])
    n = length(x0)

    x = copy(x0)
    g = A * x
    r = -g
    p = copy(r)

    residuals = [norm(g)]

    for _ in 1:n
        Ap = A * p
        denom = dot(p, Ap)
        if abs(denom) < 1e-14
            break
        end

        alpha = dot(r, r) / denom
        x = x + alpha * p

        r_new = r - alpha * Ap
        g_new = -r_new
        push!(residuals, norm(g_new))

        if norm(g_new) < tol
            r = r_new
            break
        end

        beta = dot(r_new, r_new) / max(dot(r, r), 1e-14)
        p = r_new + beta * p
        r = r_new
    end

    grad_norm_final = norm(A * x)
    passed = grad_norm_final < tol

    if verbose
        msg = passed ? "подтверждена" : "не подтверждена"
        println("[Quadratic CG n-step check | a=$(a), b=$(b)] Сходимость за n=$(n) шагов: " * msg * "; ||grad||_n = " * string(grad_norm_final))
    end

    return (passed=passed, n=n, grad_norm_final=grad_norm_final, residuals=residuals)
end
```

```julia
function history_xy(history)
    xs = [h.x[1] for h in history]
    ys = [h.x[2] for h in history]
    return xs, ys
end

function plot_trajectories_on_contour(f, histories, title_text;
                                      xlims=(-5.0, 5.0),
                                      ylims=(-5.0, 5.0),
                                      levels=25,
                                      grid_n=180,
                                      auto_zoom=false,
                                      zoom_pad=0.2,
                                      min_span=5.0,
                                      zoom_max_points=typemax(Int),
                                      aspect_ratio_setting=:auto)
    if auto_zoom
        all_x = Float64[]
        all_y = Float64[]
        for (_, hist) in histories
            n = min(length(hist), zoom_max_points)
            hist_part = hist[1:n]
            append!(all_x, [h.x[1] for h in hist_part])
            append!(all_y, [h.x[2] for h in hist_part])
        end
        xmin, xmax = minimum(all_x), maximum(all_x)
        ymin, ymax = minimum(all_y), maximum(all_y)
        xspan = max(xmax - xmin, min_span)
        yspan = max(ymax - ymin, min_span)
        xpad = zoom_pad * xspan
        ypad = zoom_pad * yspan
        xlims = (xmin - xpad, xmax + xpad)
        ylims = (ymin - ypad, ymax + ypad)
    end

    xr = LinRange(xlims[1], xlims[2], grid_n)
    yr = LinRange(ylims[1], ylims[2], grid_n)
    Z = [f([x, y]) for y in yr, x in xr]

    p = contour(xr, yr, Z;
                levels=levels,
                color=:grays,
                colorbar=false,
                xlabel="x1",
                ylabel="x2",
                title=title_text,
                aspect_ratio=aspect_ratio_setting,
                xlims=xlims,
                ylims=ylims,
                legend=:outerbottom,
                legend_column=2)

    colors = [:blue, :red, :yellow, :green, :purple, :brown, :magenta, :teal]
    markers = [:circle, :square, :diamond, :utriangle, :dtriangle, :star5, :xcross, :pentagon]

    i = 1
    for (name, hist) in histories
        xs, ys = history_xy(hist)
        color_i = colors[1 + mod(i - 1, length(colors))]
        marker_i = markers[1 + mod(i - 1, length(markers))]

        # Линия траектории
        plot!(p, xs, ys;
              label=name,
              linewidth=2,
              color=color_i,
              linealpha=0.9)

        # Все точки итераций с одинаковым размером маркера
        scatter!(p, xs, ys;
                 label="",
                 marker=marker_i,
                 markersize=4,
                 markerstrokewidth=0.8,
                 markerstrokecolor=:black,
                 color=color_i)

        i += 1
    end

    x0 = histories[1][2][1].x
    scatter!(p, [x0[1]], [x0[2]];
             label="Начальная точка",
             color=:black,
             marker=:star5,
             markersize=8,
             markerstrokewidth=1.5,
             markerstrokecolor=:white)

    return p
end
```

```julia
alpha_q = 1.0
betas = [1.0, 5.0, 10.0]
x0_quad = [10.0, 10.0]

max_iter_quad = 3000
tol_quad = 1e-4

cg_methods = [
    ("Хестенса-Стифеля", :hestenes_stiefel),
    ("Диксона", :dixon),
    ("Даяна", :dai_yuan),
    ("Полака-Рибьера", :polak_ribiere)
]

plots_quad = Any[]

for beta_q in betas
    f_quad(x) = alpha_q * x[1]^2 + beta_q * x[2]^2
    grad_quad(x) = [2.0 * alpha_q * x[1], 2.0 * beta_q * x[2]]

    histories = Any[]

    for (method_name, rule) in cg_methods
        _, hist_cg = conjugate_gradient(
            f_quad, grad_quad, x0_quad, max_iter_quad, tol_quad;
            beta_rule=rule,
            run_label="Quadratic beta=$(Int(beta_q)) | $(method_name)"
        )
        push!(histories, (method_name, hist_cg))
    end

    # Это проверка классического линейного CG для квадратичной формы (теорема про n шагов).
    cg_quadratic_n_step_check(alpha_q, beta_q, x0_quad; tol=1e-8, verbose=true)

    p = plot_trajectories_on_contour(
        f_quad,
        histories,
        "Квадратичная: a=1, b=$(Int(beta_q)), x0=$(x0_quad)";
        xlims=(-110.0, 110.0),
        ylims=(-110.0, 110.0),
        levels=22,
        grid_n=170,
        auto_zoom=true,
        zoom_pad=0.25,
        min_span=6.0,
        aspect_ratio_setting=:equal
    )

    push!(plots_quad, p)
end

plot(plots_quad...; layout=(1, 3), size=(2100, 620))
```

```julia
rosenbrock(x) = (1 - x[1])^2 + 100 * (x[2] - x[1]^2)^2

function grad_rosenbrock(x)
    g1 = -2 * (1 - x[1]) - 400 * x[1] * (x[2] - x[1]^2)
    g2 = 200 * (x[2] - x[1]^2)
    return [g1, g2]
end

function rastrigin(x)
    A = 10.0
    return 2A + (x[1]^2 - A * cos(2 * pi * x[1])) + (x[2]^2 - A * cos(2 * pi * x[2]))
end

function grad_rastrigin(x)
    A = 10.0
    g1 = 2 * x[1] + 2 * pi * A * sin(2 * pi * x[1])
    g2 = 2 * x[2] + 2 * pi * A * sin(2 * pi * x[2])
    return [g1, g2]
end

function schwefel(x)
    c = 418.9829
    return 2c - (x[1] * sin(sqrt(abs(x[1]))) + x[2] * sin(sqrt(abs(x[2]))))
end

function grad_schwefel(x)
    g = zeros(2)
    for i in 1:2
        xi = x[i]
        s = sqrt(abs(xi) + 1e-12)
        d_term = sin(s) + xi * cos(s) * sign(xi) / (2s)
        g[i] = -d_term
    end
    return g
end
```

```julia
max_iter_nl = 5000
tol_nl = 1e-5

experiments = [
    (
        "Rosenbrock",
        rosenbrock,
        grad_rosenbrock,
        [-2.0, 2.0],
        (-5.0, 5.0),
        (-1.0, 20.0),
        false
    ),
    (
        "Rastrigin",
        rastrigin,
        grad_rastrigin,
        [2.0, 2.0],
        (-5.12, 5.12),
        (-5.12, 5.12),
        false
    ),
    (
        "Schwefel",
        schwefel,
        grad_schwefel,
        [120.0, -90.0],
        (-140.0, 140.0),
        (-140.0, 140.0),
        true
    )
]

cg_methods = [
    ("Хестенса-Стифеля", :hestenes_stiefel),
    ("Диксона", :dixon),
    ("Даяна", :dai_yuan),
    ("Полака-Рибьера", :polak_ribiere)
]

plots_nl = Any[]

for (name, f_cur, g_cur, x0_cur, xlim_cur, ylim_cur, use_auto_zoom) in experiments
    histories = Any[]

    for (method_name, rule) in cg_methods
        _, hist_cg = conjugate_gradient(
            f_cur, g_cur, x0_cur, max_iter_nl, tol_nl;
            beta_rule=rule,
            run_label="$(name) | $(method_name)"
        )
        push!(histories, (method_name, hist_cg))
    end

    auto_zoom_cur = name == "Rosenbrock" ? false : use_auto_zoom

    p = plot_trajectories_on_contour(
        f_cur,
        histories,
        "$(name), x0 = $(x0_cur)";
        xlims=xlim_cur,
        ylims=ylim_cur,
        levels=25,
        grid_n=180,
        auto_zoom=auto_zoom_cur,
        zoom_pad=0.35,
        min_span=20.0
    )

    push!(plots_nl, p)
end

plot(plots_nl...; layout=(1, 3), size=(2100, 800))
```

```julia
println("\n============================================================")
println("КВАДРАТИЧНАЯ ФУНКЦИЯ: f(x) = a*x1^2 + b*x2^2")
println("Точный шаг: alpha = -g'd / d'Hd  =>  проверка сходимости CG за n=2 шага")
println("============================================================")

function run_cg_quadratic_exact(a, b, x0; beta_rule=:polak_ribiere, tol=1e-12, max_iter=50)
    H = Diagonal([2.0 * a, 2.0 * b])
    f(x) = a * x[1]^2 + b * x[2]^2
    grad(x) = H * x

    x = copy(x0)
    g = grad(x)
    d = -g

    steps = 0
    for _ in 1:max_iter
        if norm(g) < tol
            break
        end

        Hd = H * d
        denom = dot(d, Hd)
        if abs(denom) < 1e-14
            break
        end

        alpha = -dot(g, d) / denom
        x = x + alpha * d

        g_new = grad(x)
        y = g_new - g

        if norm(g_new) < tol
            g = g_new
            steps += 1
            break
        end

        beta = if beta_rule == :hestenes_stiefel
            dot(g_new, y) / max(dot(d, y), 1e-14)
        elseif beta_rule == :dixon
            -dot(g_new, g_new) / max(dot(d, g), 1e-14)
        elseif beta_rule == :dai_yuan
            -dot(g_new, y) / max(dot(d, y), 1e-14)
        elseif beta_rule == :polak_ribiere
            dot(g_new, y) / max(dot(g, g), 1e-14)
        else
            error("unknown beta rule")
        end

        d = -g_new + beta * d
        g = g_new
        steps += 1
    end

    return (x=x, f=f(x), steps=steps, grad_norm=norm(g))
end

method_specs = [
    ("Хестенса-Стифеля", :hestenes_stiefel),
    ("Диксона", :dixon),
    ("Даяна", :dai_yuan),
    ("Полака-Рибьера", :polak_ribiere)
]

for b in (1.0, 5.0, 10.0)
    println("\n--- a=1, b=$(Int(b)) ---")
    for (name, rule) in method_specs
        res = run_cg_quadratic_exact(1.0, b, [10.0, 10.0]; beta_rule=rule)
        badge = res.steps <= 2 ? "*** за n=2 шага ***" : ""
        println(rpad(name, 20), " шагов=", lpad(string(res.steps), 3),
                "   x=", round.(res.x; digits=6),
                "   f=", round(res.f; digits=12), "   ", badge)
    end
end
```
