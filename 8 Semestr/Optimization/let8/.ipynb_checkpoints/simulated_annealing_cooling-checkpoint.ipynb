## Имитация отжига: 3 метода охлаждения (Julia)

Требования:
- 2 функции: **простая** (для проверки) и **мультимодальная** (≥3 локальных минимума).
- 3 режима охлаждения: **геометрический** (синий), **логарифмический** (красный), **сверхбыстрый** (жёлтый).
- Показать **движение точек** от начальной точки к минимуму (траектории на графике).
- Вывести на русском **время до сходимости** для каждого метода.


```julia
import Pkg

# Если пакеты уже стоят, Pkg.add можно закомментировать.
for p in ["Plots", "Random", "Printf", "Statistics"]
    try
        @eval import $(Symbol(p))
    catch
        Pkg.add(p)
    end
end

using Random
using Printf
using Statistics
using Plots
```

## Функции

- **Простая**: выпуклая квадратичная чаша (один минимум).
- **Мультимодальная**: функция **Himmelblau** — имеет 4 минимума (точно ≥3 локальных минимума), удобно визуализировать в 2D.


```julia
# Простая функция (для проверки корректности алгоритма)
f_simple(x) = (x[1] - 1.0)^2 + 2.0*(x[2] + 2.0)^2

# Himmelblau (мультимодальная, 4 локальных минимума)
function f_himmelblau(x)
    X, Y = x[1], x[2]
    return (X^2 + Y - 11)^2 + (X + Y^2 - 7)^2
end
```

## Имитация отжига (Simulated Annealing)

Используем:
- случайное предложение шага: \(x' = x + \sigma(T)\,\xi\), где \(\xi\sim\mathcal{N}(0, I)\)
- правило принятия Метрополиса: \(p=\min(1, \exp(-(f(x')-f(x))/T))\)
- остановка по "сходимости": нет улучшений лучшего значения `stall_iters` итераций **или** температура упала ниже `T_min`.

### Охлаждение
- **Геометрическое**: \(T_k = T_0\,\alpha^k\)
- **Логарифмическое**: \(T_k = \dfrac{T_0}{\log(k + c)}\)
- **Сверхбыстрое** (вариант "very fast" в духе Ingber): \(T_k = T_0\,\exp(-c\,k^{1/d})\), где \(d\) — размерность (у нас 2)


```julia
abstract type CoolingSchedule end

struct GeometricCooling <: CoolingSchedule
    T0::Float64
    alpha::Float64
end

struct LogarithmicCooling <: CoolingSchedule
    T0::Float64
    c::Float64
end

struct SuperFastCooling <: CoolingSchedule
    T0::Float64
    c::Float64
    d::Int
end

temperature(s::GeometricCooling, k::Int) = s.T0 * (s.alpha^k)
temperature(s::LogarithmicCooling, k::Int) = s.T0 / log(k + s.c)
temperature(s::SuperFastCooling, k::Int) = s.T0 * exp(-s.c * (k^(1.0 / s.d)))

function simulated_annealing(f, x0::AbstractVector{<:Real};
    schedule::CoolingSchedule,
    max_iter::Int = 50_000,
    T_min::Float64 = 1e-8,
    stall_iters::Int = 3_000,
    step0::Float64 = 0.5,
    bounds = ((-6.0, 6.0), (-6.0, 6.0)),
    rng::AbstractRNG = Random.default_rng()
)
    x = Float64.(x0)
    fx = f(x)

    best_x = copy(x)
    best_f = fx

    traj = Vector{Vector{Float64}}()
    push!(traj, copy(x))

    last_improve = 0
    t_start = time_ns()

    for k in 1:max_iter
        T = temperature(schedule, k)
        if !(isfinite(T)) || T <= T_min
            break
        end

        # Амплитуда шага уменьшается вместе с температурой (но не слишком резко)
        σ = step0 * sqrt(max(T / temperature(schedule, 1), 1e-12))

        x_new = x .+ σ .* randn(rng, length(x))
        # Ограничим область поиска (для устойчивой визуализации)
        for i in 1:length(x_new)
            lo, hi = bounds[i]
            x_new[i] = clamp(x_new[i], lo, hi)
        end

        f_new = f(x_new)
        Δ = f_new - fx

        accept = (Δ <= 0) || (rand(rng) < exp(-Δ / T))
        if accept
            x = x_new
            fx = f_new
            push!(traj, copy(x))
        end

        if fx < best_f - 1e-12
            best_f = fx
            best_x = copy(x)
            last_improve = k
        end

        if (k - last_improve) >= stall_iters
            break
        end
    end

    elapsed_s = (time_ns() - t_start) / 1e9
    return (best_x = best_x, best_f = best_f, traj = traj, elapsed_s = elapsed_s)
end
```

## Запуск трёх методов и вывод времени (на русском)

Цвета по требованию:
- синий — геометрический
- красный — логарифмический
- жёлтый — сверхбыстрый


```julia
Random.seed!(42)

# Выбираем функцию для демонстрации траекторий:
f = f_himmelblau

# Начальная точка (можно менять)
x0 = [-4.5, 4.0]

# Параметры охлаждения подобраны так, чтобы:
# - траектории были видимыми
# - сходимость наступала за разумное время
geom = GeometricCooling(2.0, 0.9993)
logc = LogarithmicCooling(2.0, 10.0)
super = SuperFastCooling(2.0, 0.06, 2)

common_kwargs = (
    max_iter = 120_000,
    T_min = 1e-8,
    stall_iters = 7_000,
    step0 = 0.9,
    bounds = ((-6.0, 6.0), (-6.0, 6.0))
)

res_geom = simulated_annealing(f, x0; schedule=geom, common_kwargs...)
res_log  = simulated_annealing(f, x0; schedule=logc, common_kwargs...)
res_sup  = simulated_annealing(f, x0; schedule=super, common_kwargs...)

@printf("Геометрическое охлаждение: время сходимости = %.4f сек, найдено f(x)=%.6f при x=(%.4f, %.4f)\n",
    res_geom.elapsed_s, res_geom.best_f, res_geom.best_x[1], res_geom.best_x[2])
@printf("Логарифмическое охлаждение: время сходимости = %.4f сек, найдено f(x)=%.6f при x=(%.4f, %.4f)\n",
    res_log.elapsed_s, res_log.best_f, res_log.best_x[1], res_log.best_x[2])
@printf("Сверхбыстрое охлаждение: время сходимости = %.4f сек, найдено f(x)=%.6f при x=(%.4f, %.4f)\n",
    res_sup.elapsed_s, res_sup.best_f, res_sup.best_x[1], res_sup.best_x[2])
```

## График: уровни функции + траектории

На одном графике показываем уровни функции и движение точек:
- **синий** — геометрический
- **красный** — логарифмический
- **жёлтый** — сверхбыстрый


```julia
function traj_xy(traj)
    xs = [p[1] for p in traj]
    ys = [p[2] for p in traj]
    return xs, ys
end

xgrid = range(-6, 6, length=250)
ygrid = range(-6, 6, length=250)
Z = [f([x, y]) for y in ygrid, x in xgrid]

p = contour(
    xgrid, ygrid, Z;
    levels=35,
    c=:grays,
    linewidth=1,
    title="Имитация отжига: траектории для 3 охлаждений (Himmelblau)",
    xlabel="x",
    ylabel="y",
    legend=:topright
)

xg, yg = traj_xy(res_geom.traj)
xl, yl = traj_xy(res_log.traj)
xs, ys = traj_xy(res_sup.traj)

plot!(p, xg, yg; color=:blue,  lw=2, label="Геометрическое")
plot!(p, xl, yl; color=:red,   lw=2, label="Логарифмическое")
plot!(p, xs, ys; color=:yellow, lw=2, label="Сверхбыстрое")

scatter!(p, [x0[1]], [x0[2]]; color=:black, ms=6, label="Старт")
scatter!(p, [res_geom.best_x[1]], [res_geom.best_x[2]]; color=:blue, ms=6, marker=:star5, label="Мин (геом)")
scatter!(p, [res_log.best_x[1]],  [res_log.best_x[2]];  color=:red,  ms=6, marker=:star5, label="Мин (лог)")
scatter!(p, [res_sup.best_x[1]],  [res_sup.best_x[2]];  color=:yellow, ms=6, marker=:star5, label="Мин (сверх)")

p
```
