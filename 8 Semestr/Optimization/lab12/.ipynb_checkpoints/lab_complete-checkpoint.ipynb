```python
from PIL import Image, ImageDraw

polygons = [
    {'vertices': [(179, 203), (162, 80), (110, 108)],
     'RGBA': (0, 0, 255, 125)},
    {'vertices': [(239, 65), (216, 179), (11, 79)], 'RGBA': (0, 255, 0, 125)},
    {'vertices': [(209, 123), (23, 245), (194, 187)], 'RGBA': (255, 0, 0, 125)}
]


def draw(polygons, size):
    """ Функция для рисования многоугольников """

    img = Image.new('RGB', (size[0], size[1]), (255, 255, 255))
    drw = ImageDraw.Draw(img, 'RGBA')

    for pol in polygons:
        drw.polygon(pol['vertices'], pol['RGBA'])

    return (img.convert("RGB"))


draw(polygons, size=(300, 300))
```

```python
import numpy as np


def random_triangles(N, size, vertices=3, colour="random", alpha="random"):
    """ Функция создания популяции рандомных треугольников """

    collection = []

    for i in range(N):
        coords = [(np.random.randint(0, size[0]), np.random.randint(0, size[1]))
                  for i in range(vertices)]

        if alpha == "random":
            a = np.random.randint(0, 50)
        else:
            a = alpha

        if colour == "white":
            rgba = (255, 255, 255, a)
        elif colour == "black":
            rgba = (0, 0, 0, a)
        elif colour == "random":
            rgba = (np.random.randint(0, 256), np.random.randint(
                0, 256), np.random.randint(0, 256), a)

        triangle = {"vertices": coords, "RGBA": rgba}
        collection.append(triangle)

    return (collection)

collection = random_triangles(N=50, size=(
    200, 200), vertices=5, colour="random", alpha="random")
img = draw(collection, size=(200, 200))
img
```

```python
def pixel_difference(candidate, target):
    """ Функция для попиксельного вычисления отличий между картинками"""
    diff = ImageChops.difference(candidate, target)
    totdiff = np.array(diff.getdata()).sum()
    return (totdiff)


diff = pixel_difference(img, target)
diff
```

```python
def max_pixel_difference(target):
    """ Функция вычисления максимально возможной разницы между изображениями"""
    white = Image.new('RGB', target.size, (255, 255, 255))
    diff = ImageChops.difference(white, target)
    maxdiff = np.array(diff.getdata()).sum()
    return (maxdiff)


maxdiff = max_pixel_difference(target)
fitness = (1 - diff / maxdiff) * 100
print("Fitness = " + str(np.round(fitness, 4)) + "%")
```

```python
import copy
import matplotlib.pyplot as plt

def mutation(original, size, mutation_type="parameter", num_verts=3, colour="random", alpha="random"):
    """
    Функция мутации с разными типами:
    - "parameter": мутация параметров многоугольников (координаты, цвет, прозрачность)
    - "delete": удаление случайного многоугольника
    - "insert": вставка нового случайного многоугольника
    - "swap": перемещение многоугольника в другое место
    - "combined": комбинация всех типов
    """
    mutant = copy.deepcopy(original)
    
    if len(mutant) == 0:
        return mutant
    
    mutation_type_actual = mutation_type
    if mutation_type == "combined":
        mutation_type_actual = np.random.choice(["parameter", "delete", "insert", "swap"])
    
    if mutation_type_actual == "parameter":
        polidx = np.random.randint(len(mutant))
        
        if np.random.random() < 0.5:
            num_verts_poly = len(mutant[polidx]['vertices'])
            vert_idx = np.random.randint(num_verts_poly)
            new_x = np.clip(mutant[polidx]['vertices'][vert_idx][0] + np.random.randint(-20, 21), 
                            -10, size[0] + 10)
            new_y = np.clip(mutant[polidx]['vertices'][vert_idx][1] + np.random.randint(-20, 21), 
                            -10, size[1] + 10)
            mutant[polidx]['vertices'][vert_idx] = (new_x, new_y)
        
        if np.random.random() < 0.5:
            r, g, b, a = mutant[polidx]['RGBA']
            r = np.clip(r + np.random.randint(-30, 31), 0, 255)
            g = np.clip(g + np.random.randint(-30, 31), 0, 255)
            b = np.clip(b + np.random.randint(-30, 31), 0, 255)
            mutant[polidx]['RGBA'] = (r, g, b, a)
        
        if np.random.random() < 0.3:
            r, g, b, a = mutant[polidx]['RGBA']
            a = np.clip(a + np.random.randint(-10, 11), 0, 255)
            mutant[polidx]['RGBA'] = (r, g, b, a)
    
    elif mutation_type_actual == "delete":
        if len(mutant) > 1:
            idx_to_delete = np.random.randint(len(mutant))
            mutant.pop(idx_to_delete)
    
    elif mutation_type_actual == "insert":
        new_polygon = random_polygons(1, size, num_verts=num_verts, 
                                      colour=colour, alpha=alpha)[0]
        insert_pos = np.random.randint(0, len(mutant) + 1)
        mutant.insert(insert_pos, new_polygon)
    
    elif mutation_type_actual == "swap":
        if len(mutant) > 1:
            idx_from = np.random.randint(len(mutant))
            idx_to = np.random.randint(len(mutant))
            if idx_from != idx_to:
                polygon = mutant.pop(idx_from)
                mutant.insert(idx_to, polygon)
    
    return mutant
```

```python
def pop_fitness(pop, target):
    """
    Вычисляет приспособленность каждой популяции
    """

    size = target.size
    maxdiff = max_pixel_difference(target)

    fitvec = []

    for org in pop:
        img = draw(org[0], size)
        diff = pixel_difference(img, target)
        fitness = (1 - diff / maxdiff) * 100
        fitvec.append(fitness)

    return (fitvec)


def individ_fitness(individ, target):
    """
    Вычисление приспособленности индивида
    """

    size = target.size
    maxdiff = max_pixel_difference(target)

    img = draw(individ, size)
    diff = pixel_difference(img, target)
    fitness = (1 - diff / maxdiff) * 100

    return (fitness)
```

```python
def random_polygons(N, size, num_verts=3, colour="random", alpha="random"):
    """ Создание рандомных многоугольников """

    collection = []

    for i in range(N):
        max_tri_size = min(size[0], size[1]) // 8
        cx = np.random.randint(0, size[0])
        cy = np.random.randint(0, size[1])
        coords = [(cx + np.random.randint(-max_tri_size, max_tri_size),
                   cy + np.random.randint(-max_tri_size, max_tri_size)) for i in range(num_verts)]

        if alpha == "random":
            a = np.random.randint(0, 50)
        else:
            a = alpha

        if colour == "white":
            rgba = (255, 255, 255, a)
        elif colour == "black":
            rgba = (0, 0, 0, a)
        elif colour == "random":
            rgba = (np.random.randint(0, 256), np.random.randint(
                0, 256), np.random.randint(0, 256), a)

        triangle = {"vertices": coords, "RGBA": rgba}
        collection.append(triangle)

    return (collection)


def create_pop(Npop, N, size, num_verts=3, colour_init="black", alpha_init=100):
    """
    Создание популяции рандомных особей
    Каждая особь имеет N рандомно созданных многоугольников
    """

    pop = []
    for i in range(Npop):
        individ = random_polygons(
            N, size, num_verts=num_verts, colour=colour_init, alpha=alpha_init)
        fitness = individ_fitness(individ, target)
        pop.append((individ, fitness))

    pop = sorted(pop, key=lambda x: (x[1]), reverse=True)
    return pop


pop = create_pop(Npop=20, N=100, size=(200, 200), num_verts=2,
                 colour_init="random", alpha_init="random")
```

```python
def select_parent_roulette(pop):
    """
    Выбор родителя методом рулетки (roulette wheel selection)
    Вероятность выбора особи пропорциональна её fitness
    """
    fitnesses = np.array([ind[1] for ind in pop])
    
    min_fitness = np.min(fitnesses)
    if min_fitness <= 0:
        fitnesses = fitnesses - min_fitness + 1
    
    probabilities = fitnesses / np.sum(fitnesses)
    
    selected_idx = np.random.choice(len(pop), p=probabilities)
    
    return pop[selected_idx]


def select_parent_tournament(pop, tournament_size=3):
    """
    Турнирный отбор родителя
    Выбирает случайных tournament_size особей и возвращает лучшую из них
    """
    tournament_indices = np.random.choice(len(pop), size=min(tournament_size, len(pop)), replace=False)
    tournament = [pop[i] for i in tournament_indices]
    
    best = max(tournament, key=lambda x: x[1])
    
    return best
```

```python
def crossover(parent1, parent2, size, pmut=0.1):
    """
    Одноточечный кроссинговер
    Скрещивает два родителя, создавая потомка из части многоугольников первого родителя
    и части многоугольников второго родителя
    """
    p1_polygons = parent1[0]
    p2_polygons = parent2[0]
    
    crossover_point = np.random.randint(1, min(len(p1_polygons), len(p2_polygons)))
    
    child_polygons = copy.deepcopy(p1_polygons[:crossover_point]) + copy.deepcopy(p2_polygons[crossover_point:])
    
    if pmut > 0 and np.random.random() < pmut:
        child_polygons = mutation(child_polygons, size, mutation_type="parameter")
    
    return child_polygons
```

```python
def selection_uniform_replacement(pop, offspring, num_elites=0):
    """
    Равномерная случайная замена
    Заменяет случайных особей в популяции на потомков
    """
    Npop = len(pop)
    new_pop = copy.deepcopy(pop)
    
    if num_elites > 0:
        elites = new_pop[:num_elites]
    else:
        elites = []

    replaceable_indices = list(range(num_elites, Npop))
    np.random.shuffle(replaceable_indices)
    
    num_to_replace = min(len(offspring), len(replaceable_indices))
    for i in range(num_to_replace):
        idx = replaceable_indices[i]
        new_pop[idx] = offspring[i]
    
    if len(offspring) > num_to_replace:
        new_pop.extend(offspring[num_to_replace:])
        new_pop = new_pop[:Npop]
    
    new_pop = sorted(new_pop, key=lambda x: x[1], reverse=True)
    
    return new_pop


def selection_elitist_replacement(pop, offspring, num_elites=0):
    """
    Элитарная замена
    Сохраняет лучших особей и заменяет худших на потомков
    """
    Npop = len(pop)
    
    if num_elites > 0:
        elites = copy.deepcopy(pop[:num_elites])
    else:
        elites = []
    
    num_to_replace = Npop - num_elites
    num_offspring = min(len(offspring), num_to_replace)
    
    new_pop = elites + offspring[:num_offspring]
    
    if len(new_pop) < Npop:
        remaining = Npop - len(new_pop)
        candidates = pop[num_elites:]
        np.random.shuffle(candidates)
        new_pop.extend(candidates[:remaining])
    

    new_pop = sorted(new_pop, key=lambda x: x[1], reverse=True)
    
    return new_pop


def new_generation(pop, fitness, size, pmut=0.1, num_of_olds=0.1, 
                   parent_selection="roulette", tournament_size=3,
                   mutation_type="combined", selection_method="elitist",
                   num_verts=3, colour="random", alpha="random"):
    """
    Функция создания нового (следующего) поколения
    с различными операторами выбора родителей и отбора
    """
    Npop = len(pop)
    num_elites = int(Npop * num_of_olds)
    
    offspring = []
    
    while len(offspring) < Npop - num_elites:
        if parent_selection == "roulette":
            parent1 = select_parent_roulette(pop)
            parent2 = select_parent_roulette(pop)
        elif parent_selection == "tournament":
            parent1 = select_parent_tournament(pop, tournament_size)
            parent2 = select_parent_tournament(pop, tournament_size)
        else:
            top_half = pop[:max(1, Npop // 2)]
            parent1 = top_half[np.random.randint(len(top_half))]
            parent2 = top_half[np.random.randint(len(top_half))]
        
        child_polygons = crossover(parent1, parent2, size, pmut=0)  # Мутацию применяем отдельно
        
        if np.random.random() < pmut:
            child_polygons = mutation(child_polygons, size, mutation_type=mutation_type,
                                     num_verts=num_verts, colour=colour, alpha=alpha)
        
        child_fitness = individ_fitness(child_polygons, target)
        
        offspring.append((child_polygons, child_fitness))
    
    if selection_method == "uniform":
        new_pop = selection_uniform_replacement(pop, offspring, num_elites=num_elites)
    elif selection_method == "elitist":
        new_pop = selection_elitist_replacement(pop, offspring, num_elites=num_elites)
    else:
        new_pop = []
        for i in range(num_elites):
            new_pop.append(copy.deepcopy(pop[i]))
        new_pop.extend(offspring[:Npop - num_elites])
        new_pop = sorted(new_pop, key=lambda x: x[1], reverse=True)
    
    return new_pop
```

```python
import pandas as pd
import time
import os


def mean(list):
    sum = 0
    for elem in list:
        sum += elem[1]
    return sum/len(list)


def genetic_triangle_painting(target, Npop, Ntri=100, num_verts=3, maxgen=10, every=20, 
                              pmut=0.1, num_of_olds=0.1, colour_init="random", alpha_init="random", 
                              parent_selection="roulette", tournament_size=3,
                              mutation_type="combined", selection_method="elitist",
                              logs=True, outdir="output_images"):

    size = target.size
    
    if not os.path.exists(outdir):
        os.makedirs(outdir)

    pop_time = time.time()
    pop = create_pop(Npop=Npop, N=Ntri, size=size, num_verts=num_verts,
                     colour_init=colour_init, alpha_init=alpha_init)
    fitness = pop_fitness(pop, target)
    pop_time = time.time() - pop_time

    info = []
    maxfit = 0

    for i in range(maxgen+1):
        maxfit = pop[0][1]
        if i % every == 0:
            avgfit = mean(pop)

            best = pop[0][0]
            best_img = draw(best, size=size)
            outpath = outdir + "/generation_" + str(i) + ".png"
            best_img.save(outpath)
            info.append([i, maxfit, avgfit, pop_time, outpath])
            if logs == True:
                print("Generation: " + str(i) + "    Max. Fitness: " + str(np.round(maxfit, 2)) +
                      "%    Avg. Fitness: " + str(np.round(avgfit, 2)) + "%" + "   Time: "+str(np.round(pop_time, 2)))

        pop_time = time.time()
        newpop = new_generation(pop, fitness, size=size,
                                pmut=pmut, num_of_olds=num_of_olds,
                                parent_selection=parent_selection, tournament_size=tournament_size,
                                mutation_type=mutation_type, selection_method=selection_method,
                                num_verts=num_verts, colour=colour_init, alpha=alpha_init)
        fitness = pop_fitness(newpop, target)
        pop_time = time.time() - pop_time
        pop = newpop

    dfout = pd.DataFrame(
        info, columns=["generation", "max_fitness", "avg_fitness", "time", "outpath"])
    return (dfout)
```

```python
result_optimized = genetic_triangle_painting(
    target=target,
    Npop=80,
    Ntri=150,
    num_verts=3,
    maxgen=300,
    every=25,
    pmut=0.12,
    num_of_olds=0.15,
    colour_init="random",
    alpha_init="random",
    parent_selection="tournament",
    tournament_size=5,
    mutation_type="combined",
    selection_method="elitist",
    logs=True,
    outdir="output_images_optimized"
)
print("\nЛучший результат:", result_optimized['max_fitness'].max(), "%")

best_opt = result_optimized.loc[result_optimized['max_fitness'].idxmax()]
best_img_opt = Image.open(best_opt['outpath'])

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(target)
plt.title('Оригинальное изображение')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(best_img_opt)
plt.title(f'Результат\nПоколение {int(best_opt["generation"])}, Fitness: {best_opt["max_fitness"]:.2f}%')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(ImageChops.difference(target, best_img_opt))
plt.title('Разница')
plt.axis('off')

plt.tight_layout()
plt.show()
```
