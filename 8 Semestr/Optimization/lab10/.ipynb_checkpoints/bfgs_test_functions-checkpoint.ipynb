# Метод BFGS для тестовых функций

## Функции:
1. **Розенброка (Rosenbrock)**: $f(x, y) = (1-x)^2 + 100(y-x^2)^2$
2. **Растригина (Rastrigin)**: $f(x, y) = 20 + (x^2 - 10\cos(2\pi x)) + (y^2 - 10\cos(2\pi y))$
3. **Швефеля (Schwefel)**: $f(x, y) = 837.9658 - (x\sin(\sqrt{|x|}) + y\sin(\sqrt{|y|}))$


```python
using LinearAlgebra
using Plots
using Printf
plotly()
```

```python
# Функция Розенброка
function rosenbrock(x)
    return (1 - x[1])^2 + 100 * (x[2] - x[1]^2)^2
end

# Градиент Розенброка
function rosenbrock_gradient(x)
    df_dx1 = -2*(1 - x[1]) - 400*x[1]*(x[2] - x[1]^2)
    df_dx2 = 200*(x[2] - x[1]^2)
    return [df_dx1, df_dx2]
end
```

```python
# Функция Растригина
function rastrigin(x)
    n = length(x)
    A = 10.0
    return A * n + sum(x.^2 .- A * cos.(2 * π * x))
end

# Градиент Растригина
function rastrigin_gradient(x)
    A = 10.0
    df_dx1 = 2*x[1] + 2*π*A*sin(2*π*x[1])
    df_dx2 = 2*x[2] + 2*π*A*sin(2*π*x[2])
    return [df_dx1, df_dx2]
end
```

```python
# Функция Швефеля
function schwefel(x)
    n = length(x)
    return 418.9829 * n - sum(x .* sin.(sqrt.(abs.(x))))
end

# Градиент Швефеля
function schwefel_gradient(x)
    function grad_component(xi)
        abs_xi = abs(xi)
        sqrt_abs = sqrt(abs_xi)
        if abs_xi < 1e-10
            return 0.0
        end
        return -sin(sqrt_abs) - (xi * cos(sqrt_abs)) / (2 * sqrt_abs)
    end
    return [grad_component(x[1]), grad_component(x[2])]
end
```

```python
# Линейный поиск (метод золотого сечения)
function golden_section_search(f, x, p; max_iter=100, tol=1e-6)
    phi = (1 + sqrt(5)) / 2
    a, b = 0.0, 1.0
    
    # Расширяем интервал, если нужно
    while f(x + b * p) < f(x + (b * phi) * p)
        a = b
        b = b * phi
        if b > 100.0  # Ограничение на максимальный шаг
            break
        end
    end
    
    # Метод золотого сечения
    for _ in 1:max_iter
        if abs(b - a) < tol
            break
        end
        
        c = b - (b - a) / phi
        d = a + (b - a) / phi
        
        if f(x + c * p) < f(x + d * p)
            b = d
        else
            a = c
        end
    end
    
    return (a + b) / 2
end
```

```python
function bfgs(f, grad_func, x0; max_iters=1000, epsilon=1e-4)
    x = copy(x0)
    trajectory = [copy(x)]
    n = length(x)
    println("x0 = $x0")
    
    H = Matrix{Float64}(I, n, n)
    println("H_0 = $H")
    
    grad = grad_func(x)
    println("grad_0 = $grad")
    
    for i in 1:max_iters
        println("\nШаг k = $(i-1)")
        temp = norm(grad)
        println("Проверяем критерий остановки: ||grad|| = $(round(temp, digits=6))")
        
        if temp < epsilon
            println("Минимум найден")
            break
        end
        
        p = -H * grad
        println("Направление p = $(round.(p, digits=6))")
        
        # Линейный поиск
        alpha = golden_section_search(x -> f(x), x, p)
        println("Шаг alpha = $(round(alpha, digits=6))")
        
        x_new = x + alpha * p
        println("Новая точка x_new = $(round.(x_new, digits=6))")
        
        grad_new = grad_func(x_new)
        println("Новый градиент grad_new = $(round.(grad_new, digits=6))")
        
        s = x_new - x
        println("Разность координат delta_x (s) = $(round.(s, digits=6))")
        
        y = grad_new - grad
        println("Разность градиентов delta_g (y) = $(round.(y, digits=6))")
        
        rho = 1.0 / dot(y, s)
        println("Коэффициент rho = $(round(rho, digits=6))")
        
        if isfinite(rho) && rho > 0
            I_n = Matrix{Float64}(I, n, n)
            H = (I_n - rho * s * y') * H * (I_n - rho * y * s') + rho * s * s'
            println("Обновленная матрица H = \n$(round.(H, digits=6))")
        end
        
        x = x_new
        grad = grad_new
        push!(trajectory, copy(x))
    end
    
    return x, trajectory
end
```

```python
function visualize_bfgs_method(f, grad_func, name, x0, bounds_pad=2.0)
    println("ФУНКЦИЯ: $name | Старт: $x0")
    
    t1 = time()
    x_bfgs, hist_bfgs = bfgs(f, grad_func, x0)
    time_bfgs = round(time() - t1, digits=6)
    iters1 = length(hist_bfgs) - 1
    
    println("\nBFGS:")
    println("Итераций = $iters1 | Минимум = $(round.(x_bfgs, digits=4)) | Время = $time_bfgs сек")
    println("Значение функции в минимуме: $(round(f(x_bfgs), digits=6))")
    
    all_x1 = [p[1] for p in hist_bfgs]
    all_y1 = [p[2] for p in hist_bfgs]
    
    min_x = minimum(all_x1)
    max_x = maximum(all_x1)
    min_y = minimum(all_y1)
    max_y = maximum(all_y1)
    
    xr = range(min_x - bounds_pad, max_x + bounds_pad, length=100)
    yr = range(min_y - bounds_pad, max_y + bounds_pad, length=100)
    Z = [f([x, y]) for y in yr, x in xr]
    
    p1 = contour(xr, yr, Z, levels=50, color=:grays, fill=false, title="$name (2D)", xlabel="x1", ylabel="x2")
    plot!(p1, all_x1, all_y1, color=:blue, lw=2, marker=:circle, markersize=3, label="BFGS")
    scatter!(p1, [x_bfgs[1]], [x_bfgs[2]], color=:blue, markersize=7, marker=:cross, label="")
    scatter!(p1, [x0[1]], [x0[2]], color=:green, markersize=6, label="")
    
    p2 = surface(xr, yr, Z, color=:grays, alpha=0.5, colorbar=false, title="3D", xlabel="x1", ylabel="x2")
    plot!(p2, all_x1, all_y1, [f(p) for p in hist_bfgs], color=:blue, lw=3, label="BFGS")
    scatter!(p2, [x0[1]], [x0[2]], [f(x0)], color=:green, markersize=5, label="")
    scatter!(p2, [x_bfgs[1]], [x_bfgs[2]], [f(x_bfgs)], color=:blue, markersize=7, marker=:cross, label="")
    
    display(plot(p1, p2, layout=(1,2), size=(1100, 500)))
end
```

## Функция Розенброка
Минимум находится в точке (1, 1), значение функции = 0


```python
x0_rosenbrock = [-1.2, 1.0]
visualize_bfgs_method(rosenbrock, rosenbrock_gradient, "Розенброка", x0_rosenbrock, 3.0)
```

## Функция Растригина
Минимум находится в точке (0, 0), значение функции = 0


```python
x0_rastrigin = [2.0, 2.0]
visualize_bfgs_method(rastrigin, rastrigin_gradient, "Растригина", x0_rastrigin, 5.0)
```

## Функция Швефеля
Минимум находится в точке (420.9687, 420.9687), значение функции ≈ 0


```python
x0_schwefel = [300.0, 300.0]
visualize_bfgs_method(schwefel, schwefel_gradient, "Швефеля", x0_schwefel, 150.0)
```
