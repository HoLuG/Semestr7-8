## Цель работы

Исследовать метод штрафных функций (метод внешней точки) для решения задачи условной оптимизации и проиллюстрировать релаксационную последовательность решений при $r \to \infty$ в двумерном и трёхмерном случаях.



## Теория (кратко)

Рассматривается задача условной оптимизации

$$\min f(x), \quad g(x) \le 0, \; x \in \mathbb{R}^2.$$

В методе внешних штрафных функций вводится вспомогательная функция

$$Q(x, r) = f(x) + r\,G(g(x)), \quad r > 0,$$

где $G(g(x)) > 0$ при нарушении ограничения $g(x) > 0$ и $G(g(x)) = 0$ в допустимой области $g(x) \le 0$. При $r \to \infty$ решение задачи безусловной оптимизации $\min Q(x,r)$ стремится к решению исходной задачи с ограничением.



```julia
using LinearAlgebra
using Plots

gr()

# Целевая квадратичная функция (параболоид) с параметрами α и β
function f(x::AbstractVector, α::Real=1.0, β::Real=1.0)
    return α*(x[1] - 1.0)^2 + β*(x[2] - 2.0)^2
end

# Градиент целевой функции
function ∇f(x::AbstractVector, α::Real=1.0, β::Real=1.0)
    return [2α*(x[1] - 1.0), 2β*(x[2] - 2.0)]
end

# Ограничение-неравенство g(x) ≤ 0  (полуплоскость x₁ + x₂ ≥ 1)
# Здесь g(x) = 1 - x₁ - x₂ ≤ 0
function g(x::AbstractVector)
    return 1.0 - x[1] - x[2]
end

# Штрафная функция для ограничения: G(g) = max(0, g)^2
G(s) = s > 0 ? s^2 : 0.0

# Градиент штрафной части
function ∇G_term(x::AbstractVector, r::Real)
    s = g(x)
    if s > 0
        # ∇(r * s^2) = 2 r s ∇g(x),  ∇g(x) = (-1, -1)
        return 2r*s .* [-1.0, -1.0]
    else
        return zeros(2)
    end
end

# Вспомогательная функция штрафа
function Q(x::AbstractVector, r::Real; α::Real=1.0, β::Real=1.0)
    return f(x, α, β) + r * G(g(x))
end

# Градиент Q
function ∇Q(x::AbstractVector, r::Real; α::Real=1.0, β::Real=1.0)
    return ∇f(x, α, β) + ∇G_term(x, r)
end

# Простой градиентный спуск для минимизации Q(x,r)
function gradient_descent_penalty(x0::AbstractVector, r::Real; α::Real=1.0, β::Real=1.0,
                                  η::Real=0.05, max_iter::Int=2000, tol::Real=1e-6)
    x = collect(Float64, x0)
    path = [copy(x)]
    for k in 1:max_iter
        gk = ∇Q(x, r; α=α, β=β)
        if norm(gk) < tol
            break
        end
        x_new = x - η * gk
        push!(path, copy(x_new))
        if norm(x_new - x) < tol
            x = x_new
            break
        end
        x = x_new
    end
    return x, path
end
```

```julia
# Одномерный случай: f: R -> R, g(x) ≤ 0, Q(x,r) = f(x) + r·G(g(x))

α1 = 1.0
f1(x; α=α1) = α*(x - 1.0)^2           # целевая парабола
g1(x) = 1.0 - x                      # ограничение g(x) ≤ 0  <=>  x ≥ 1
G1(s) = s > 0 ? s^2 : 0.0
Q1(x, r; α=α1) = f1(x; α=α) + r * G1(g1(x))

rs1 = [1.0, 10.0, 100.0, 200.0]
colors1 = [:blue, :green, :orange, :red]

x_range1 = range(-2.0, 4.0, length=400)

# заранее считаем максимум по всем r, чтобы выставить y‑диапазон
vals_all = [Q1(x, r; α=α1) for x in x_range1, r in rs1]
ymin = 0.0
ymax = maximum(vals_all) * 1.05

p1d = plot(xlabel="x", ylabel="Q(x,r)",
           title="Метод внешних штрафов: семейство парабол Q(x,r) (R¹ → R)",
           legend=:topleft, ylim=(ymin, ymax))

# допустимая область x ≥ 1 (заштриховано)
plot!(p1d, [1.0, maximum(x_range1)], [ymin, ymin];
      fillrange=ymax, alpha=0.08, color=:gray, label="допустимая область x ≥ 1")

for (i, r) in enumerate(rs1)
    plot!(p1d, x_range1, Q1.(x_range1, r; α=α1),
          color=colors1[i], label="Q(x,r=$(Int(r)))")
end

p1d
```

```julia
# Трёхмерный случай: поверхности Q(x,r) (R² → R)

α = 1.0
β = 3.0

# используем те же r и цвета, что и в двумерной задаче
rs3 = [1.0, 10.0, 100.0, 200.0]
colors3 = [:blue, :green, :orange, :red]

x1_range3 = range(-3.0, 4.0, length=80)
x2_range3 = range(-3.0, 4.0, length=80)

# Базовая поверхность f(x) (для ориентира)
Z_f = [f([x1, x2], α, β) for x2 in x2_range3, x1 in x1_range3]

p3d = surface(x1_range3, x2_range3, Z_f,
              xlabel="x₁", ylabel="x₂", zlabel="значение",
              title="Метод внешних штрафов: поверхности Q(x,r)",
              c=:plasma, alpha=0.5, label="f(x)")

# Параболоиды Q(x,r) для разных r
for (i, r) in enumerate(rs3)
    Z_Q = [Q([x1, x2], r; α=α, β=β) for x2 in x2_range3, x1 in x1_range3]
    surface!(p3d, x1_range3, x2_range3, Z_Q,
             c=colors3[i], alpha=0.25, label="Q(x,r=$(Int(r)))")
end

# Минимумы Q(x,r) (берём точки из двумерного кода, если он был выполнен)
for (i, r) in enumerate(rs3)
    if i <= length(mins)
        x_star = mins[i]
        scatter!(p3d, [x_star[1]], [x_star[2]], [Q(x_star, r; α=α, β=β)],
                 markersize=6, color=colors3[i], marker=:circle,
                 label="мин. при r = $(Int(r))")
    end
end

# Линия ограничения g(x)=0
x1_line = range(-1.0, 3.0, length=50)
x2_line = 1 .- x1_line
z_line = [f([x1, x2], α, β) for (x1, x2) in zip(x1_line, x2_line)]
plot!(p3d, x1_line, x2_line, z_line, color=:black, linewidth=3, label="g(x)=0")

p3d
```
