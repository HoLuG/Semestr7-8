## Генетический алгоритм для 2D функций (Rosenbrock / Rastrigin / Schwefel)

Ниже — отдельная реализация ГА для **двумерных** benchmark-функций.

- Кодирование: **бинарная хромосома** (по `bits_per_var` бит на каждую координату).
- Декодирование: линейное отображение целого в диапазон 
  \(x \in [x_{min}, x_{max}]\), \(y \in [y_{min}, y_{max}]\).
- Отбор: турнирный.
- Кроссинговер: **1-точечный** и **2-точечный** (для экспериментов можно фиксировать точки разреза).
- Мутация: независимый bit-flip каждого бита с вероятностью `p_mut_bit`.


```julia
using Random
using Statistics
using Plots

# Чтобы графики корректно сохранялись в headless-режиме (если нужно):
# ENV["GKSwstype"] = "100"

# -----------------------------
# 2D benchmark-функции
# -----------------------------
rosenbrock(x, y) = (1 - x)^2 + 100 * (y - x^2)^2
rastrigin(x, y) = 20 + (x^2 - 10cos(2π * x)) + (y^2 - 10cos(2π * y))
schwefel(x, y) = 2 * 418.9829 - (x * sin(sqrt(abs(x))) + y * sin(sqrt(abs(y))))

struct GA2DProblem
    name::String
    f::Function
    x_min::Float64
    x_max::Float64
    y_min::Float64
    y_max::Float64
    optimum_xy::Tuple{Float64, Float64}
    optimum_f::Float64
end

function make_problem(name, f, x_min, x_max, y_min, y_max, opt_xy)
    opt_f = f(opt_xy[1], opt_xy[2])
    return GA2DProblem(name, f, x_min, x_max, y_min, y_max, opt_xy, opt_f)
end

problems = [
    make_problem("Rosenbrock", rosenbrock, -2.0, 2.0, -1.0, 3.0, (1.0, 1.0)),
    make_problem("Rastrigin", rastrigin, -5.12, 5.12, -5.12, 5.12, (0.0, 0.0)),
    make_problem("Schwefel", schwefel, -500.0, 500.0, -500.0, 500.0, (420.968746, 420.968746)),
]

# -----------------------------
# Бинарное кодирование (x,y)
# -----------------------------
struct BinaryEncoding2D
    bits_per_var::Int
end

total_bits(enc::BinaryEncoding2D) = 2 * enc.bits_per_var

# Bool-вектор (MSB→LSB) -> UInt
function bits_to_uint(bits::AbstractVector{Bool})
    u = UInt(0)
    for b in bits
        u = (u << 1) | (b ? UInt(1) : UInt(0))
    end
    return u
end

# UInt -> [0,1] (с учётом максимума 2^n-1)
function uint_to_unit(u::UInt, nbits::Int)
    maxu = (UInt(1) << nbits) - UInt(1)
    return Float64(u) / Float64(maxu)
end

# Декодирование хромосомы -> (x,y)
function decode_xy(ch::AbstractVector{Bool}, prob::GA2DProblem, enc::BinaryEncoding2D)
    n = enc.bits_per_var
    @assert length(ch) == 2n
    bx = @view ch[1:n]
    by = @view ch[n+1:2n]

    ux = bits_to_uint(bx)
    uy = bits_to_uint(by)

    tx = uint_to_unit(ux, n)
    ty = uint_to_unit(uy, n)

    x = prob.x_min + tx * (prob.x_max - prob.x_min)
    y = prob.y_min + ty * (prob.y_max - prob.y_min)
    return (x, y)
end

# Приспособленность для минимизации: больше = лучше
fitness(ch, prob, enc) = -prob.f(decode_xy(ch, prob, enc)...)

# -----------------------------
# Операторы ГА
# -----------------------------
random_chromosome(enc::BinaryEncoding2D, rng::AbstractRNG) = rand(rng, Bool, total_bits(enc))

function tournament_select(pop::Vector{Vector{Bool}}, fits::Vector{Float64}, rng::AbstractRNG; k::Int=3)
    idxs = rand(rng, 1:length(pop), k)
    best = idxs[1]
    for i in idxs[2:end]
        if fits[i] > fits[best]
            best = i
        end
    end
    return pop[best]
end

function one_point_crossover_at(p1::Vector{Bool}, p2::Vector{Bool}, point::Int)
    n = length(p1)
    @assert n == length(p2)
    @assert 1 ≤ point ≤ n - 1
    c1 = vcat(p1[1:point], p2[point+1:end])
    c2 = vcat(p2[1:point], p1[point+1:end])
    return c1, c2
end

function two_point_crossover_at(p1::Vector{Bool}, p2::Vector{Bool}, a::Int, b::Int)
    n = length(p1)
    @assert n == length(p2)
    @assert 1 ≤ a < b ≤ n - 1
    c1 = copy(p1)
    c2 = copy(p2)
    c1[a+1:b] = p2[a+1:b]
    c2[a+1:b] = p1[a+1:b]
    return c1, c2
end

function mutate_bitflip!(ch::Vector{Bool}, p_mut_bit::Float64, rng::AbstractRNG)
    @assert 0.0 ≤ p_mut_bit ≤ 1.0
    for i in eachindex(ch)
        if rand(rng) < p_mut_bit
            ch[i] = !ch[i]
        end
    end
    return ch
end

# -----------------------------
# Запуск ГА (возвращает историю лучшего)
# -----------------------------
function run_ga_2d(prob::GA2DProblem;
    enc::BinaryEncoding2D = BinaryEncoding2D(16),
    pop_size::Int = 40,
    generations::Int = 400,
    elite_count::Int = 2,
    p_cross::Float64 = 0.9,
    p_mut_bit::Float64 = 0.01,
    crossover::Symbol = :one_point,  # :one_point или :two_point
    crossover_points = nothing,      # Int для :one_point, (a,b) для :two_point, или nothing (случайно)
    tournament_k::Int = 3,
    seed::Int = 42,
)
    rng = MersenneTwister(seed)
    nbits = total_bits(enc)

    pop = [random_chromosome(enc, rng) for _ in 1:pop_size]

    best_f_hist = Float64[]
    best_xy_hist = Tuple{Float64, Float64}[]

    function update_best!(pop)
        vals = [prob.f(decode_xy(ch, prob, enc)...) for ch in pop]
        i = argmin(vals)
        push!(best_f_hist, vals[i])
        push!(best_xy_hist, decode_xy(pop[i], prob, enc))
        return vals
    end

    vals = update_best!(pop)

    for _ in 1:generations
        fits = [-v for v in vals]

        elite_idx = partialsortperm(vals, 1:elite_count)  # минимизация
        new_pop = [copy(pop[i]) for i in elite_idx]

        while length(new_pop) < pop_size
            p1 = tournament_select(pop, fits, rng; k=tournament_k)
            p2 = tournament_select(pop, fits, rng; k=tournament_k)

            c1 = copy(p1)
            c2 = copy(p2)

            if rand(rng) < p_cross
                if crossover == :one_point
                    point = crossover_points === nothing ? rand(rng, 1:nbits-1) : crossover_points
                    c1, c2 = one_point_crossover_at(p1, p2, point)
                elseif crossover == :two_point
                    if crossover_points === nothing
                        a = rand(rng, 1:nbits-2)
                        b = rand(rng, a+1:nbits-1)
                        c1, c2 = two_point_crossover_at(p1, p2, a, b)
                    else
                        a, b = crossover_points
                        c1, c2 = two_point_crossover_at(p1, p2, a, b)
                    end
                else
                    error("Unknown crossover=$(crossover)")
                end
            end

            mutate_bitflip!(c1, p_mut_bit, rng)
            push!(new_pop, c1)

            if length(new_pop) < pop_size
                mutate_bitflip!(c2, p_mut_bit, rng)
                push!(new_pop, c2)
            end
        end

        pop = new_pop
        vals = update_best!(pop)
    end

    best_i = argmin(best_f_hist)
    return (
        best_xy = best_xy_hist[best_i],
        best_f = best_f_hist[best_i],
        best_f_hist = best_f_hist,
        best_xy_hist = best_xy_hist,
        enc = enc,
    )
end
```

```julia
# -----------------------------
# Визуализация: 3D поверхность + 2D проекция + траектория лучшего
# -----------------------------
function plot_surface_and_contour(prob::GA2DProblem, result; grid::Int=120, title_suffix::String="")
    xs = range(prob.x_min, prob.x_max; length=grid)
    ys = range(prob.y_min, prob.y_max; length=grid)
    Z = [prob.f(x, y) for y in ys, x in xs]

    p3 = surface(xs, ys, Z;
        xlabel="x", ylabel="y", zlabel="f(x,y)",
        title="$(prob.name) — 3D поверхность$(title_suffix)",
        alpha=0.85, color=:viridis, legend=:topright)

    p2 = contourf(xs, ys, Z;
        xlabel="x", ylabel="y",
        title="$(prob.name) — 2D проекция (контуры)$(title_suffix)",
        levels=40, color=:viridis, legend=:topright)

    # траектория лучшего решения
    path = result.best_xy_hist
    px = first.(path)
    py = last.(path)
    pz = [prob.f(x, y) for (x, y) in path]

    plot!(p3, px, py, pz; color=:red, linewidth=2, label="best path")
    scatter!(p3, [px[end]], [py[end]], [pz[end]]; marker=:star5, ms=10, color=:green, label="best")

    plot!(p2, px, py; color=:red, linewidth=2, label="best path")
    scatter!(p2, [px[end]], [py[end]]; marker=:star5, ms=10, color=:green, label="best")

    # известный минимум
    ox, oy = prob.optimum_xy
    scatter!(p3, [ox], [oy], [prob.optimum_f]; marker=:diamond, ms=7, color=:white, markerstrokecolor=:black, label="known")
    scatter!(p2, [ox], [oy]; marker=:diamond, ms=7, color=:white, markerstrokecolor=:black, label="known")

    return plot(p3, p2; layout=(1,2), size=(1400, 600))
end

function plot_convergence_curves(prob::GA2DProblem, labeled_results; title_text::String)
    p = plot(title=title_text, xlabel="generation", ylabel="f_best - f*", yscale=:log10,
        linewidth=2, legend=:topright, size=(1000, 650))

    fstar = prob.optimum_f
    eps = 1e-12
    for (label, res) in labeled_results
        gaps = [max(v - fstar, eps) for v in res.best_f_hist]
        plot!(p, 0:length(gaps)-1, gaps; label=label)
    end
    return p
end

# Быстрый базовый прогон + графики для каждой функции
Random.seed!(42)
enc = BinaryEncoding2D(16)

for prob in problems
    grid = prob.name == "Schwefel" ? 90 : 130

    res = run_ga_2d(prob; enc=enc, pop_size=50, generations=500, elite_count=2,
        p_cross=0.9, p_mut_bit=0.01, crossover=:one_point, crossover_points=nothing, seed=42)

    println("\n--- $(prob.name) ---")
    println("best_xy = ", res.best_xy)
    println("best_f  = ", res.best_f)
    println("known_f = ", prob.optimum_f)

    p = plot_surface_and_contour(prob, res; grid=grid)
    display(p)
end
```

## Эксперименты по мутации и точкам кроссинговера

Ниже — сравнение сходимости:

- **при разных вероятностях bit-flip мутации** `p_mut_bit`
- **при разных точках кроссинговера** (для 1-точечного) и разных парах точек (для 2-точечного)

Метрика на графике: \(f_{best}(t) - f^*\) в логарифмической шкале.


```julia
# -----------------------------
# Sweeps: mutation / crossover-point
# -----------------------------
function mutation_sweep(prob::GA2DProblem;
    enc::BinaryEncoding2D=BinaryEncoding2D(16),
    p_muts = [0.0, 0.002, 0.005, 0.01, 0.02, 0.05],
    crossover::Symbol=:one_point,
    crossover_points=nothing,
    seed0::Int=100,
)
    labeled = Tuple{String, Any}[]
    for (i, pm) in enumerate(p_muts)
        res = run_ga_2d(prob; enc=enc, pop_size=50, generations=500, elite_count=2,
            p_cross=0.9, p_mut_bit=pm, crossover=crossover, crossover_points=crossover_points,
            seed=seed0 + i)
        push!(labeled, ("p_mut_bit=$(pm)", res))
    end
    return labeled
end

# Варианты точек кроссинговера (1-точечный): позиции как доля длины хромосомы
function one_point_positions(enc::BinaryEncoding2D, ratios = [0.2, 0.35, 0.5, 0.65, 0.8])
    n = total_bits(enc)
    pts = [clamp(round(Int, r*(n-1)), 1, n-1) for r in ratios]
    return unique(pts)
end

# Варианты пар точек (2-точечный): (a,b) как доли
function two_point_pairs(enc::BinaryEncoding2D, pairs = [(0.2,0.6), (0.25,0.7), (0.3,0.8)])
    n = total_bits(enc)
    out = Tuple{Int,Int}[]
    for (ra, rb) in pairs
        a = clamp(round(Int, ra*(n-1)), 1, n-2)
        b = clamp(round(Int, rb*(n-1)), a+1, n-1)
        push!(out, (a,b))
    end
    return unique(out)
end

function crossover_sweep_one_point(prob::GA2DProblem;
    enc::BinaryEncoding2D=BinaryEncoding2D(16),
    p_mut_bit::Float64=0.01,
    ratios=[0.2, 0.35, 0.5, 0.65, 0.8],
    seed0::Int=200,
)
    pts = one_point_positions(enc, ratios)
    labeled = Tuple{String, Any}[]
    for (i, pt) in enumerate(pts)
        res = run_ga_2d(prob; enc=enc, pop_size=50, generations=500, elite_count=2,
            p_cross=0.9, p_mut_bit=p_mut_bit, crossover=:one_point, crossover_points=pt,
            seed=seed0 + i)
        push!(labeled, ("cp=$(pt)", res))
    end
    return labeled
end

function crossover_sweep_two_point(prob::GA2DProblem;
    enc::BinaryEncoding2D=BinaryEncoding2D(16),
    p_mut_bit::Float64=0.01,
    pair_ratios=[(0.2,0.6), (0.25,0.7), (0.3,0.8)],
    seed0::Int=300,
)
    prs = two_point_pairs(enc, pair_ratios)
    labeled = Tuple{String, Any}[]
    for (i, pr) in enumerate(prs)
        res = run_ga_2d(prob; enc=enc, pop_size=50, generations=500, elite_count=2,
            p_cross=0.9, p_mut_bit=p_mut_bit, crossover=:two_point, crossover_points=pr,
            seed=seed0 + i)
        push!(labeled, ("cp=$(pr)", res))
    end
    return labeled
end

# -----------------------------
# Запуск экспериментов и графики
# -----------------------------
enc = BinaryEncoding2D(16)

for prob in problems
    println("\n=== Experiments: $(prob.name) ===")

    # 1) Сходимость при разных мутациях (фиксируем кроссинговер)
    mut_res = mutation_sweep(prob; enc=enc, p_muts=[0.0, 0.002, 0.005, 0.01, 0.02, 0.05], crossover=:one_point)
    p_mut = plot_convergence_curves(prob, mut_res; title_text="$(prob.name): convergence vs mutation")
    display(p_mut)

    # 2) Сходимость при разных точках 1-точечного кроссинговера
    cross_res = crossover_sweep_one_point(prob; enc=enc, p_mut_bit=0.01, ratios=[0.2, 0.35, 0.5, 0.65, 0.8])
    p_cp = plot_convergence_curves(prob, cross_res; title_text="$(prob.name): convergence vs 1-point crossover position")
    display(p_cp)

    # 3) (доп.) 2-точечный кроссинговер: разные пары точек
    cross2_res = crossover_sweep_two_point(prob; enc=enc, p_mut_bit=0.01, pair_ratios=[(0.2,0.6), (0.25,0.7), (0.3,0.8)])
    p_cp2 = plot_convergence_curves(prob, cross2_res; title_text="$(prob.name): convergence vs 2-point crossover (pairs)")
    display(p_cp2)
end
```
