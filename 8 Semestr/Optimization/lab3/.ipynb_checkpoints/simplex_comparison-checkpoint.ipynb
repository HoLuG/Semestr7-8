# Сравнение метода простого симплекса и метода Нельдера-Мида

Этот notebook сравнивает два метода оптимизации:
- **Метод простого симплекса**: базовая версия с отражением и сжатием
- **Метод Нельдера-Мида**: расширенная версия с отражением, растяжением, сжатием и уменьшением

Тестирование проводится на функциях:
- Розенброк
- Растригин
- Швефель


```julia
using Plots
using LinearAlgebra

gr()  # Используем gr() для лучшей совместимости
```

```julia
# Тестовые функции
function rosenbrock(x)
    return (1.0 - x[1])^2 + 100.0*(x[2] - x[1]^2)^2
end

function rastrigin(x)
    n = length(x)
    return 20 + sum(x[i]^2 - 10*cos(2*pi*x[i]) for i in 1:n)
end

function schwefel(x)
    n = length(x)
    return 418.9829*n - sum(x[i]*sin(sqrt(abs(x[i]))) for i in 1:n)
end
```

```julia
# Функция для создания начального симплекса для простого метода (правильный симплекс)
function create_initial_simplex_simple(x0, n, alpha=1.0)
    # Для 2D используем формулу из PDF
    if n == 2
        x1 = deepcopy(x0)
        x2 = x0 .+ alpha .* [(sqrt(3)+1)/(2*sqrt(2)), (sqrt(3)-1)/(2*sqrt(2))]
        x3 = x0 .+ alpha .* [(sqrt(3)-1)/(2*sqrt(2)), (sqrt(3)+1)/(2*sqrt(2))]
        return [x1, x2, x3]
    else
        # Для других размерностей используем стандартный способ
        simplex = [deepcopy(x0)]
        q = (sqrt(n+1) - 1) / (n * sqrt(2))
        p = q + 1 / sqrt(2)
        for i in 1:n
            x_new = deepcopy(x0)
            for j in 1:n
                if j == i
                    x_new[j] += alpha * p
                else
                    x_new[j] += alpha * q
                end
            end
            push!(simplex, x_new)
        end
        return simplex
    end
end

# Функция для создания начального симплекса для Нельдера-Мида (простой)
function create_initial_simplex_nm(x0, n, alpha=0.05)
    # Простой симплекс как в PDF: x0, x0+[alpha,0,...], x0+[0,alpha,...]
    simplex = [deepcopy(x0)]
    for i in 1:n
        x_new = deepcopy(x0)
        x_new[i] += alpha
        push!(simplex, x_new)
    end
    return simplex
end
```

```julia
function simple_simplex_method(f, x0, eps=1e-3, alpha=1.0, max_iterations=5000)
    n = length(x0)
    simplex = create_initial_simplex_simple(x0, n, alpha)
    
    # Траектории для визуализации
    trajectory = [copy(x0)]
    all_simplexes = [copy(simplex)]
    
    iteration = 0
    
    while iteration < max_iterations
        iteration += 1
        
        if norm(simplex[1] - simplex[2]) < eps
            break
        end
        
        # Вычисляем значения функции
        vals = f.(simplex)
        worst_idx = argmax(vals)
        
        # Меняем местами худшую и первую точку
        simplex[1], simplex[worst_idx] = simplex[worst_idx], simplex[1]
        
        # Отражение (как в PDF: x4 = points[2] + points[3] - points[1])
        x_reflected = simplex[2] .+ simplex[3] .- simplex[1]
        f_reflected = f(x_reflected)
        f_good = f(simplex[2])
        f_better = f(simplex[3])
        
        if f_reflected >= f_good && f_reflected >= f_better
            # Отражение не лучше - сжимаем к x_reflected
            simplex[1] = x_reflected
            simplex[2] = x_reflected .+ (simplex[2] .- x_reflected) ./ 2
            simplex[3] = x_reflected .+ (simplex[3] .- x_reflected) ./ 2
            push!(trajectory, copy(x_reflected))
        else
            # Отражение лучше - принимаем его
            simplex[1] = x_reflected
            push!(trajectory, copy(x_reflected))
        end
        
        # Сохраняем симплекс для визуализации
        push!(all_simplexes, copy(simplex))
    end
    
    # Возвращаем лучшую точку (первая точка после сортировки)
    result = simplex[1]
    
    return result, trajectory, all_simplexes, iteration
end
```

```julia
function nelder_mead_method(f, x0, eps=1e-3, alpha=1.0, max_iterations=5000)
    n = length(x0)
    simplex = create_initial_simplex_nm(x0, n, 0.05)
    
    α = 1.0   # Коэффициент отражения (alpha)
    β = 2.0   # Коэффициент растяжения (betha)
    ρ = 0.5   # Коэффициент сжатия (ro)
    δ = 0.5   # Коэффициент уменьшения (delta)
    
    # Траектории для визуализации
    trajectory = [copy(x0)]
    all_simplexes = [copy(simplex)]
    
    iteration = 0
    
    while iteration < max_iterations
        iteration += 1
        
        simplex = sort(simplex, by=f)
        
        best = simplex[1]
        good = simplex[2]
        worst = simplex[3]
        
        # Проверка на сходимость (как в PDF)
        if norm(best - worst) < eps
            break
        end
        
        # Центр двух лучших точек (как в PDF: centroid = (best + good) / 2)
        centroid = (best .+ good) ./ 2
        
        # 1. Отражение
        x_reflected = centroid .+ α .* (centroid .- worst)
        f_reflected = f(x_reflected)
        f_best = f(best)
        f_good = f(good)
        f_worst = f(worst)
        
        if f_reflected < f_best
            # Растяжение (если отражение лучше лучшей точки)
            x_expanded = centroid .+ β .* (x_reflected .- centroid)
            f_expanded = f(x_expanded)
            
            if f_expanded < f_reflected
                simplex[3] = x_expanded
                push!(trajectory, copy(x_expanded))
            else
                simplex[3] = x_reflected
                push!(trajectory, copy(x_reflected))
            end
            
        elseif f_reflected < f_good
            # Отражение хорошее - принимаем его
            simplex[3] = x_reflected
            push!(trajectory, copy(x_reflected))
            
        else
            # Сжатие
            x_contracted = centroid .+ ρ .* (worst .- centroid)
            f_contracted = f(x_contracted)
            
            if f_contracted < f_worst
                simplex[3] = x_contracted
                push!(trajectory, copy(x_contracted))
            else
                # Уменьшение (если сжатие не помогло)
                simplex[2] = best .+ δ .* (simplex[2] .- best)
                simplex[3] = best .+ δ .* (simplex[3] .- best)
                push!(trajectory, copy(best))
            end
        end
        
        push!(all_simplexes, copy(simplex))
    end
    
    # Возвращаем лучшую точку (первая точка после сортировки)
    result = simplex[1]
    
    return result, trajectory, all_simplexes, iteration
end
```

```julia
# Функция для создания сравнительной 3D визуализации (для 2D)
function plot_comparison_3d(f, x0, result_simple, result_nm, 
                           traj_simple, traj_nm, 
                           simplexes_simple, simplexes_nm,
                           title_str, x_range, y_range)
    # Подготовка данных для 3D графика
    x_grid = range(x_range[1], x_range[2], length=100)
    y_grid = range(y_range[1], y_range[2], length=100)
    X = repeat(reshape(x_grid, 1, :), length(y_grid), 1)
    Y = repeat(y_grid, 1, length(x_grid))
    Z = map((a,b) -> f([a,b]), X, Y)
    
    plt = surface(x_grid, y_grid, Z, color=:thermal, alpha=0.5, legend=true)
    
    # Рисуем симплексы простого метода (красные треугольники)
    step_simple = max(1, length(simplexes_simple) ÷ 1)  # Показываем больше симплексов
    for (idx, simplex) in enumerate(simplexes_simple[1:step_simple:end])
        if length(simplex) >= 3
            x_coords = [simplex[1][1], simplex[2][1], simplex[3][1], simplex[1][1]]
            y_coords = [simplex[1][2], simplex[2][2], simplex[3][2], simplex[1][2]]
            z_coords = [f(simplex[1]), f(simplex[2]), f(simplex[3]), f(simplex[1])]
            plot!(plt, x_coords, y_coords, z_coords, linecolor=:red, linewidth=2.0, 
                  label=(idx == 1 ? "Простой симплекс" : ""), alpha=0.8, linestyle=:solid)
        end
    end
    
    # Рисуем симплексы Нельдера-Мида (синие треугольники)
    step_nm = max(1, length(simplexes_nm) ÷ 1)  # Показываем больше симплексов
    for (idx, simplex) in enumerate(simplexes_nm[1:step_nm:end])
        if length(simplex) >= 3
            x_coords = [simplex[1][1], simplex[2][1], simplex[3][1], simplex[1][1]]
            y_coords = [simplex[1][2], simplex[2][2], simplex[3][2], simplex[1][2]]
            z_coords = [f(simplex[1]), f(simplex[2]), f(simplex[3]), f(simplex[1])]
            plot!(plt, x_coords, y_coords, z_coords, linecolor=:blue, linewidth=2.0, 
                  label=(idx == 1 ? "Нельдер-Мид" : ""), alpha=0.8, linestyle=:dash)
        end
    end
    
    # Начальная точка
    scatter3d!(plt, [x0[1]], [x0[2]], [f(x0)], color=:yellow, markersize=8, label="Старт")
    
    # Конечные точки
    scatter3d!(plt, [result_simple[1]], [result_simple[2]], [f(result_simple)], 
              color=:purple, markersize=8, label="Финиш (простой)")
    scatter3d!(plt, [result_nm[1]], [result_nm[2]], [f(result_nm)], 
              color=:cyan, markersize=8, label="Финиш (Нельдер-Мид)")
    
    title!(plt, title_str)
    xlabel!(plt, "x")
    ylabel!(plt, "y")
    zlabel!(plt, "f(x,y)")
    
    return plt
end
```

```julia
# Функция для создания сравнительного 2D контурного графика
function plot_comparison_2d(f, x0, result_simple, result_nm,
                            traj_simple, traj_nm,
                            simplexes_simple, simplexes_nm,
                            title_str, x_range, y_range)
    x_grid = range(x_range[1], x_range[2], length=100)
    y_grid = range(y_range[1], y_range[2], length=100)
    X = repeat(reshape(x_grid, 1, :), length(y_grid), 1)
    Y = repeat(y_grid, 1, length(x_grid))
    Z = map((a,b) -> f([a,b]), X, Y)
    
    plt = contour(x_grid, y_grid, Z, 
        color=:thermal, 
        levels=20,
        xlabel="x₁", 
        ylabel="x₂",
        title=title_str,
        legend=:topleft
    )
    
    # Рисуем симплексы простого метода (красные треугольники)
    step_simple = max(1, length(simplexes_simple) ÷ 1)  # Показываем больше симплексов
    for (idx, simplex) in enumerate(simplexes_simple[1:step_simple:end])
        if length(simplex) >= 3
            x_coords = [simplex[1][1], simplex[2][1], simplex[3][1], simplex[1][1]]
            y_coords = [simplex[1][2], simplex[2][2], simplex[3][2], simplex[1][2]]
            plot!(plt, x_coords, y_coords, color=:red, linewidth=2.5, 
                  label=(idx == 1 ? "Простой симплекс" : ""), alpha=0.8, linestyle=:solid)
        end
    end
    
    # Рисуем симплексы Нельдера-Мида (синие треугольники)
    step_nm = max(1, length(simplexes_nm) ÷ 1)  # Показываем больше симплексов
    for (idx, simplex) in enumerate(simplexes_nm[1:step_nm:end])
        if length(simplex) >= 3
            x_coords = [simplex[1][1], simplex[2][1], simplex[3][1], simplex[1][1]]
            y_coords = [simplex[1][2], simplex[2][2], simplex[3][2], simplex[1][2]]
            plot!(plt, x_coords, y_coords, color=:blue, linewidth=2.5, 
                  label=(idx == 1 ? "Нельдер-Мид" : ""), alpha=0.8, linestyle=:dash)
        end
    end
    
    # Начальная точка
    scatter!(plt, [x0[1]], [x0[2]], color=:yellow, markersize=8, label="Старт")
    
    # Конечные точки
    scatter!(plt, [result_simple[1]], [result_simple[2]], color=:purple, markersize=8, label="Финиш (простой)")
    scatter!(plt, [result_nm[1]], [result_nm[2]], color=:cyan, markersize=8, label="Финиш (Нельдер-Мид)")
    
    plot!(plt, colorbar=true)
    # Путь простого симплекса
    plot!(plt,
        [p[1] for p in traj_simple],
        [p[2] for p in traj_simple],
        color=:red, linewidth=2, label="Путь (простой)"
    )
    
    # Путь Нелдера–Мида
    plot!(plt,
        [p[1] for p in traj_nm],
        [p[2] for p in traj_nm],
        color=:blue, linewidth=2, linestyle=:dash, label="Путь (Нелдер-Мид)"
    )
    
    return plt
end
```

```julia
# Функция для сравнения методов
function compare_methods(f, f_name, x0, x_range, y_range)
    println("\n" * "="^70)
    println("СРАВНЕНИЕ МЕТОДОВ для функции: $f_name")
    println("Начальная точка: $x0")
    println("="^70)
    
    # Запускаем оба метода
    result_simple, traj_simple, simplexes_simple, iter_simple = simple_simplex_method(f, x0)
    result_nm, traj_nm, simplexes_nm, iter_nm = nelder_mead_method(f, x0)
    
    f_simple = f(result_simple)
    f_nm = f(result_nm)
    
    # Выводим результаты
    println("\n--- Метод простого симплекса ---")
    println("Результат: $result_simple")
    println("Значение функции: $f_simple")
    println("Количество итераций: $iter_simple")
    println("Количество симплексов: $(length(simplexes_simple))")
    println("Количество точек траектории: $(length(traj_simple))")
    
    println("\n--- Метод Нельдера-Мида ---")
    println("Результат: $result_nm")
    println("Значение функции: $f_nm")
    println("Количество итераций: $iter_nm")
    println("Количество симплексов: $(length(simplexes_nm))")
    println("Количество точек траектории: $(length(traj_nm))")
    
    println("\n--- Сравнительная характеристика ---")
    println("Лучшее значение функции:")
    if f_simple < f_nm
        println("  ✓ Простой симплекс: $f_simple (лучше на $(f_nm - f_simple))")
    elseif f_nm < f_simple
        println("  ✓ Нельдер-Мид: $f_nm (лучше на $(f_simple - f_nm))")
    else
        println("  = Оба метода дали одинаковый результат: $f_simple")
    end
    
    println("Скорость сходимости (итерации):")
    if iter_simple < iter_nm
        println("  ✓ Простой симплекс: $iter_simple итераций (быстрее в $(iter_nm / iter_simple) раз)")
    elseif iter_nm < iter_simple
        println("  ✓ Нельдер-Мид: $iter_nm итераций (быстрее в $(iter_simple / iter_nm) раз)")
    else
        println("  = Оба метода использовали одинаковое количество итераций: $iter_simple")
    end
    
    println("Количество вычислений функции (приблизительно):")
    evals_simple = iter_simple * (length(x0) + 1)  # Примерная оценка
    evals_nm = iter_nm * (length(x0) + 2)  # Нельдер-Мид может делать больше вычислений
    println("  Простой симплекс: ~$evals_simple")
    println("  Нельдер-Мид: ~$evals_nm")
    
    # Создание графиков
    title_3d = "Сравнение методов - $f_name (3D)"
    title_2d = "Сравнение методов - $f_name (2D)"
    
    plt_3d = plot_comparison_3d(f, x0, result_simple, result_nm,
                               traj_simple, traj_nm,
                               simplexes_simple, simplexes_nm,
                               title_3d, x_range, y_range)
    
    plt_2d = plot_comparison_2d(f, x0, result_simple, result_nm,
                               traj_simple, traj_nm,
                               simplexes_simple, simplexes_nm,
                               title_2d, x_range, y_range)
    
    # Отображение графиков
    display(plt_3d)
    display(plt_2d)
    
    return result_simple, result_nm, traj_simple, traj_nm, plt_3d, plt_2d
end
```

## Тест 1: Функция Розенброк


```julia
x0_rosenbrock = [-2.0, -2.0]
result_r1_simple, result_r1_nm, traj_r1_simple, traj_r1_nm, plt_3d_r1, plt_2d_r1 = 
    compare_methods(rosenbrock, "Розенброк", x0_rosenbrock, (-2.0, 2.0), (-2.0, 2.0))
```

## Тест 2: Функция Растригин


```julia
x0_rastrigin = [0.5, 0.5]
result_r2_simple, result_r2_nm, traj_r2_simple, traj_r2_nm, plt_3d_r2, plt_2d_r2 = 
    compare_methods(rastrigin, "Растригин", x0_rastrigin, (-5.0, 5.0), (-5.0, 5.0))
```

## Тест 3: Функция Швефель


```julia
x0_schwefel = [0.5, 0.5]
result_r3_simple, result_r3_nm, traj_r3_simple, traj_r3_nm, plt_3d_r3, plt_2d_r3 = 
    compare_methods(schwefel, "Швефель", x0_schwefel, (-10.0, 10.0), (-10.0, 10.0))
```
