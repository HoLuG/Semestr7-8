```julia
using Plots
using LinearAlgebra

gr()

function rosenbrock(x)
    return (1.0 - x[1])^2 + 100.0*(x[2] - x[1]^2)^2
end

function rastrigin(x)
    n = length(x)
    return 20 + sum(x[i]^2 - 10*cos(2*pi*x[i]) for i in 1:n)
end

function schwefel(x)
    n = length(x)
    return 418.9829*n - sum(x[i]*sin(sqrt(abs(x[i]))) for i in 1:n)
end

# Вариант А:
function hooke_jeeves_variant_a(f, x0, eps=1e-6, delta0=0.1)
    n = length(x0)
    delta = fill(delta0, n)
    x = copy(x0)
    
    trajectory = [copy(x)]  # Релаксационная последовательность
    search_points = [copy(x)]  # Точки поиска направления
    max_points = []  # Красные точки - где функция не убывает
    trajectory_colors = [:green]  # Цвета точек траектории
    
    iteration = 0
    max_iterations = 10000
    
    while maximum(delta) > eps && iteration < max_iterations
        iteration += 1
        x_prev = copy(x)
        f_prev = f(x_prev)
        
        # Исследовательский поиск - смотрим во все стороны (по 1 клетке)
        x_new = copy(x)
        improved = false
        
        for i in 1:n
            # +delta
            x_test = copy(x_new)
            x_test[i] += delta[i]
            f_test = f(x_test)
            push!(search_points, copy(x_test))  # Отмечаем все проверенные точки
            
            if f_test < f(x_new)
                x_new = x_test
                improved = true
            else
                # -delta
                x_test = copy(x_new)
                x_test[i] -= delta[i]
                f_test = f(x_test)
                push!(search_points, copy(x_test))  # Отмечаем все проверенные точки
                
                if f_test < f(x_new)
                    x_new = x_test
                    improved = true
                end
            end
        end
        
        if f(x_new) < f_prev
            x = copy(x_new)
            push!(trajectory, copy(x))
            push!(trajectory_colors, :green)
        else

            push!(max_points, copy(x))
            delta = delta / 2.0
            
            if maximum(delta) < eps || abs(f(x) - f_prev) < eps
                break
            end
        end
    end
    
    return x, trajectory, search_points, max_points, trajectory_colors
end

function exploratory_search(f, x, delta, include_start=true)
    n = length(x)
    x_new = copy(x)
    direction_points = include_start ? [copy(x)] : []  # Начальную точку добавляем только если нужно

    for i in 1:n
        start_val = f(x_new)
        temp = x_new[i]
        
        # Пробуем +delta
        x_test_plus = copy(x_new)
        x_test_plus[i] = temp + delta[i]
        f_plus = f(x_test_plus)
        push!(direction_points, copy(x_test_plus))  # Добавляем проверенную точку
        
        # Пробуем -delta
        x_test_minus = copy(x_new)
        x_test_minus[i] = temp - delta[i]
        f_minus = f(x_test_minus)
        push!(direction_points, copy(x_test_minus))  # Добавляем проверенную точку

        # Выбираем лучшую точку
        if f_plus < start_val
            x_new = x_test_plus
        elseif f_minus < start_val
            x_new = x_test_minus
        end
        # Если обе хуже, остаёмся на месте (x_new не меняется)
    end

    return x_new, direction_points
end

# Вариант Б:
function hooke_jeeves_variant_b(f, x0, eps=1e-6, delta0=0.1)
    n = length(x0)
    delta = fill(delta0, n)
    x = copy(x0)
    
    trajectory = [copy(x)]  # Релаксационная последовательность
    search_points = [copy(x)]  # Точки поиска направления
    max_points = []  # Красные точки - где функция не убывает
    trajectory_colors = [:green]
    
    iteration = 0
    max_iterations = 10000
    
    while maximum(delta) > eps && iteration < max_iterations
        iteration += 1
        x_prev = copy(x)
        f_prev = f(x_prev)
        
        # Исследовательский поиск - находим направление ОДИН РАЗ
        # Не добавляем начальную точку, так как она уже есть в search_points
        x_exp, direction_search_points = exploratory_search(f, x, delta, false)
        append!(search_points, direction_search_points)
        
        if f(x_exp) >= f(x)
            push!(max_points, copy(x))
            delta = delta / 2.0
            continue
        end
        
        direction = x_exp - x
        step_size = norm(direction)
        
        if step_size > eps
            direction_normalized = direction / step_size
            
            x_current = copy(x_exp)
            f_current = f(x_current)
            improved = true
            
            while improved
                x_next = x_current + direction_normalized * step_size
                f_next = f(x_next)
                
                if f_next < f_current
                    x_current = x_next
                    f_current = f_next
                    push!(trajectory, copy(x_current))
                    push!(trajectory_colors, :green)
                else
                    improved = false
                end
            end
            
            x_rollback = x_current - direction_normalized * step_size / 2.0
            f_rollback = f(x_rollback)
            
            if f_rollback < f_current
                x = copy(x_rollback)
            else
                x = copy(x_current)
            end
            
            push!(trajectory, copy(x))
            push!(trajectory_colors, :green)
        else
            x = copy(x_exp)
            push!(trajectory, copy(x))
            push!(trajectory_colors, :green)
        end
        
        if norm(x - x_prev) < eps || abs(f(x) - f_prev) < eps
            break
        end
    end
    
    return x, trajectory, search_points, max_points, trajectory_colors
end

# Функция для создания 3D визуализации
function plot_optimization_3d(f, x0, result, trajectory, search_points, max_points, trajectory_colors, title_str, x_range, y_range)
    # Подготовка данных для 3D графика
    x_grid = range(x_range[1], x_range[2], length=100)
    y_grid = range(y_range[1], y_range[2], length=100)
    X = repeat(reshape(x_grid, 1, :), length(y_grid), 1)
    Y = repeat(y_grid, 1, length(x_grid))
    Z = map((a,b) -> f([a,b]), X, Y)
    
    plt = surface(x_grid, y_grid, Z, color=:thermal, alpha=0.5, legend=true)
    
    if length(trajectory) > 1
        for i in 1:length(trajectory)-1
            color = trajectory_colors[i]
            plot!(plt, 
                [trajectory[i][1], trajectory[i+1][1]], 
                [trajectory[i][2], trajectory[i+1][2]], 
                [f(trajectory[i]), f(trajectory[i+1])], 
                linecolor=color, 
                linewidth=2,
                label=(i == 1 ? "Релаксационная последовательность" : "")
            )
        end
    end
    
    if length(trajectory) > 0
        scatter3d!(plt, 
            [p[1] for p in trajectory], 
            [p[2] for p in trajectory], 
            [f(p) for p in trajectory], 
            color=:green, 
            markersize=4,
            label="Убывание"
        )
    end
    
    if length(search_points) > 0
        scatter3d!(plt, 
            [p[1] for p in search_points], 
            [p[2] for p in search_points], 
            [f(p) for p in search_points], 
            color=:blue, 
            markersize=3,
            label="Поиск направления"
        )
    end
    
    if length(max_points) > 0
        scatter3d!(plt, 
            [p[1] for p in max_points], 
            [p[2] for p in max_points], 
            [f(p) for p in max_points], 
            color=:red, 
            markersize=5,
            label="Максимум"
        )
    end
    

    scatter3d!(plt, [x0[1]], [x0[2]], [f(x0)], color=:yellow, markersize=8, label="Старт")
    scatter3d!(plt, [result[1]], [result[2]], [f(result)], color=:purple, markersize=8, label="Финиш")
    
    title!(plt, title_str)
    xlabel!(plt, "x")
    ylabel!(plt, "y")
    zlabel!(plt, "f(x,y)")
    
    return plt
end


function plot_optimization_2d(f, x0, result, trajectory, search_points, max_points, trajectory_colors, title_str, x_range, y_range)
    x_grid = range(x_range[1], x_range[2], length=100)
    y_grid = range(y_range[1], y_range[2], length=100)
    X = repeat(reshape(x_grid, 1, :), length(y_grid), 1)
    Y = repeat(y_grid, 1, length(x_grid))
    Z = map((a,b) -> f([a,b]), X, Y)
    
    plt = contour(x_grid, y_grid, Z, 
        color=:thermal, 
        levels=20,
        xlabel="x₁", 
        ylabel="x₂",
        title=title_str,
        legend=:topleft
    )
    
    if length(trajectory) > 1
        for i in 1:length(trajectory)-1
            color = trajectory_colors[i]
            plot!(plt, 
                [trajectory[i][1], trajectory[i+1][1]], 
                [trajectory[i][2], trajectory[i+1][2]], 
                color=color, 
                linewidth=2,
                label=(i == 1 ? "Релаксационная последовательность" : "")
            )
        end
    end
    
    if length(trajectory) > 0
        scatter!(plt, 
            [p[1] for p in trajectory], 
            [p[2] for p in trajectory], 
            color=:green, 
            markersize=4,
            label="Убывание"
        )
    end
    
    if length(search_points) > 0
        scatter!(plt, 
            [p[1] for p in search_points], 
            [p[2] for p in search_points], 
            color=:blue, 
            markersize=3,
            label="Поиск направления"
        )
    end
    
    if length(max_points) > 0
        scatter!(plt, 
            [p[1] for p in max_points], 
            [p[2] for p in max_points], 
            color=:red, 
            markersize=5,
            label="Максимум"
        )
    end
    
    scatter!(plt, [x0[1]], [x0[2]], color=:yellow, markersize=8, label="Старт")
    scatter!(plt, [result[1]], [result[2]], color=:purple, markersize=8, label="Финиш")
    
    plot!(plt, colorbar=true)
    
    return plt
end

function test_function(f, f_name, x0, x_range, y_range, variant="a", show_plots=true)
    println("\n" * "="^60)
    println("Тестирование функции: $f_name")
    println("Начальная точка: $x0")
    println("Вариант: $variant")
    println("="^60)
    
    if variant == "a"
        result, trajectory, search_points, max_points, trajectory_colors = hooke_jeeves_variant_a(f, x0)
    else
        result, trajectory, search_points, max_points, trajectory_colors = hooke_jeeves_variant_b(f, x0)
    end
    
    println("Результат: $result")
    println("Значение функции: $(f(result))")
    println("Количество итераций: $(length(trajectory))")
    println("Количество точек поиска: $(length(search_points))")
    println("Количество точек максимума: $(length(max_points))")
    
    title_3d = "Метод Хука-Дживса (Вариант $variant) - $f_name (3D)"
    title_2d = "Метод Хука-Дживса (Вариант $variant) - $f_name (2D)"
    
    plt_3d = plot_optimization_3d(f, x0, result, trajectory, search_points, max_points, trajectory_colors, title_3d, x_range, y_range)
    plt_2d = plot_optimization_2d(f, x0, result, trajectory, search_points, max_points, trajectory_colors, title_2d, x_range, y_range)
    
    if show_plots
        display(plt_3d)
        display(plt_2d)
    end
    
    return result, trajectory, plt_3d, plt_2d
end

function compare_variants(f, f_name, x0, x_range, y_range)
    println("\n" * "="^60)
    println("СРАВНЕНИЕ ВАРИАНТОВ для функции: $f_name")
    println("Начальная точка: $x0")
    println("="^60)
    
    # Вариант А
    result_a, trajectory_a, search_points_a, max_points_a, _ = hooke_jeeves_variant_a(f, x0)
    
    # Вариант Б
    result_b, trajectory_b, search_points_b, max_points_b, _ = hooke_jeeves_variant_b(f, x0)
    
    println("\nВариант А (поиск по направлению один раз):")
    println("  Количество точек в релаксационной последовательности: $(length(trajectory_a))")
    println("  Количество точек поиска: $(length(search_points_a))")
    println("  Значение функции: $(f(result_a))")
    
    println("\nВариант Б (поиск по направлению несколько раз с проверкой убывания):")
    println("  Количество точек в релаксационной последовательности: $(length(trajectory_b))")
    println("  Количество точек поиска: $(length(search_points_b))")
    println("  Значение функции: $(f(result_b))")
    
    return nothing
end

println("Метод Хука-Дживса: Тестирование на трёх функциях")
println("\nОписание вариантов:")
println("Вариант А: Поиск по направлению выполняется один раз")
println("Вариант Б: Поиск по направлению выполняется несколько раз с проверкой убывания функции")
# Тест 1: Розенброк
x0_rosenbrock = [-2.0, -2.0]
result_a1, traj_a1, plt_3d_a1, plt_2d_a1 = test_function(rosenbrock, "Розенброк", x0_rosenbrock, (-2.0, 2.0), (-2.0, 2.0), "a")
result_b1, traj_b1, plt_3d_b1, plt_2d_b1 = test_function(rosenbrock, "Розенброк", x0_rosenbrock, (-2.0, 2.0), (-2.0, 2.0), "b")

compare_variants(rosenbrock, "Розенброк", x0_rosenbrock, (-2.0, 2.0), (-2.0, 2.0));

# Тест 2: Растригин
x0_rastrigin = [0.5, 0.5]
result_a2, traj_a2, plt_3d_a2, plt_2d_a2 = test_function(rastrigin, "Растригин", x0_rastrigin, (-5.0, 5.0), (-5.0, 5.0), "a")
result_b2, traj_b2, plt_3d_b2, plt_2d_b2 = test_function(rastrigin, "Растригин", x0_rastrigin, (-5.0, 5.0), (-5.0, 5.0), "b")

compare_variants(rastrigin, "Растригин", x0_rastrigin, (-5.0, 5.0), (-5.0, 5.0));

# Тест 3: Швефель
x0_schwefel = [0.5, 0.5]
result_a3, traj_a3, plt_3d_a3, plt_2d_a3 = test_function(schwefel, "Швефель", x0_schwefel, (-10.0, 10.0), (-10.0, 10.0), "a")
result_b3, traj_b3, plt_3d_b3, plt_2d_b3 = test_function(schwefel, "Швефель", x0_schwefel, (-10.0, 10.0), (-10.0, 10.0), "b")

compare_variants(schwefel, "Швефель", x0_schwefel, (-10.0, 10.0), (-10.0, 10.0));
```
