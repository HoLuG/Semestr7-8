```julia
using Plots
using LinearAlgebra
using Colors

gr()

# 3D функции (от 3 переменных)
function rosenbrock_3d(x)
    return (1.0 - x[1])^2 + 100.0*(x[2] - x[1]^2)^2 + (1.0 - x[2])^2 + 100.0*(x[3] - x[2]^2)^2
end

function rastrigin_3d(x)
    n = length(x)
    return 20 + sum(x[i]^2 - 10*cos(2*pi*x[i]) for i in 1:n)
end

function schwefel_3d(x)
    n = length(x)
    return 418.9829*n - sum(x[i]*sin(sqrt(abs(x[i]))) for i in 1:n)
end

# Вариант А:
function hooke_jeeves_variant_a(f, x0, eps=1e-6, delta0=0.1)
    n = length(x0)
    delta = fill(delta0, n)
    x = copy(x0)
    
    trajectory = [copy(x)]  # Релаксационная последовательность
    search_points = [copy(x)]  # Точки поиска направления
    max_points = []  # Красные точки - где функция не убывает
    trajectory_colors = [:green]  # Цвета точек траектории
    
    iteration = 0
    max_iterations = 10000
    
    while maximum(delta) > eps && iteration < max_iterations
        iteration += 1
        x_prev = copy(x)
        f_prev = f(x_prev)
        
        # Исследовательский поиск - смотрим во все стороны (по 1 клетке)
        x_new = copy(x)
        improved = false
        
        for i in 1:n
            # +delta
            x_test = copy(x_new)
            x_test[i] += delta[i]
            f_test = f(x_test)
            push!(search_points, copy(x_test))  # Отмечаем все проверенные точки
            
            if f_test < f(x_new)
                x_new = x_test
                improved = true
            else
                # -delta
                x_test = copy(x_new)
                x_test[i] -= delta[i]
                f_test = f(x_test)
                push!(search_points, copy(x_test))  # Отмечаем все проверенные точки
                
                if f_test < f(x_new)
                    x_new = x_test
                    improved = true
                end
            end
        end
        
        if f(x_new) < f_prev
            x = copy(x_new)
            push!(trajectory, copy(x))
            push!(trajectory_colors, :green)
        else

            push!(max_points, copy(x))
            delta = delta / 2.0
            
            if maximum(delta) < eps || abs(f(x) - f_prev) < eps
                break
            end
        end
    end
    
    return x, trajectory, search_points, max_points, trajectory_colors
end

function exploratory_search(f, x, delta, include_start=true)
    n = length(x)
    x_new = copy(x)
    direction_points = include_start ? [copy(x)] : []  # Начальную точку добавляем только если нужно

    for i in 1:n
        start_val = f(x_new)
        temp = x_new[i]
        
        # Пробуем +delta
        x_test_plus = copy(x_new)
        x_test_plus[i] = temp + delta[i]
        f_plus = f(x_test_plus)
        push!(direction_points, copy(x_test_plus))  # Добавляем проверенную точку
        
        # Пробуем -delta
        x_test_minus = copy(x_new)
        x_test_minus[i] = temp - delta[i]
        f_minus = f(x_test_minus)
        push!(direction_points, copy(x_test_minus))  # Добавляем проверенную точку

        # Выбираем лучшую точку
        if f_plus < start_val
            x_new = x_test_plus
        elseif f_minus < start_val
            x_new = x_test_minus
        end
        # Если обе хуже, остаёмся на месте (x_new не меняется)
    end

    return x_new, direction_points
end

# Вариант Б:
function hooke_jeeves_variant_b(f, x0, eps=1e-6, delta0=0.1)
    n = length(x0)
    delta = fill(delta0, n)
    x = copy(x0)
    
    trajectory = [copy(x)]  # Релаксационная последовательность
    search_points = [copy(x)]  # Точки поиска направления
    max_points = []  # Красные точки - где функция не убывает
    trajectory_colors = [:green]
    
    iteration = 0
    max_iterations = 10000
    
    while maximum(delta) > eps && iteration < max_iterations
        iteration += 1
        x_prev = copy(x)
        f_prev = f(x_prev)
        
        # Исследовательский поиск - находим направление ОДИН РАЗ
        # Не добавляем начальную точку, так как она уже есть в search_points
        x_exp, direction_search_points = exploratory_search(f, x, delta, false)
        append!(search_points, direction_search_points)
        
        if f(x_exp) >= f(x)
            push!(max_points, copy(x))
            delta = delta / 2.0
            continue
        end
        
        direction = x_exp - x
        step_size = norm(direction)
        
        if step_size > eps
            direction_normalized = direction / step_size
            
            x_current = copy(x_exp)
            f_current = f(x_current)
            improved = true
            
            while improved
                x_next = x_current + direction_normalized * step_size
                f_next = f(x_next)
                
                if f_next < f_current
                    x_current = x_next
                    f_current = f_next
                    push!(trajectory, copy(x_current))
                    push!(trajectory_colors, :green)
                else
                    improved = false
                end
            end
            
            x_rollback = x_current - direction_normalized * step_size / 2.0
            f_rollback = f(x_rollback)
            
            if f_rollback < f_current
                x = copy(x_rollback)
            else
                x = copy(x_current)
            end
            
            push!(trajectory, copy(x))
            push!(trajectory_colors, :green)
        else
            x = copy(x_exp)
            push!(trajectory, copy(x))
            push!(trajectory_colors, :green)
        end
        
        if norm(x - x_prev) < eps || abs(f(x) - f_prev) < eps
            break
        end
    end
    
    return x, trajectory, search_points, max_points, trajectory_colors
end

# Функция для создания 3D визуализации траекторий двух вариантов
function plot_trajectories_3d(f, x0, result_a, trajectory_a, search_points_a, 
                              result_b, trajectory_b, search_points_b, 
                              title_str, show_search_points=false)
    # Функция для получения цвета от светлого к тёмному
    function get_color_gradient(step, total_steps, base_color)
        # step от 0 до total_steps-1
        # 0 -> светлый (начало), total_steps-1 -> тёмный (конец/экстремум)
        if total_steps <= 1
            intensity = 0.0
        else
            intensity = step / (total_steps - 1)  # от 0 до 1
        end
        # Для синего: от светлого (0.6, 0.8, 1.0) к тёмному (0.0, 0.0, 0.5)
        # Для красного: от светлого (1.0, 0.7, 0.7) к тёмному (0.5, 0.0, 0.0)
        if base_color == :blue
            r = 0.6 * (1 - intensity)
            g = 0.8 * (1 - intensity)
            b = 1.0 - 0.5 * intensity
        else  # red
            r = 1.0 - 0.5 * intensity
            g = 0.7 * (1 - intensity)
            b = 0.7 * (1 - intensity)
        end
        return RGB(r, g, b)
    end
    
    # Создаём пустой 3D график
    plt = plot([], [], [],
        seriestype=:path3d,
        xlabel="x₁", 
        ylabel="x₂", 
        zlabel="x₃",
        title=title_str,
        legend=:topright,
        size=(800, 600)
    )
    
    # Рисуем траекторию варианта А (синий градиент)
    if length(trajectory_a) > 1
        for i in 1:length(trajectory_a)-1
            color = get_color_gradient(i-1, length(trajectory_a)-1, :blue)
            plot!(plt, 
                [trajectory_a[i][1], trajectory_a[i+1][1]], 
                [trajectory_a[i][2], trajectory_a[i+1][2]], 
                [trajectory_a[i][3], trajectory_a[i+1][3]], 
                linecolor=color, 
                linewidth=2.5,
                seriestype=:path3d,
                label=(i == 1 ? "Вариант А" : "")
            )
        end
    end
    
    # Рисуем траекторию варианта Б (красный градиент)
    if length(trajectory_b) > 1
        for i in 1:length(trajectory_b)-1
            color = get_color_gradient(i-1, length(trajectory_b)-1, :red)
            plot!(plt, 
                [trajectory_b[i][1], trajectory_b[i+1][1]], 
                [trajectory_b[i][2], trajectory_b[i+1][2]], 
                [trajectory_b[i][3], trajectory_b[i+1][3]], 
                linecolor=color, 
                linewidth=2.5,
                seriestype=:path3d,
                label=(i == 1 ? "Вариант Б" : "")
            )
        end
    end
    
    # Показываем точки поиска, если включен параметр
    if show_search_points
        if length(search_points_a) > 0
            scatter!(plt, 
                [p[1] for p in search_points_a], 
                [p[2] for p in search_points_a], 
                [p[3] for p in search_points_a], 
                color=:lightblue, 
                markersize=2,
                alpha=0.3,
                seriestype=:scatter3d,
                label="Поиск А"
            )
        end
        
        if length(search_points_b) > 0
            scatter!(plt, 
                [p[1] for p in search_points_b], 
                [p[2] for p in search_points_b], 
                [p[3] for p in search_points_b], 
                color=:lightcoral, 
                markersize=2,
                alpha=0.3,
                seriestype=:scatter3d,
                label="Поиск Б"
            )
        end
    end
    
    # Стартовая точка
    scatter!(plt, [x0[1]], [x0[2]], [x0[3]], 
        color=:yellow, 
        markersize=10, 
        seriestype=:scatter3d,
        label="Старт"
    )
    
    # Финишные точки
    scatter!(plt, [result_a[1]], [result_a[2]], [result_a[3]], 
        color=:darkblue, 
        markersize=10, 
        seriestype=:scatter3d,
        label="Финиш А"
    )
    scatter!(plt, [result_b[1]], [result_b[2]], [result_b[3]], 
        color=:darkred, 
        markersize=10, 
        seriestype=:scatter3d,
        label="Финиш Б"
    )
    
    return plt
end

function compare_variants_3d(f, f_name, x0, show_search_points=false, show_plot=true)
    println("\n" * "="^60)
    println("СРАВНЕНИЕ ВАРИАНТОВ для функции: $f_name (3D)")
    println("Начальная точка: $x0")
    println("="^60)
    
    # Вариант А
    result_a, trajectory_a, search_points_a, max_points_a, _ = hooke_jeeves_variant_a(f, x0)
    
    # Вариант Б
    result_b, trajectory_b, search_points_b, max_points_b, _ = hooke_jeeves_variant_b(f, x0)
    
    println("\nВариант А (поиск по направлению один раз):")
    println("  Количество точек в релаксационной последовательности: $(length(trajectory_a))")
    println("  Количество точек поиска: $(length(search_points_a))")
    println("  Значение функции: $(f(result_a))")
    println("  Результат: $result_a")
    
    println("\nВариант Б (поиск по направлению несколько раз с проверкой убывания):")
    println("  Количество точек в релаксационной последовательности: $(length(trajectory_b))")
    println("  Количество точек поиска: $(length(search_points_b))")
    println("  Значение функции: $(f(result_b))")
    println("  Результат: $result_b")
    
    title_str = "Метод Хука-Дживса - $f_name (3D)"
    plt = plot_trajectories_3d(f, x0, result_a, trajectory_a, search_points_a,
                               result_b, trajectory_b, search_points_b,
                               title_str, show_search_points)
    
    if show_plot
        display(plt)
    end
    
    return result_a, trajectory_a, result_b, trajectory_b, plt
end

println("Метод Хука-Дживса: Тестирование на трёх функциях (3D)")
println("\nОписание вариантов:")
println("Вариант А: Поиск по направлению выполняется один раз")
println("Вариант Б: Поиск по направлению выполняется несколько раз с проверкой убывания функции")
println("\nВизуализация:")
println("- Траектория варианта А: синий градиент (от светлого к тёмному)")
println("- Траектория варианта Б: красный градиент (от светлого к тёмному)")
println("- Параметр show_search_points=true показывает точки поиска")

# Тест 1: Розенброк (3D)
x0_rosenbrock_3d = [-2.0, -2.0, -2.0]
println("\n" * "="^80)
println("ТЕСТ 1: Розенброк (3D)")
println("="^80)
result_a1, traj_a1, result_b1, traj_b1, plt1 = compare_variants_3d(rosenbrock_3d, "Розенброк", x0_rosenbrock_3d, false)

# Тест 2: Растригин (3D)
x0_rastrigin_3d = [0.5, 0.5, 0.5]
println("\n" * "="^80)
println("ТЕСТ 2: Растригин (3D)")
println("="^80)
result_a2, traj_a2, result_b2, traj_b2, plt2 = compare_variants_3d(rastrigin_3d, "Растригин", x0_rastrigin_3d, false)

# Тест 3: Швефель (3D)
x0_schwefel_3d = [0.5, 0.5, 0.5]
println("\n" * "="^80)
println("ТЕСТ 3: Швефель (3D)")
println("="^80)
result_a3, traj_a3, result_b3, traj_b3, plt3 = compare_variants_3d(schwefel_3d, "Швефель", x0_schwefel_3d, false)

# Пример с включёнными точками поиска для Розенброка
println("\n" * "="^80)
println("ПРИМЕР: Розенброк (3D) с точками поиска")
println("="^80)
_, _, _, _, plt1_with_search = compare_variants_3d(rosenbrock_3d, "Розенброк (с точками поиска)", x0_rosenbrock_3d, true)
```
