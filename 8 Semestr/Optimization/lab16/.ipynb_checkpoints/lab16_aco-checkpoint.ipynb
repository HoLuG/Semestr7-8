```julia
using Random
using Statistics
using LinearAlgebra

using Plots

Random.seed!(42);
```

```julia
function tour_length(tour::AbstractVector{<:Integer}, D::AbstractMatrix{<:Real})
    n = length(tour)
    s = 0.0
    @inbounds for k in 1:(n-1)
        s += D[tour[k], tour[k+1]]
    end
    s += D[tour[end], tour[1]]
    return s
end

function roulette_index(weights::AbstractVector{<:Real}, rng::AbstractRNG=Random.default_rng())
    total = 0.0
    @inbounds for w in weights
        total += w
    end
    if !(total > 0)
        return rand(rng, 1:length(weights))
    end
    r = rand(rng) * total
    acc = 0.0
    @inbounds for i in eachindex(weights)
        acc += weights[i]
        if acc >= r
            return i
        end
    end
    return lastindex(weights)
end

function tsplib_euc_2d(p::NTuple{2,Float64}, q::NTuple{2,Float64})
    dx = p[1] - q[1]
    dy = p[2] - q[2]
    return round(Int, sqrt(dx*dx + dy*dy))
end

function distance_matrix(coords::Vector{NTuple{2,Float64}}; metric=:tsplib)
    n = length(coords)
    D = zeros(Float64, n, n)
    @inbounds for i in 1:n
        for j in i+1:n
            d = metric === :tsplib ? tsplib_euc_2d(coords[i], coords[j]) : hypot(coords[i][1]-coords[j][1], coords[i][2]-coords[j][2])
            D[i,j] = d
            D[j,i] = d
        end
    end
    return D
end
```

```julia
using Base64

MAKE_GIFS = true
GIF_DIR = "gifs"
mkpath(GIF_DIR)

safe_filename(s::AbstractString) = replace(lowercase(s), r"[^a-z0-9]+" => "_")

function ant_system_tsp_best_trace(
    D::AbstractMatrix{<:Real};
    iters::Int=200,
    N::Int=size(D,1),
    α::Float64=1.0,
    β::Float64=5.0,
    ρ::Float64=0.9,
    Q::Float64=20.0,
    τ0::Float64=1e-6,
    elitist::Int=2,
    directed_pheromone::Bool=false,
    seed::Int=1,
    save_frames::Bool=false,
    frame_stride::Int=5
)
    rng = MersenneTwister(seed)
    n = size(D, 1)

    τ = fill(τ0, n, n)
    η = zeros(Float64, n, n)
    @inbounds for i in 1:n
        for j in 1:n
            if i != j
                dij = float(D[i,j])
                η[i,j] = dij > 0 ? 1.0 / dij : 0.0
            end
        end
    end

    best_tour = collect(1:n)
    best_len = Inf
    best_history = Vector{Float64}(undef, iters)

    saved_tours = Vector{Vector{Int}}()
    saved_lens  = Float64[]
    saved_iters = Int[]

    allowed = Vector{Int}(undef, n-1)
    weights = Vector{Float64}(undef, n-1)

    for t in 1:iters
        tours = Vector{Vector{Int}}(undef, N)
        lens  = Vector{Float64}(undef, N)

        @inbounds for k in 1:N
            start = rand(rng, 1:n)
            tour = Vector{Int}(undef, n)
            tour[1] = start

            visited = falses(n)
            visited[start] = true
            cur = start

            for step in 2:n
                m = 0
                for j in 1:n
                    if !visited[j]
                        m += 1
                        allowed[m] = j
                    end
                end
                for idx in 1:m
                    j = allowed[idx]
                    weights[idx] = (τ[cur, j]^α) * (η[cur, j]^β)
                end
                chosen_idx = roulette_index(view(weights, 1:m), rng)
                next_city = allowed[chosen_idx]

                tour[step] = next_city
                visited[next_city] = true
                cur = next_city
            end

            L = tour_length(tour, D)
            tours[k] = tour
            lens[k] = L

            if L < best_len
                best_len = L
                best_tour = copy(tour)
            end
        end

        best_history[t] = best_len

        if save_frames && (t == 1 || t == iters || (t % frame_stride == 0))
            push!(saved_tours, copy(best_tour))
            push!(saved_lens, best_len)
            push!(saved_iters, t)
        end

        τ .*= (1.0 - ρ)

        @inbounds for k in 1:N
            Δ = Q / lens[k]
            tour = tours[k]
            for s in 1:(n-1)
                i = tour[s]
                j = tour[s+1]
                τ[i,j] += Δ
                if !directed_pheromone
                    τ[j,i] += Δ
                end
            end
            i = tour[end]
            j = tour[1]
            τ[i,j] += Δ
            if !directed_pheromone
                τ[j,i] += Δ
            end
        end

        if elitist > 0
            kbest = argmin(lens)
            tour = tours[kbest]
            Δe = elitist * (Q / lens[kbest])
            @inbounds for s in 1:(n-1)
                i = tour[s]
                j = tour[s+1]
                τ[i,j] += Δe
                if !directed_pheromone
                    τ[j,i] += Δe
                end
            end
            i = tour[end]
            j = tour[1]
            τ[i,j] += Δe
            if !directed_pheromone
                τ[j,i] += Δe
            end
        end
    end

    return (
        best_tour=best_tour,
        best_len=best_len,
        best_history=best_history,
        saved_iters=saved_iters,
        saved_tours=saved_tours,
        saved_lens=saved_lens
    )
end

function plot_tour_gif(coords::Vector{NTuple{2,Float64}}, tour::AbstractVector{<:Integer}; ttl::AbstractString="")
    xs = [coords[i][1] for i in tour]
    ys = [coords[i][2] for i in tour]
    push!(xs, coords[tour[1]][1])
    push!(ys, coords[tour[1]][2])

    plt = plot(xs, ys, seriestype=:path, color=:gray30, lw=2, legend=false, title=ttl)
    scatter!(plt, [c[1] for c in coords], [c[2] for c in coords], ms=2.8, color=:orange, markerstrokewidth=0.4)
    return plt
end

function show_gif_inline(path::AbstractString)
    data = base64encode(read(path))
    html = "<img src=\"data:image/gif;base64,$data\" />"
    return display("text/html", html)
end

function make_tour_gif(name::AbstractString, coords::Vector{NTuple{2,Float64}}, D::AbstractMatrix{<:Real};
    iters::Int=200,
    N::Int=size(D,1),
    α::Float64=1.0,
    β::Float64=5.0,
    ρ::Float64=0.9,
    Q::Float64=20.0,
    τ0::Float64=1e-6,
    elitist::Int=2,
    directed_pheromone::Bool=false,
    seed::Int=1,
    frame_stride::Int=5,
    fps::Int=8
)
    res = ant_system_tsp_best_trace(D; iters, N, α, β, ρ, Q, τ0, elitist, directed_pheromone, seed, save_frames=true, frame_stride)

    anim = @animate for idx in eachindex(res.saved_tours)
        t = res.saved_iters[idx]
        L = res.saved_lens[idx]
        plot_tour_gif(coords, res.saved_tours[idx]; ttl="$(name): итерация $(t), L=$(round(L, digits=2))")
    end

    out = joinpath(GIF_DIR, safe_filename(name) * ".gif")
    gif(anim, out, fps=fps)
    println("GIF сохранён: ", out)
    show_gif_inline(out)
    return out
end

function make_convergence_gif(name::AbstractString, hist::AbstractVector{<:Real}; fps::Int=12)
    anim = @animate for t in 1:length(hist)
        plot(hist[1:t], lw=2, legend=false, xlabel="Итерация", ylabel="Лучшая длина", title="$(name): t=$(t)")
    end
    out = joinpath(GIF_DIR, safe_filename(name) * "_conv.gif")
    gif(anim, out, fps=fps)
    println("GIF сохранён: ", out)
    show_gif_inline(out)
    return out
end
```

```julia
n_small = 10
coords_small = [(100.0*rand(), 100.0*rand()) for _ in 1:n_small]
D_small = distance_matrix(coords_small; metric=:euclid)

res_small = ant_system_tsp_best_trace(
    D_small;
    iters=100,
    N=n_small,
    α=1.0,
    β=5.0,
    ρ=0.9,
    Q=20.0,
    τ0=1e-6,
    elitist=2,
    seed=1,
    save_frames=false
)

best_tour_s = res_small.best_tour
best_len_s = res_small.best_len
hist_s = res_small.best_history

println("Лучшая длина (small) = ", round(best_len_s, digits=3))
println("Лучший тур (small) = ", best_tour_s)

plot(hist_s, lw=2, xlabel="Итерация", ylabel="Лучшая длина", title="Сходимость ACO (малый граф)", legend=false)
```

```julia
function plot_tour(coords::Vector{NTuple{2,Float64}}, tour::AbstractVector{<:Integer}; ttl="Маршрут")
    xs = [coords[i][1] for i in tour]
    ys = [coords[i][2] for i in tour]

    push!(xs, coords[tour[1]][1])
    push!(ys, coords[tour[1]][2])

    plt = plot(xs, ys, seriestype=:path, lw=2, legend=false, title=ttl, xlabel="x", ylabel="y")
    scatter!(plt, [c[1] for c in coords], [c[2] for c in coords], ms=4)
    return plt
end

plot_tour(coords_small, best_tour_s; ttl="Лучший тур (малый граф)")
```

```julia
function run_aco_experiment(name::AbstractString, D::AbstractMatrix{<:Real};
    coords::Union{Nothing,Vector{NTuple{2,Float64}}}=nothing,
    iters::Int=200,
    N::Union{Int,Nothing}=nothing,
    α::Float64=1.0,
    β::Float64=5.0,
    ρ::Float64=0.9,
    Q::Float64=20.0,
    τ0::Float64=1e-6,
    elitist::Int=2,
    seed::Int=33
)
    N = isnothing(N) ? size(D,1) : N

    res = ant_system_tsp_best_trace(
        D;
        iters=iters,
        N=N,
        α=α,
        β=β,
        ρ=ρ,
        Q=Q,
        τ0=τ0,
        elitist=elitist,
        seed=seed,
        save_frames=false
    )

    best_tour = res.best_tour
    best_len  = res.best_len
    hist      = res.best_history

    println("=== ", name, " ===")
    println("Лучшая длина = ", best_len)
    println("Лучший тур = ", best_tour)

    p_conv = plot(hist, lw=2, xlabel="Итерация", ylabel="Лучшая длина", title="Сходимость ACO — $(name)", legend=false)

    if coords === nothing
        return p_conv
    else
        p_tour = plot_tour(coords, best_tour; ttl="Лучший тур — $(name), L=$(best_len)")
        return plot(p_conv, p_tour, layout=(1,2), size=(1100,420))
    end
end
```

```julia
function ring_plus_center_coords(n_ring::Int=30, n_center::Int=12; R::Float64=100.0, center_sigma::Float64=12.0, seed::Int=10)
    rng = MersenneTwister(seed)
    coords = NTuple{2,Float64}[]

    for k in 0:(n_ring-1)
        θ = 2π * k / n_ring
        x = R * cos(θ)
        y = R * sin(θ)
        push!(coords, (x,y))
    end

    for _ in 1:n_center
        x = center_sigma * randn(rng)
        y = center_sigma * randn(rng)
        push!(coords, (x,y))
    end

    return coords
end

coords_ring = ring_plus_center_coords(32, 14; R=120.0, center_sigma=15.0, seed=11)
D_ring = distance_matrix(coords_ring; metric=:euclid)
run_aco_experiment("Кольцо + центр", D_ring; coords=coords_ring, iters=250, N=length(coords_ring), seed=12)
```

## Необычные примеры: “ловушка локальных рёбер” и “кластеры + мосты”

Ниже — два специально сконструированных примера (не из TSPLIB), которые хорошо демонстрируют поведение ACO:

- **Кластеры + мосты**: есть два кластера, а переходы между ними почти всегда очень дорогие, кроме **двух “мостов”**. Оптимальный тур обязан использовать эти мосты.
- **Ловушка локальных рёбер**: внутри небольшой группы вершин очень дешёвые рёбра (муравьи быстро их усиливают), но выгодные “выходы наружу” есть только из 1–2 вершин. На ранних итерациях алгоритм часто ошибается и платит большой штраф, пока не “выучит” правильные точки выхода.

```julia
function clusters_with_bridges_distance(nA::Int=18, nB::Int=18; seed::Int=100,
    within_min::Float64=1.0, within_max::Float64=10.0,
    between_min::Float64=120.0, between_max::Float64=200.0,
    bridge_min::Float64=15.0, bridge_max::Float64=30.0
)
    rng = MersenneTwister(seed)
    n = nA + nB
    D = zeros(Float64, n, n)

    A = collect(1:nA)
    B = collect((nA+1):n)

    # внутри
    for i in 1:n
        for j in (i+1):n
            same = (i <= nA && j <= nA) || (i > nA && j > nA)
            if same
                d = within_min + (within_max-within_min)*rand(rng)
            else
                d = between_min + (between_max-between_min)*rand(rng)
            end
            D[i,j] = d
            D[j,i] = d
        end
    end

    # выбираем 2 моста
    a0, a1 = rand(rng, A, 2)
    b0, b1 = rand(rng, B, 2)
    bridges = [(a0,b0), (a1,b1)]

    for (i,j) in bridges
        d = bridge_min + (bridge_max-bridge_min)*rand(rng)
        D[i,j] = d
        D[j,i] = d
    end

    return D, (A=A, B=B), bridges
end

function local_edges_trap_distance(n_out::Int=18, n_trap::Int=6; seed::Int=200,
    out_min::Float64=5.0, out_max::Float64=20.0,
    trap_min::Float64=1.0, trap_max::Float64=3.0,
    trap_to_out_bad_min::Float64=80.0, trap_to_out_bad_max::Float64=140.0,
    good_exit_min::Float64=15.0, good_exit_max::Float64=30.0
)
    rng = MersenneTwister(seed)
    n = n_out + n_trap
    D = zeros(Float64, n, n)

    OUT = collect(1:n_out)
    TRAP = collect((n_out+1):n)

    # внутри OUT
    for i in OUT
        for j in (i+1):n
            if j in OUT
                d = out_min + (out_max-out_min)*rand(rng)
                D[i,j] = d
                D[j,i] = d
            end
        end
    end

    # внутри TRAP
    for i in TRAP
        for j in (i+1):n
            if j in TRAP
                d = trap_min + (trap_max-trap_min)*rand(rng)
                D[i,j] = d
                D[j,i] = d
            end
        end
    end

    # TRAP <-> OUT (плохие)
    for i in TRAP
        for j in OUT
            d = trap_to_out_bad_min + (trap_to_out_bad_max-trap_to_out_bad_min)*rand(rng)
            D[i,j] = d
            D[j,i] = d
        end
    end

    # 2 хорошие "двери"
    t_out1, t_out2 = rand(rng, TRAP, 2)
    o1, o2 = rand(rng, OUT, 2)

    good_edges = [(t_out1, o1), (t_out2, o2)]
    for (i,j) in good_edges
        d = good_exit_min + (good_exit_max-good_exit_min)*rand(rng)
        D[i,j] = d
        D[j,i] = d
    end

    info = (OUT=OUT, TRAP=TRAP, good_edges=good_edges)
    return D, info
end
```

```julia
# Тест: Кластеры + мосты
D_bridge, groups_bridge, bridges = clusters_with_bridges_distance(18, 18; seed=101)
println("Мосты (индексы вершин): ", bridges)
run_aco_experiment("Кластеры + мосты (взвешенный граф)", D_bridge; coords=nothing, iters=350, N=size(D_bridge,1), seed=40)
```

```julia
# Тест: Ловушка локальных рёбер
D_trap, info_trap = local_edges_trap_distance(18, 6; seed=202)
println("Хорошие выходы (trap->out): ", info_trap.good_edges)
run_aco_experiment("Ловушка локальных рёбер (взвешенный граф)", D_trap; coords=nothing, iters=350, N=size(D_trap,1), seed=41)
```
