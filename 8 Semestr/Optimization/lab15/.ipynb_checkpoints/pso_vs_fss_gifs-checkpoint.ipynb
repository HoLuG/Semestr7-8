```julia
using Random, Statistics
using Plots

gr()
```

```julia
schwefel(x) = 418.9828872724338 * length(x) - sum(xi * sin(sqrt(abs(xi))) for xi in x)
rosenbrock(x) = (1.0 - x[1])^2 + 100.0 * (x[2] - x[1]^2)^2
rastrigin(x) = 10.0 * length(x) + sum(xi^2 - 10.0 * cos(2pi * xi) for xi in x)

BENCHMARKS = Dict(
    "Schwefel" => (f = schwefel, bounds = (-500.0, 500.0), optimum = [420.968746, 420.968746]),
    "Rosenbrock" => (f = rosenbrock, bounds = (-2.0, 2.0), optimum = [1.0, 1.0]),
    "Rastrigin" => (f = rastrigin, bounds = (-5.12, 5.12), optimum = [0.0, 0.0])
);
```

```julia
mutable struct PSOResult
    gbest_position::Vector{Float64}
    gbest_value::Float64
    gbest_history::Vector{Float64}
    pos_history::Vector{Matrix{Float64}}  # 2 x n
end

function pso(
    f;
    bounds::Tuple{Float64, Float64},
    n_particles::Int = 50,
    n_iter::Int = 250,
    w::Float64 = 0.72,
    c1::Float64 = 1.49,
    c2::Float64 = 1.49,
    vmax_ratio::Float64 = 0.20,
    seed::Int = 42
)
    Random.seed!(seed)
    dim = 2
    lo, hi = bounds
    span = hi - lo
    vmax = vmax_ratio * span

    X = lo .+ span .* rand(dim, n_particles)
    V = (2rand(dim, n_particles) .- 1.0) .* vmax

    P = copy(X)
    pbest = [f(X[:, i]) for i in 1:n_particles]
    best_id = argmin(pbest)
    G = copy(P[:, best_id])
    gbest = pbest[best_id]

    g_hist = Float64[gbest]
    pos_hist = Matrix{Float64}[copy(X)]

    for _ in 1:n_iter
        for i in 1:n_particles
            r1 = rand(dim)
            r2 = rand(dim)
            V[:, i] .= w .* V[:, i] .+ c1 .* r1 .* (P[:, i] .- X[:, i]) .+ c2 .* r2 .* (G .- X[:, i])
            V[:, i] .= clamp.(V[:, i], -vmax, vmax)

            X[:, i] .+= V[:, i]
            X[:, i] .= clamp.(X[:, i], lo, hi)

            cur = f(X[:, i])
            if cur < pbest[i]
                pbest[i] = cur
                P[:, i] .= X[:, i]
            end
        end

        best_id = argmin(pbest)
        if pbest[best_id] < gbest
            gbest = pbest[best_id]
            G .= P[:, best_id]
        end

        push!(g_hist, gbest)
        push!(pos_hist, copy(X))
    end

    return PSOResult(copy(G), gbest, g_hist, pos_hist)
end
```

```julia
mutable struct FSSResult
    best_position::Vector{Float64}
    best_value::Float64
    best_history::Vector{Float64}
    pos_history::Vector{Matrix{Float64}}  # 2 x n
end

function fss(
    f;
    bounds::Tuple{Float64, Float64},
    n_fish::Int = 50,
    n_iter::Int = 250,
    step_ind::Float64 = 0.08,
    step_vol::Float64 = 0.008,
    w_min::Float64 = 1.0,
    w_max::Float64 = 3.0,
    w_scale::Float64 = 1.0,
    seed::Int = 42
)
    Random.seed!(seed)
    dim = 2
    lo, hi = bounds
    span = hi - lo

    X = lo .+ span .* rand(dim, n_fish)
    F = [f(X[:, i]) for i in 1:n_fish]
    W = fill(w_min, n_fish)

    best_id = argmin(F)
    best_pos = copy(X[:, best_id])
    best_val = F[best_id]

    best_hist = Float64[best_val]
    pos_hist = Matrix{Float64}[copy(X)]

    for t in 1:n_iter
        α = 1.0 - (t - 1) / max(n_iter - 1, 1)
        si = step_ind * α * span
        sv = step_vol * α * span

        ΔX = zeros(dim, n_fish)
        ΔF = zeros(n_fish)

        for i in 1:n_fish
            trial = X[:, i] .+ (2rand(dim) .- 1.0) .* si
            trial .= clamp.(trial, lo, hi)
            ftrial = f(trial)
            if ftrial < F[i]
                ΔX[:, i] .= trial .- X[:, i]
                ΔF[i] = F[i] - ftrial
                X[:, i] .= trial
                F[i] = ftrial
            end
        end

        maxΔF = maximum(ΔF)
        if maxΔF > 0
            for i in 1:n_fish
                W[i] = clamp(W[i] + w_scale * (ΔF[i] / maxΔF), w_min, w_max)
            end
        end

        denom = sum(ΔF)
        if denom > 0
            I = sum(ΔX[:, i] .* ΔF[i] for i in 1:n_fish) ./ denom
            for i in 1:n_fish
                X[:, i] .+= I
                X[:, i] .= clamp.(X[:, i], lo, hi)
                F[i] = f(X[:, i])
            end
        end

        B = sum(X[:, i] .* W[i] for i in 1:n_fish) ./ sum(W)
        direction = denom > 0 ? -1.0 : 1.0
        for i in 1:n_fish
            u = rand(dim)
            X[:, i] .+= direction .* sv .* u .* (X[:, i] .- B)
            X[:, i] .= clamp.(X[:, i], lo, hi)
            F[i] = f(X[:, i])
        end

        best_id = argmin(F)
        if F[best_id] < best_val
            best_val = F[best_id]
            best_pos .= X[:, best_id]
        end

        push!(best_hist, best_val)
        push!(pos_hist, copy(X))
    end

    return FSSResult(copy(best_pos), best_val, best_hist, pos_hist)
end
```

```julia
function landscape_grid(f, bounds; n::Int = 180)
    lo, hi = bounds
    xs = collect(range(lo, hi; length = n))
    ys = collect(range(lo, hi; length = n))
    Z = [f([x, y]) for y in ys, x in xs]
    return xs, ys, Z
end

function make_pso_vs_fss_gif(problem_name; n_agents = 50, n_iter = 200, seed = 7, every = 2, fps = 12, grid_n = 160)
    problem = BENCHMARKS[problem_name]
    f = problem.f
    bounds = problem.bounds
    opt = problem.optimum

    pso_res = pso(f; bounds = bounds, n_particles = n_agents, n_iter = n_iter, seed = seed)
    fss_res = fss(f; bounds = bounds, n_fish = n_agents, n_iter = n_iter, seed = seed)

    xs, ys, Z = landscape_grid(f, bounds; n = grid_n)
    idxs = collect(1:every:length(pso_res.pos_history))
    if idxs[end] != length(pso_res.pos_history)
        push!(idxs, length(pso_res.pos_history))
    end

    anim = @animate for k in idxs
        Xp = pso_res.pos_history[k]
        Xf = fss_res.pos_history[k]

        p1 = contour(xs, ys, Z; levels = 28, c = :viridis, fill = false,
                    title = "PSO (it=$(k-1))", xlabel = "x1", ylabel = "x2", legend = false)
        scatter!(p1, Xp[1, :], Xp[2, :]; ms = 4.8, c = :tomato, alpha = 0.95,
                markerstrokecolor = :black, markerstrokewidth = 0.4)
        scatter!(p1, [pso_res.gbest_position[1]], [pso_res.gbest_position[2]];
                marker = :star5, ms = 9, c = :red, markerstrokecolor = :black, markerstrokewidth = 0.6)
        scatter!(p1, [opt[1]], [opt[2]]; marker = :diamond, ms = 8, c = :white,
                markerstrokecolor = :black, markerstrokewidth = 0.8)

        p2 = contour(xs, ys, Z; levels = 28, c = :viridis, fill = false,
                    title = "FSS (it=$(k-1))", xlabel = "x1", ylabel = "x2", legend = false)
        scatter!(p2, Xf[1, :], Xf[2, :]; ms = 4.8, c = :limegreen, alpha = 0.95,
                markerstrokecolor = :black, markerstrokewidth = 0.4)
        scatter!(p2, [fss_res.best_position[1]], [fss_res.best_position[2]];
                marker = :star5, ms = 9, c = :orange, markerstrokecolor = :black, markerstrokewidth = 0.6)
        scatter!(p2, [opt[1]], [opt[2]]; marker = :diamond, ms = 8, c = :white,
                markerstrokecolor = :black, markerstrokewidth = 0.8)

        plot(p1, p2; layout = (1, 2), size = (1200, 520), margin = 6Plots.mm)
    end

    outpath = "pso_vs_fss_$(lowercase(problem_name)).gif"
    gif(anim, outpath; fps = fps)
end
```

```julia
make_pso_vs_fss_gif("Rastrigin"; n_agents = 50, n_iter = 200, seed = 7, every = 2, fps = 12)
```
