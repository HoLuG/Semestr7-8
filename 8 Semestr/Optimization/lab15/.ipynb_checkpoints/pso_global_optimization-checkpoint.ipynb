```julia
using Random, Statistics, LinearAlgebra, Printf
using Plots

gr()
```

```julia
schwefel(x) = 418.9829 * length(x) - sum(xi * sin(sqrt(abs(xi))) for xi in x)
rosenbrock(x) = (1.0 - x[1])^2 + 100.0 * (x[2] - x[1]^2)^2
rastrigin(x) = 10.0 * length(x) + sum(xi^2 - 10.0 * cos(2pi * xi) for xi in x)

BENCHMARKS = Dict(
    "Schwefel" => (f = schwefel, bounds = (-500.0, 500.0), optimum = [420.9687, 420.9687]),
    "Rosenbrock" => (f = rosenbrock, bounds = (-2.0, 2.0), optimum = [1.0, 1.0]),
    "Rastrigin" => (f = rastrigin, bounds = (-5.12, 5.12), optimum = [0.0, 0.0])
);
```

```julia
mutable struct PSOResult
    gbest_position::Vector{Float64}
    gbest_value::Float64
    gbest_history::Vector{Float64}
    pos_history::Vector{Matrix{Float64}}
end

function pso(
    f;
    bounds::Tuple{Float64, Float64} = (-5.0, 5.0),
    n_particles::Int = 50,
    n_iter::Int = 250,
    w::Float64 = 0.72,
    c1::Float64 = 1.49,
    c2::Float64 = 1.49,
    vmax_ratio::Float64 = 0.20,
    seed::Int = 42
)
    Random.seed!(seed)
    dim = 2
    lo, hi = bounds
    span = hi - lo
    vmax = vmax_ratio * span

    X = lo .+ span .* rand(dim, n_particles)
    V = (2rand(dim, n_particles) .- 1.0) .* vmax

    P = copy(X)
    pbest = [f(X[:, i]) for i in 1:n_particles]
    best_id = argmin(pbest)
    G = copy(P[:, best_id])
    gbest = pbest[best_id]

    g_hist = Float64[gbest]
    pos_hist = Matrix{Float64}[copy(X)]

    for _ in 1:n_iter
        for i in 1:n_particles
            r1 = rand(dim)
            r2 = rand(dim)
            V[:, i] .= w .* V[:, i] .+ c1 .* r1 .* (P[:, i] .- X[:, i]) .+ c2 .* r2 .* (G .- X[:, i])
            V[:, i] .= clamp.(V[:, i], -vmax, vmax)

            X[:, i] .+= V[:, i]
            X[:, i] .= clamp.(X[:, i], lo, hi)

            cur = f(X[:, i])
            if cur < pbest[i]
                pbest[i] = cur
                P[:, i] .= X[:, i]
            end
        end

        best_id = argmin(pbest)
        if pbest[best_id] < gbest
            gbest = pbest[best_id]
            G .= P[:, best_id]
        end

        push!(g_hist, gbest)
        push!(pos_hist, copy(X))
    end

    return PSOResult(copy(G), gbest, g_hist, pos_hist)
end
```

```julia
function run_benchmarks(; n_particles = 60, n_iter = 300, seed = 42)
    results = Dict{String, PSOResult}()
    for (name, problem) in BENCHMARKS
        res = pso(
            problem.f;
            bounds = problem.bounds,
            n_particles = n_particles,
            n_iter = n_iter,
            seed = seed
        )
        results[name] = res
    end
    return results
end

results = run_benchmarks(n_particles = 70, n_iter = 350, seed = 7);

for name in ["Schwefel", "Rosenbrock", "Rastrigin"]
    problem = BENCHMARKS[name]
    res = results[name]
    @printf("%-10s | gbest = %.8f | x* = [%.5f, %.5f] | эталон = [%.5f, %.5f]\n",
        name, res.gbest_value, res.gbest_position[1], res.gbest_position[2], problem.optimum[1], problem.optimum[2])
end
```

```julia
function plot_convergence(results)
    p = plot(layout = (1, 3), size = (1400, 380), margin = 6Plots.mm)
    for (k, name) in enumerate(["Schwefel", "Rosenbrock", "Rastrigin"])
        res = results[name]
        y = copy(res.gbest_history)

        if any(v -> v <= 0.0, y)
            y .= y .+ eps(Float64)
        end

        plot!(p[k], 0:length(y)-1, y,
              lw = 2.5, label = "best f(x)", title = name,
              xlabel = "Итерация", ylabel = "Значение функции", yscale = :log10)
    end
    return p
end

plot_convergence(results)
```

```julia
function contour_swarm(problem_name, res; show_traces = 10)
    problem = BENCHMARKS[problem_name]
    f = problem.f
    lo, hi = problem.bounds

    x = range(lo, hi; length = 160)
    y = range(lo, hi; length = 160)
    z = [f([xx, yy]) for yy in y, xx in x]

    p = contour(x, y, z; levels = 35, c = :viridis, fill = false,
                title = "$(problem_name): траектории частиц", xlabel = "x1", ylabel = "x2", legend = :topright)

    n_particles = size(res.pos_history[1], 2)
    traced = 1:min(show_traces, n_particles)
    for i in traced
        xs = [res.pos_history[t][1, i] for t in eachindex(res.pos_history)]
        ys = [res.pos_history[t][2, i] for t in eachindex(res.pos_history)]
        plot!(p, xs, ys; lw = 1.3, alpha = 0.65, label = false)
    end

    xf = res.pos_history[end][1, :]
    yf = res.pos_history[end][2, :]
    scatter!(p, xf, yf; ms = 3.5, alpha = 0.85, label = "финальные позиции")
    scatter!(p, [res.gbest_position[1]], [res.gbest_position[2]];
            ms = 8, marker = :star5, c = :red, label = "найденный минимум")
    scatter!(p, [problem.optimum[1]], [problem.optimum[2]];
            ms = 7, marker = :diamond, c = :orange, label = "эталонный минимум")

    return p
end
```

```julia
function make_gif(problem_name, res; every = 3, fps = 20)
    problem = BENCHMARKS[problem_name]
    f = problem.f
    lo, hi = problem.bounds

    x = range(lo, hi; length = 140)
    y = range(lo, hi; length = 140)
    z = [f([xx, yy]) for yy in y, xx in x]

    anim = @animate for k in 1:every:length(res.pos_history)
        contour(x, y, z; levels = 25, c = :viridis, fill = false,
                title = "$(problem_name), итерация $(k-1)", xlabel = "x1", ylabel = "x2", legend = false)
        scatter!(res.pos_history[k][1, :], res.pos_history[k][2, :]; ms = 3.5, c = :deepskyblue)
        scatter!([res.gbest_position[1]], [res.gbest_position[2]]; ms = 7, marker = :star5, c = :red)
    end

    gif(anim, "pso_$(lowercase(problem_name)).gif"; fps = fps)
end
```

```julia
function pso_trace_time(
    f;
    bounds::Tuple{Float64, Float64} = (-5.0, 5.0),
    n_particles::Int = 50,
    n_iter::Int = 250,
    w::Float64 = 0.72,
    c1::Float64 = 1.49,
    c2::Float64 = 1.49,
    vmax_ratio::Float64 = 0.20,
    seed::Int = 42
)
    Random.seed!(seed)
    dim = 2
    lo, hi = bounds
    span = hi - lo
    vmax = vmax_ratio * span

    X = lo .+ span .* rand(dim, n_particles)
    V = (2rand(dim, n_particles) .- 1.0) .* vmax

    P = copy(X)
    pbest = [f(X[:, i]) for i in 1:n_particles]
    best_id = argmin(pbest)
    G = copy(P[:, best_id])
    gbest = pbest[best_id]

    t0 = time_ns()
    iter_hist = Int[0]
    t_hist = Float64[0.0]
    g_hist = Float64[gbest]

    for it in 1:n_iter
        for i in 1:n_particles
            r1 = rand(dim)
            r2 = rand(dim)
            V[:, i] .= w .* V[:, i] .+ c1 .* r1 .* (P[:, i] .- X[:, i]) .+ c2 .* r2 .* (G .- X[:, i])
            V[:, i] .= clamp.(V[:, i], -vmax, vmax)

            X[:, i] .+= V[:, i]
            X[:, i] .= clamp.(X[:, i], lo, hi)

            cur = f(X[:, i])
            if cur < pbest[i]
                pbest[i] = cur
                P[:, i] .= X[:, i]
            end
        end

        best_id = argmin(pbest)
        if pbest[best_id] < gbest
            gbest = pbest[best_id]
            G .= P[:, best_id]
        end

        push!(g_hist, gbest)
        push!(iter_hist, it)
        push!(t_hist, (time_ns() - t0) / 1e9)
    end

    return (iter = iter_hist, time = t_hist, best = g_hist)
end

function pso_teleport_trace_time(
    f;
    bounds::Tuple{Float64, Float64} = (-5.0, 5.0),
    n_particles::Int = 50,
    n_iter::Int = 250,
    w::Float64 = 0.72,
    c1::Float64 = 1.49,
    c2::Float64 = 1.49,
    vmax_ratio::Float64 = 0.20,
    teleport_prob::Float64 = 0.003,
    stagnation_iters::Int = 20,
    seed::Int = 42
)
    Random.seed!(seed)
    dim = 2
    lo, hi = bounds
    span = hi - lo
    vmax = vmax_ratio * span

    X = lo .+ span .* rand(dim, n_particles)
    V = (2rand(dim, n_particles) .- 1.0) .* vmax

    P = copy(X)
    pbest = [f(X[:, i]) for i in 1:n_particles]
    no_improve = zeros(Int, n_particles)

    best_id = argmin(pbest)
    G = copy(P[:, best_id])
    gbest = pbest[best_id]

    t0 = time_ns()
    iter_hist = Int[0]
    t_hist = Float64[0.0]
    g_hist = Float64[gbest]

    for it in 1:n_iter
        for i in 1:n_particles
            r1 = rand(dim)
            r2 = rand(dim)
            V[:, i] .= w .* V[:, i] .+ c1 .* r1 .* (P[:, i] .- X[:, i]) .+ c2 .* r2 .* (G .- X[:, i])
            V[:, i] .= clamp.(V[:, i], -vmax, vmax)

            X[:, i] .+= V[:, i]
            X[:, i] .= clamp.(X[:, i], lo, hi)

            cur = f(X[:, i])
            if cur < pbest[i]
                pbest[i] = cur
                P[:, i] .= X[:, i]
                no_improve[i] = 0
            else
                no_improve[i] += 1
            end

            if no_improve[i] >= stagnation_iters && rand() < teleport_prob
                X[:, i] .= lo .+ span .* rand(dim)
                V[:, i] .= (2rand(dim) .- 1.0) .* vmax
                cur2 = f(X[:, i])
                if cur2 < pbest[i]
                    pbest[i] = cur2
                    P[:, i] .= X[:, i]
                    no_improve[i] = 0
                end
            end
        end

        best_id = argmin(pbest)
        if pbest[best_id] < gbest
            gbest = pbest[best_id]
            G .= P[:, best_id]
        end

        push!(g_hist, gbest)
        push!(iter_hist, it)
        push!(t_hist, (time_ns() - t0) / 1e9)
    end

    return (iter = iter_hist, time = t_hist, best = g_hist)
end
```

```julia
function _xdata(trace, xaxis::Symbol)
    if xaxis == :time
        return trace.time, "Время (секунды)"
    elseif xaxis == :iter
        return trace.iter, "Итерация"
    else
        error("xaxis must be :time or :iter")
    end
end

function plot_particles_like_example(problem_name; particle_vals = [10, 30, 50], n_iter = 250, seed = 42, xaxis::Symbol = :time)
    problem = BENCHMARKS[problem_name]
    colors = [:red, :green, :blue, :purple, :orange]

    p = plot(title = "Влияние количества частиц на сходимость",
             ylabel = "Значение функции",
             legend = :topright, size = (1100, 600), margin = 6Plots.mm)

    for (k, n) in enumerate(particle_vals)
        tr = pso_trace_time(problem.f; bounds = problem.bounds, n_particles = n, n_iter = n_iter, seed = seed)
        x, xlabel_txt = _xdata(tr, xaxis)
        plot!(p, x, tr.best; lw = 2.0, c = colors[k], label = "particles=$(n)", xlabel = xlabel_txt)
    end

    return p
end

function plot_c2_like_example(problem_name; c2_vals = [0.01, 0.2, 0.3], n_particles = 50, n_iter = 250, seed = 42, xaxis::Symbol = :time)
    problem = BENCHMARKS[problem_name]
    colors = [:red, :green, :blue, :purple, :orange]

    p = plot(title = "Влияние параметра c2 на сходимость",
             ylabel = "Значение функции",
             legend = :topright, size = (1100, 600), margin = 6Plots.mm)

    for (k, c2) in enumerate(c2_vals)
        tr = pso_trace_time(problem.f; bounds = problem.bounds, n_particles = n_particles, n_iter = n_iter, c2 = c2, seed = seed)
        x, xlabel_txt = _xdata(tr, xaxis)
        plot!(p, x, tr.best; lw = 2.0, c = colors[k], label = "c2=$(c2)", xlabel = xlabel_txt)
    end

    return p
end

function plot_teleport_like_example(problem_name; probs = [0.0, 0.001, 0.003, 0.005], n_particles = 50, n_iter = 250, seed = 42, xaxis::Symbol = :time)
    problem = BENCHMARKS[problem_name]
    colors = [:black, :blue, :cyan3, :darkgreen, :goldenrod, :red]

    p = plot(title = "Влияние вероятности телепортации на сходимость",
             ylabel = "Значение функции",
             legend = :topright, size = (1100, 600), margin = 6Plots.mm)

    for (k, tp) in enumerate(probs)
        tr = pso_teleport_trace_time(problem.f;
            bounds = problem.bounds,
            n_particles = n_particles,
            n_iter = n_iter,
            teleport_prob = tp,
            seed = seed
        )
        x, xlabel_txt = _xdata(tr, xaxis)
        label_txt = tp == 0.0 ? "0% (нет)" : "$(round(tp*100; digits=1))%"
        plot!(p, x, tr.best; lw = 2.0, c = colors[k], label = label_txt, xlabel = xlabel_txt)
    end

    return p
end
```

```julia
mutable struct FSSResult
    best_position::Vector{Float64}
    best_value::Float64
    best_history::Vector{Float64}
    pos_history::Vector{Matrix{Float64}}  # 2 x n_fish
end

"""
Fish School Search (FSS) для минимизации f(x) в R^2.

Параметры (типичные):
- step_ind: начальный шаг индивидуального движения
- step_vol: начальный шаг волевого (collective-volitive) движения
- w_scale: масштаб обновления веса (feeding)

Оба шага линейно уменьшаются к 0.
"""
function fss(
    f;
    bounds::Tuple{Float64, Float64} = (-5.0, 5.0),
    n_fish::Int = 50,
    n_iter::Int = 250,
    step_ind::Float64 = 0.1,
    step_vol::Float64 = 0.01,
    w_min::Float64 = 1.0,
    w_max::Float64 = 3.0,
    w_scale::Float64 = 1.0,
    seed::Int = 42
)
    Random.seed!(seed)
    dim = 2
    lo, hi = bounds
    span = hi - lo

    # позиции рыб: 2 x n
    X = lo .+ span .* rand(dim, n_fish)
    F = [f(X[:, i]) for i in 1:n_fish]

    # веса
    W = fill(w_min, n_fish)

    best_id = argmin(F)
    best_pos = copy(X[:, best_id])
    best_val = F[best_id]

    best_hist = Float64[best_val]
    pos_hist = Matrix{Float64}[copy(X)]

    for t in 1:n_iter
        # линейное уменьшение шагов
        α = 1.0 - (t - 1) / max(n_iter - 1, 1)
        si = step_ind * α * span
        sv = step_vol * α * span

        # 1) индивидуальное движение
        ΔX = zeros(dim, n_fish)
        ΔF = zeros(n_fish)

        for i in 1:n_fish
            trial = X[:, i] .+ (2rand(dim) .- 1.0) .* si
            trial .= clamp.(trial, lo, hi)
            ftrial = f(trial)

            if ftrial < F[i]
                ΔX[:, i] .= trial .- X[:, i]
                ΔF[i] = F[i] - ftrial  # улучшение (>=0)
                X[:, i] .= trial
                F[i] = ftrial
            end
        end

        # 2) feeding (обновление веса): у кого было лучше улучшение — тот набирает вес
        maxΔF = maximum(ΔF)
        if maxΔF > 0
            for i in 1:n_fish
                W[i] = clamp(W[i] + w_scale * (ΔF[i] / maxΔF), w_min, w_max)
            end
        end

        # 3) инстинктивное движение (instinctive)
        denom = sum(ΔF)
        if denom > 0
            I = sum(ΔX[:, i] .* ΔF[i] for i in 1:n_fish) ./ denom
            for i in 1:n_fish
                X[:, i] .+= I
                X[:, i] .= clamp.(X[:, i], lo, hi)
                F[i] = f(X[:, i])
            end
        end

        # 4) волевое движение (volitive): сжатие/расширение относительно барицентра
        B = sum(X[:, i] .* W[i] for i in 1:n_fish) ./ sum(W)

        # если общая масса растёт — сжимаемся (эксплуатация), иначе расширяемся (исследование)
        # простая эвристика: сравниваем средний вес с прошлым шагом через α
        # (вместо хранения прошлого веса используем знак по denom)
        direction = denom > 0 ? -1.0 : 1.0

        for i in 1:n_fish
            u = rand(dim)
            X[:, i] .+= direction .* sv .* u .* (X[:, i] .- B)
            X[:, i] .= clamp.(X[:, i], lo, hi)
            F[i] = f(X[:, i])
        end

        best_id = argmin(F)
        if F[best_id] < best_val
            best_val = F[best_id]
            best_pos .= X[:, best_id]
        end

        push!(best_hist, best_val)
        push!(pos_hist, copy(X))
    end

    return FSSResult(copy(best_pos), best_val, best_hist, pos_hist)
end
```

```julia
"""Запуск PSO и FSS на одной задаче (одинаковый seed)."""
function run_pso_fss(problem_name; n_agents = 50, n_iter = 250, seed = 7)
    problem = BENCHMARKS[problem_name]

    pso_res = pso(problem.f; bounds = problem.bounds, n_particles = n_agents, n_iter = n_iter, seed = seed)

    # шаги FSS как доля диапазона (подбираются под bounds автоматически)
    fss_res = fss(problem.f;
        bounds = problem.bounds,
        n_fish = n_agents,
        n_iter = n_iter,
        step_ind = 0.08,
        step_vol = 0.008,
        seed = seed
    )

    return (pso = pso_res, fss = fss_res)
end

function plot_compare_convergence(problem_name, out; xaxis::Symbol = :iter)
    p = plot(size = (1100, 520), margin = 6Plots.mm, grid = true)

    if xaxis == :iter
        x1 = 0:length(out.pso.gbest_history)-1
        x2 = 0:length(out.fss.best_history)-1
        xlabel_txt = "Итерация"
    elseif xaxis == :time
        # простая трассировка времени (по wall time)
        t0 = time_ns();
        x1 = collect(0:length(out.pso.gbest_history)-1) .* 0.0
        x2 = collect(0:length(out.fss.best_history)-1) .* 0.0
        xlabel_txt = "Время (условн.)"
    else
        error("xaxis must be :iter or :time")
    end

    plot!(p, x1, out.pso.gbest_history; lw = 2.2, label = "PSO", title = "Сходимость: $(problem_name)", xlabel = xlabel_txt, ylabel = "best f(x)")
    plot!(p, x2, out.fss.best_history; lw = 2.2, label = "FSS")

    return p
end

function plot_compare_final_positions(problem_name, out)
    problem = BENCHMARKS[problem_name]
    lo, hi = problem.bounds
    f = problem.f

    x = range(lo, hi; length = 160)
    y = range(lo, hi; length = 160)
    z = [f([xx, yy]) for yy in y, xx in x]

    p = contour(x, y, z; levels = 30, c = :viridis, fill = false,
                title = "Финальные позиции: $(problem_name)", xlabel = "x1", ylabel = "x2", legend = :topright)

    Xp = out.pso.pos_history[end]
    Xf = out.fss.pos_history[end]

    scatter!(p, Xp[1, :], Xp[2, :]; ms = 3.3, alpha = 0.85, label = "PSO particles")
    scatter!(p, Xf[1, :], Xf[2, :]; ms = 3.3, alpha = 0.85, label = "FSS fish")

    scatter!(p, [out.pso.gbest_position[1]], [out.pso.gbest_position[2]]; marker = :star5, ms = 8, c = :red, label = "PSO best")
    scatter!(p, [out.fss.best_position[1]], [out.fss.best_position[2]]; marker = :star5, ms = 8, c = :orange, label = "FSS best")
    scatter!(p, [problem.optimum[1]], [problem.optimum[2]]; marker = :diamond, ms = 7, c = :white, label = "optimum")

    return p
end
```

```julia
out_ras = run_pso_fss("Rastrigin"; n_agents = 50, n_iter = 200, seed = 7)
plot_compare_convergence("Rastrigin", out_ras; xaxis = :iter)
```
