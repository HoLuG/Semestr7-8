# Лабораторная 3: проверка адекватности через логическую модель (СДНФ/СКНФ)

Цель: проверить, можно ли **детерминированно** описать результаты кластеризации (метки классов `y`) с помощью **логической модели** на основе бинаризованных признаков.

Идея проверки:

- На **рабочей** выборке (50%) строим модель кластеризации (без использования `Outcome`).
- Получаем метки кластеров `y` для рабочей и контрольной выборок (контрольная 50%).
- Бинаризуем исходные признаки по медианам (порогам), получая булевы переменные \(x_1,\dots,x_p\).
- Строим **совершенную форму** (СДНФ или СКНФ) для `y` и выполняем упрощение.
- На контрольной проверяем совпадение по классам 0/1 (точность по 0 и по 1 + суммарная).

Дополнительно:

- Проверяем «отцепление» признаков: удаляем признаки по одному и смотрим, ухудшается ли проверка на контрольной.
- При необходимости выполняем полный перебор подмножеств признаков (для 8 признаков это 256 вариантов).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.dpi"] = 140
np.set_printoptions(suppress=True, precision=4)
```

```python
df = pd.read_csv("diabetes.csv")
features = [c for c in df.columns if c != "Outcome"]
X = df[features].to_numpy(dtype=float)

print("Shape:", df.shape)
print("Features:", features)
print("Outcome counts (справочно):", df["Outcome"].value_counts().to_dict())
```

```python
def zscore_fit(X):
    mu = X.mean(axis=0)
    sigma = X.std(axis=0, ddof=0)
    sigma = np.where(sigma == 0, 1.0, sigma)
    return mu, sigma

def zscore_transform(X, mu, sigma):
    return (X - mu) / sigma

def binarize_fit_median(X):
    t = np.median(X, axis=0)
    return t

def binarize_transform(X, t):
    # x_i = 1 <=> X[:,i] <= median_i
    return (X <= t).astype(np.uint8)

def split_train_test_stratified(y, test_size=0.5, seed=42):
    rng = np.random.default_rng(seed)
    idx0 = np.where(y == 0)[0]
    idx1 = np.where(y == 1)[0]
    rng.shuffle(idx0)
    rng.shuffle(idx1)

    n0_test = int(round(len(idx0) * test_size))
    n1_test = int(round(len(idx1) * test_size))

    test_idx = np.concatenate([idx0[:n0_test], idx1[:n1_test]])
    train_idx = np.concatenate([idx0[n0_test:], idx1[n1_test:]])
    rng.shuffle(test_idx)
    rng.shuffle(train_idx)
    return train_idx, test_idx

def kmeans2_fit(X, max_iter=200, seed=1):
    rng = np.random.default_rng(seed)
    n = X.shape[0]
    # init: pick random point and farthest from it
    i0 = rng.integers(0, n)
    c1 = X[i0].copy()
    d2 = ((X - c1) ** 2).sum(axis=1)
    c2 = X[np.argmax(d2)].copy()
    C = np.vstack([c1, c2])

    for _ in range(max_iter):
        dist = ((X[:, None, :] - C[None, :, :]) ** 2).sum(axis=2)
        labels = np.argmin(dist, axis=1)
        newC = C.copy()
        for k in (0, 1):
            pts = X[labels == k]
            if len(pts) > 0:
                newC[k] = pts.mean(axis=0)
        if np.allclose(newC, C, atol=1e-8):
            C = newC
            break
        C = newC
    return C, labels

def kmeans2_predict(X, C):
    dist = ((X[:, None, :] - C[None, :, :]) ** 2).sum(axis=2)
    return np.argmin(dist, axis=1)

def silhouette_score_manual(X, labels):
    # O(n^2) but n=768 OK
    n = X.shape[0]
    if len(np.unique(labels)) < 2:
        return 0.0
    D = np.sqrt(((X[:, None, :] - X[None, :, :]) ** 2).sum(axis=2))
    s = np.zeros(n)
    for i in range(n):
        same = labels == labels[i]
        other = ~same
        a = D[i, same].mean() if same.sum() > 1 else 0.0
        # mean distance to the other cluster (only 2 clusters)
        b = D[i, other].mean() if other.sum() > 0 else 0.0
        den = max(a, b)
        s[i] = 0.0 if den == 0 else (b - a) / den
    return float(s.mean())

def davies_bouldin_manual(X, labels):
    # 2 clusters
    uniq = np.unique(labels)
    if len(uniq) < 2:
        return float("inf")
    centers = np.vstack([X[labels == k].mean(axis=0) for k in uniq])
    S = np.array([
        np.sqrt(((X[labels == k] - centers[i]) ** 2).sum(axis=1)).mean()
        for i, k in enumerate(uniq)
    ])
    M = np.sqrt(((centers[0] - centers[1]) ** 2).sum())
    if M == 0:
        return float("inf")
    R01 = (S[0] + S[1]) / M
    return float(R01)  # for k=2, DB is just this

def calinski_harabasz_manual(X, labels):
    uniq = np.unique(labels)
    k = len(uniq)
    n = X.shape[0]
    if k < 2 or n <= k:
        return 0.0
    overall = X.mean(axis=0)
    B = 0.0
    W = 0.0
    for cl in uniq:
        Xi = X[labels == cl]
        ci = Xi.mean(axis=0)
        ni = Xi.shape[0]
        B += ni * ((ci - overall) ** 2).sum()
        W += ((Xi - ci) ** 2).sum()
    if W == 0:
        return 0.0
    return float((B / (k - 1)) / (W / (n - k)))

def score_heuristic(sil, db, ch, m, balance):
    return float(sil - 0.15 * np.log1p(db) + 0.02 * np.log1p(ch) - 0.03 * m + 0.10 * balance)
```

```python
def contingency_counts(xi, xj):
    # xi, xj are 0/1 arrays with meaning: 1 means event A holds (<= median)
    xi = xi.astype(bool)
    xj = xj.astype(bool)
    a = int(np.sum(xi & xj))
    b = int(np.sum(xi & (~xj)))
    c = int(np.sum((~xi) & xj))
    d = int(np.sum((~xi) & (~xj)))
    return a, b, c, d

def phi_contingency(a, b, c, d):
    den = (a + b) * (c + d) * (a + c) * (b + d)
    if den == 0:
        return 0.0
    return (a * d - b * c) / np.sqrt(den)

def yule_q(a, b, c, d):
    den = a * d + b * c
    if den == 0:
        return 0.0
    return (a * d - b * c) / den

def yule_y(a, b, c, d):
    ad = a * d
    bc = b * c
    if ad == 0 and bc == 0:
        return 0.0
    return (np.sqrt(ad) - np.sqrt(bc)) / (np.sqrt(ad) + np.sqrt(bc))

def lift_k(a, b, c, d):
    n = a + b + c + d
    if n == 0:
        return 1.0
    pA = (a + b) / n
    pA_given_B = a / (a + c) if (a + c) > 0 else 0.0
    if pA == 0:
        return 1.0
    return pA_given_B / pA

def similarity_matrix(Xb, measure="Y"):
    # Xb: (n,p) in {0,1}
    p = Xb.shape[1]
    S = np.eye(p)
    for i in range(p):
        for j in range(i + 1, p):
            a, b, c, d = contingency_counts(Xb[:, i], Xb[:, j])
            if measure == "K":
                kij = lift_k(a, b, c, d)
                kji = lift_k(a + c, b + d, a, b)  # not used, just to keep symmetric via max below
                s = max(kij, kji)
            elif measure == "PHI":
                s = abs(phi_contingency(a, b, c, d))
            elif measure == "Q":
                s = abs(yule_q(a, b, c, d))
            elif measure == "Y":
                s = abs(yule_y(a, b, c, d))
            else:
                raise ValueError("Unknown measure")
            S[i, j] = S[j, i] = float(s)
    return S

def mst_groups(S, target_m):
    # Maximum spanning tree -> cut to get exactly target_m components
    p = S.shape[0]
    edges = []
    for i in range(p):
        for j in range(i + 1, p):
            edges.append((S[i, j], i, j))
    edges.sort(reverse=True, key=lambda x: x[0])

    parent = list(range(p))
    rank = [0] * p

    def find(x):
        while parent[x] != x:
            parent[x] = parent[parent[x]]
            x = parent[x]
        return x

    def union(a, b):
        ra, rb = find(a), find(b)
        if ra == rb:
            return False
        if rank[ra] < rank[rb]:
            ra, rb = rb, ra
        parent[rb] = ra
        if rank[ra] == rank[rb]:
            rank[ra] += 1
        return True

    # Build maximum spanning forest with (p - target_m) edges
    need_edges = p - target_m
    used = []
    for w, i, j in edges:
        if union(i, j):
            used.append((w, i, j))
            if len(used) == need_edges:
                break

    comps = {}
    for i in range(p):
        r = find(i)
        comps.setdefault(r, []).append(i)

    groups = list(comps.values())
    # equivalent threshold: smallest kept edge weight (if any)
    thr_equiv = used[-1][0] if used else 1.0
    return groups, float(thr_equiv)

def build_composites(Xz, groups):
    # each group -> mean of standardized columns
    G = np.zeros((Xz.shape[0], len(groups)))
    for gi, cols in enumerate(groups):
        G[:, gi] = Xz[:, cols].mean(axis=1)
    return G

def choose_reduction_by_score(X_train_raw, m_values=range(2, 9), measures=("K", "PHI", "Q", "Y")):
    # Fit binarization + similarity on TRAIN only
    t = binarize_fit_median(X_train_raw)
    Xb = binarize_transform(X_train_raw, t)

    best = None
    for meas in measures:
        S = similarity_matrix(Xb, measure=meas)
        for m in m_values:
            groups, thr_equiv = mst_groups(S, target_m=m)
            mu, sigma = zscore_fit(X_train_raw)
            Xz = zscore_transform(X_train_raw, mu, sigma)
            G = build_composites(Xz, groups)
            muG, sigG = zscore_fit(G)
            Gz = zscore_transform(G, muG, sigG)

            C, y = kmeans2_fit(Gz, seed=1)
            n0 = int(np.sum(y == 0))
            n1 = int(np.sum(y == 1))
            balance = min(n0, n1) / max(n0, n1) if max(n0, n1) > 0 else 0.0
            sil = silhouette_score_manual(Gz, y)
            db = davies_bouldin_manual(Gz, y)
            ch = calinski_harabasz_manual(Gz, y)
            sc = score_heuristic(sil, db, ch, m=m, balance=balance)

            cand = {
                "measure": meas,
                "m": int(m),
                "thr_equiv": float(thr_equiv),
                "groups": groups,
                "bin_thr": t,
                "mu": mu,
                "sigma": sigma,
                "muG": muG,
                "sigG": sigG,
                "centers": C,
                "score": sc,
                "sil": sil,
                "db": db,
                "ch": ch,
                "balance": balance,
            }
            if best is None or cand["score"] > best["score"]:
                best = cand

    return best

def transform_for_model(X_raw, model):
    # Use fitted params from model
    Xz = zscore_transform(X_raw, model["mu"], model["sigma"])
    G = build_composites(Xz, model["groups"])
    Gz = zscore_transform(G, model["muG"], model["sigG"])
    return Gz
```

```python
# 1) Строим кластеризацию на рабочей выборке и проверяем на контрольной

# Сначала грубо получим y для стратификации: кластеризация на всех данных по тем же правилам
# (Outcome не используется, это нужно только чтобы разделить 50/50 равнозначно по y)
base_model = choose_reduction_by_score(X)
G_all = transform_for_model(X, base_model)
centers_all = base_model["centers"]
y_all = kmeans2_predict(G_all, centers_all)

print("Chosen (by score on full data):", {k: base_model[k] for k in ["measure","m","thr_equiv","score","sil","db","ch","balance"]})
print("Cluster sizes (full):", {0: int((y_all==0).sum()), 1: int((y_all==1).sum())})

train_idx, test_idx = split_train_test_stratified(y_all, test_size=0.5, seed=42)
X_train = X[train_idx]
X_test = X[test_idx]

# Теперь честно: модель кластеризации обучаем только на рабочей выборке
model = choose_reduction_by_score(X_train)
G_train = transform_for_model(X_train, model)
C_train, y_train = kmeans2_fit(G_train, seed=1)

# фиксируем центры и применяем к контрольной (без переобучения)
G_test = transform_for_model(X_test, model)
y_test = kmeans2_predict(G_test, C_train)

print("Chosen (by score on TRAIN):", {k: model[k] for k in ["measure","m","thr_equiv","score","sil","db","ch","balance"]})
print("Cluster sizes (train):", {0: int((y_train==0).sum()), 1: int((y_train==1).sum())})
print("Cluster sizes (test):", {0: int((y_test==0).sum()), 1: int((y_test==1).sum())})
```

```python
# 2) Бинаризация признаков (для логической модели)

bin_thr = binarize_fit_median(X_train)  # пороги только по рабочей выборке
Xb_train = binarize_transform(X_train, bin_thr)
Xb_test = binarize_transform(X_test, bin_thr)

print("Binary matrix shape:", Xb_train.shape)
print("Example row:", Xb_train[0])
```

```python
def bits_to_int(bits):
    v = 0
    for b in bits:
        v = (v << 1) | int(b)
    return v

def int_to_bits(v, p):
    return [(v >> (p - 1 - i)) & 1 for i in range(p)]

def build_truth_table(Xb, y):
    # returns dict: code -> (n_total, n1)
    tbl = {}
    for bits, yy in zip(Xb, y):
        code = bits_to_int(bits)
        n, n1 = tbl.get(code, (0, 0))
        tbl[code] = (n + 1, n1 + int(yy == 1))
    return tbl

def table_stats(tbl):
    # purity per observed code
    pure = 0
    amb = 0
    for code, (n, n1) in tbl.items():
        if n1 == 0 or n1 == n:
            pure += 1
        else:
            amb += 1
    return pure, amb

def majority_mapping(tbl):
    # code -> 0/1 by majority (ties -> 0)
    mp = {}
    for code, (n, n1) in tbl.items():
        mp[code] = 1 if n1 > (n - n1) else 0
    return mp

def predict_by_mapping(Xb, mapping, fallback=0):
    yhat = np.zeros(Xb.shape[0], dtype=np.int64)
    unseen = 0
    for i, bits in enumerate(Xb):
        code = bits_to_int(bits)
        if code in mapping:
            yhat[i] = mapping[code]
        else:
            yhat[i] = fallback
            unseen += 1
    return yhat, unseen

def confusion(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    tp = int(np.sum((y_true == 1) & (y_pred == 1)))
    tn = int(np.sum((y_true == 0) & (y_pred == 0)))
    fp = int(np.sum((y_true == 0) & (y_pred == 1)))
    fn = int(np.sum((y_true == 1) & (y_pred == 0)))
    return tn, fp, fn, tp

def per_class_accuracy(tn, fp, fn, tp):
    acc0 = tn / (tn + fp) if (tn + fp) else 0.0
    acc1 = tp / (tp + fn) if (tp + fn) else 0.0
    acc = (tp + tn) / (tp + tn + fp + fn) if (tp + tn + fp + fn) else 0.0
    bal = 0.5 * (acc0 + acc1)
    return acc, acc0, acc1, bal

train_tbl = build_truth_table(Xb_train, y_train)
pure, amb = table_stats(train_tbl)
print(f"Observed combinations on train: {len(train_tbl)} / {2**Xb_train.shape[1]}")
print(f"Pure combos: {pure}, ambiguous combos: {amb}")

mapping = majority_mapping(train_tbl)
fallback = int(np.mean(y_train) >= 0.5)

yhat_train, unseen_train = predict_by_mapping(Xb_train, mapping, fallback=fallback)
yhat_test, unseen_test = predict_by_mapping(Xb_test, mapping, fallback=fallback)

tn, fp, fn, tp = confusion(y_train, yhat_train)
acc, acc0, acc1, bal = per_class_accuracy(tn, fp, fn, tp)
print("TRAIN majority-table:", {"acc": acc, "acc0": acc0, "acc1": acc1, "bal": bal, "unseen": unseen_train})

tn, fp, fn, tp = confusion(y_test, yhat_test)
acc, acc0, acc1, bal = per_class_accuracy(tn, fp, fn, tp)
print("TEST  majority-table:", {"acc": acc, "acc0": acc0, "acc1": acc1, "bal": bal, "unseen": unseen_test})
```

```python
# 3) Совершенные формы (СДНФ / СКНФ) и упрощение (Куайн–МакКласки)

def minterm_str(bits, names=None):
    if names is None:
        names = [f"x{i+1}" for i in range(len(bits))]
    parts = []
    for b, nm in zip(bits, names):
        parts.append(nm if b == 1 else f"¬{nm}")
    return " ∧ ".join(parts) if parts else "1"

def maxterm_str(bits, names=None):
    # for CNF: for y=0 row, maxterm is OR of literals (xi if bit=0 else ¬xi)
    if names is None:
        names = [f"x{i+1}" for i in range(len(bits))]
    parts = []
    for b, nm in zip(bits, names):
        parts.append(nm if b == 0 else f"¬{nm}")
    return " ∨ ".join(parts) if parts else "0"

def sdnf_from_mapping(mapping, p, names=None):
    ones = [code for code, val in mapping.items() if val == 1]
    if not ones:
        return "0", ones
    terms = [f"({minterm_str(int_to_bits(code, p), names)})" for code in sorted(ones)]
    return " ∨ ".join(terms), ones

def sknf_from_mapping(mapping, p, names=None):
    zeros = [code for code, val in mapping.items() if val == 0]
    if not zeros:
        return "1", zeros
    terms = [f"({maxterm_str(int_to_bits(code, p), names)})" for code in sorted(zeros)]
    return " ∧ ".join(terms), zeros

# Quine–McCluskey for DNF minimization, supporting don't-cares

def qm_minimize(ones, dcs, p):
    # terms encoded as (mask, bits) where mask=1 means don't-care for that position
    def bits_term(v):
        return (0, v)

    def combine(t1, t2):
        m1, b1 = t1
        m2, b2 = t2
        if m1 != m2:
            return None
        diff = b1 ^ b2
        # must differ in exactly one non-masked bit
        if diff == 0:
            return None
        if diff & (diff - 1):
            return None
        if m1 & diff:
            return None
        return (m1 | diff, b1 & ~diff)

    # initial groups by popcount
    current = [bits_term(v) for v in sorted(set(ones + dcs))]

    prime_implicants = set()
    while True:
        used = set()
        next_set = set()
        groups = {}
        for t in current:
            m, b = t
            # popcount only on fixed bits
            pc = int(bin(b & ~m).count("1"))
            groups.setdefault(pc, []).append(t)

        keys = sorted(groups.keys())
        for k in keys:
            for t1 in groups.get(k, []):
                for t2 in groups.get(k + 1, []):
                    cmb = combine(t1, t2)
                    if cmb is not None:
                        used.add(t1)
                        used.add(t2)
                        next_set.add(cmb)

        for t in current:
            if t not in used:
                prime_implicants.add(t)

        if not next_set:
            break
        current = sorted(next_set)

    # coverage table: PI -> covered minterms
    def covers(pi, v):
        m, b = pi
        return (v & ~m) == (b & ~m)

    ones_set = sorted(set(ones))
    cover = {pi: [v for v in ones_set if covers(pi, v)] for pi in prime_implicants}
    cover = {pi: cov for pi, cov in cover.items() if cov}

    # Petrick's method (small p, few terms)
    # Build product of sums over minterms: for each minterm choose one PI that covers it
    mt_to_pis = {v: [pi for pi, cov in cover.items() if v in cov] for v in ones_set}
    if any(len(v) == 0 for v in mt_to_pis.values()):
        return []  # should not happen

    # represent a solution as set of PIs
    sols = [set([pi]) for pi in mt_to_pis[ones_set[0]]]
    for mt in ones_set[1:]:
        new_sols = []
        for s in sols:
            for pi in mt_to_pis[mt]:
                ns = set(s)
                ns.add(pi)
                new_sols.append(ns)
        # prune supersets by size then literal count
        def lit_count(sol):
            c = 0
            for m, _ in sol:
                c += p - int(bin(m).count("1"))
            return c
        # keep Pareto-front a bit
        new_sols.sort(key=lambda s: (len(s), lit_count(s)))
        pruned = []
        for s in new_sols:
            if any(s2.issubset(s) for s2 in pruned):
                continue
            pruned.append(s)
            if len(pruned) > 200:
                break
        sols = pruned

    # choose best
    def lit_count(sol):
        c = 0
        for m, _ in sol:
            c += p - int(bin(m).count("1"))
        return c

    best = min(sols, key=lambda s: (len(s), lit_count(s)))
    return sorted(best)


def pi_to_str(pi, p, names=None):
    if names is None:
        names = [f"x{i+1}" for i in range(p)]
    m, b = pi
    parts = []
    for i in range(p):
        bit_mask = 1 << (p - 1 - i)
        if m & bit_mask:
            continue
        lit = names[i] if (b & bit_mask) else f"¬{names[i]}"
        parts.append(lit)
    return " ∧ ".join(parts) if parts else "1"


def eval_pi_list(pi_list, Xb):
    # Evaluate DNF: OR of implicants
    p = Xb.shape[1]
    v = np.array([bits_to_int(row) for row in Xb], dtype=np.int64)
    out = np.zeros(len(v), dtype=np.uint8)
    for m, b in pi_list:
        # (v & ~m) == (b & ~m)
        ok = ((v & ~m) == (b & ~m))
        out |= ok.astype(np.uint8)
    return out


p = Xb_train.shape[1]
var_names = [f"x{i+1}" for i in range(p)]

sdnf_str, ones = sdnf_from_mapping(mapping, p, names=var_names)
sknf_str, zeros = sknf_from_mapping(mapping, p, names=var_names)

print("Perfect SDNF length (terms):", len(ones))
print("Perfect SKNF length (clauses):", len(zeros))

# Don't-cares: combinations not observed in train
observed = set(train_tbl.keys())
all_codes = set(range(2**p))
dcs = sorted(list(all_codes - observed))

pi_list = qm_minimize(ones=ones, dcs=dcs, p=p)
print("Minimized DNF implicants:", len(pi_list))

expr = " ∨ ".join([f"({pi_to_str(pi, p, var_names)})" for pi in pi_list]) if pi_list else "0"
print("Minimized DNF:")
print(expr)

# evaluate minimized DNF
pred_train = eval_pi_list(pi_list, Xb_train)
pred_test = eval_pi_list(pi_list, Xb_test)

tn, fp, fn, tp = confusion(y_train, pred_train)
print("TRAIN minimized DNF:", dict(zip(["acc","acc0","acc1","bal"], per_class_accuracy(tn, fp, fn, tp))))

tn, fp, fn, tp = confusion(y_test, pred_test)
print("TEST  minimized DNF:", dict(zip(["acc","acc0","acc1","bal"], per_class_accuracy(tn, fp, fn, tp))))

print("\nСДНФ (совершенная, по наблюдаемым комбинациям):")
print(sdnf_str[:800] + (" ..." if len(sdnf_str) > 800 else ""))
print("\nСКНФ (совершенная, по наблюдаемым комбинациям):")
print(sknf_str[:800] + (" ..." if len(sknf_str) > 800 else ""))
```

```python
# 4) «Отцепление» признаков
#    Критерий: не ухудшаем Balanced Accuracy на контрольной, стараемся уменьшить число переменных.

def run_logic_pipeline_with_features(feature_idx, seed=42):
    Xtr = X_train[:, feature_idx]
    Xte = X_test[:, feature_idx]
    t = binarize_fit_median(Xtr)
    Xbtr = binarize_transform(Xtr, t)
    Xbte = binarize_transform(Xte, t)

    tbl = build_truth_table(Xbtr, y_train)
    mp = majority_mapping(tbl)
    obs = set(tbl.keys())
    p = Xbtr.shape[1]
    ones = [code for code, val in mp.items() if val == 1]
    dcs = sorted(list(set(range(2**p)) - obs))
    pi_list = qm_minimize(ones=ones, dcs=dcs, p=p)

    pred_te = eval_pi_list(pi_list, Xbte)
    tn, fp, fn, tp = confusion(y_test, pred_te)
    acc, acc0, acc1, bal = per_class_accuracy(tn, fp, fn, tp)
    return {
        "p": p,
        "bal": bal,
        "acc": acc,
        "acc0": acc0,
        "acc1": acc1,
        "n_imp": len(pi_list),
        "pi": pi_list,
        "thr": t,
        "idx": feature_idx,
    }

full_idx = list(range(X.shape[1]))
base = run_logic_pipeline_with_features(full_idx)
print("Base (all features):", {k: base[k] for k in ["p","bal","acc","acc0","acc1","n_imp"]})

# greedy removal
current = full_idx.copy()
current_res = base
improved = True
while improved and len(current) > 1:
    improved = False
    best_candidate = None
    for j in current:
        cand_idx = [i for i in current if i != j]
        res = run_logic_pipeline_with_features(cand_idx)
        # keep if not worse in bal, tie-break by fewer features then fewer implicants
        if res["bal"] >= current_res["bal"] - 1e-12:
            if best_candidate is None:
                best_candidate = res
            else:
                key = (res["bal"], -res["p"], -res["n_imp"])
                key_best = (best_candidate["bal"], -best_candidate["p"], -best_candidate["n_imp"])
                if key > key_best:
                    best_candidate = res

    if best_candidate is not None and best_candidate["p"] < current_res["p"]:
        current = best_candidate["idx"]
        current_res = best_candidate
        improved = True

print("Greedy reduced:", {k: current_res[k] for k in ["p","bal","acc","acc0","acc1","n_imp"]})
print("Selected features:")
for i in current:
    print("-", features[i])

# OPTIONAL: полный перебор 256 подмножеств (для 8 признаков)
# Включить = True если хочешь получить таблицу по всем вариантам.
DO_EXHAUSTIVE = False
if DO_EXHAUSTIVE:
    results = []
    p0 = X.shape[1]
    for mask in range(1, 2**p0):
        idx = [i for i in range(p0) if (mask >> (p0 - 1 - i)) & 1]
        res = run_logic_pipeline_with_features(idx)
        results.append({"mask": mask, "p": res["p"], "bal": res["bal"], "acc": res["acc"], "n_imp": res["n_imp"]})
    resdf = pd.DataFrame(results).sort_values(["bal","p"],[False, True]).head(15)
    display(resdf)
```

## Что считаем «адекватностью» в этом ноутбуке

- Если по бинаризованным признакам удаётся построить **детерминированное** правило (логическую формулу), которое на контрольной выборке даёт **высокую точность отдельно по 0 и по 1** (и высокую balanced accuracy), то результат кластеризации можно считать **устойчиво объяснимым** через выбранную бинаризацию.

- Если при «отцеплении» признака качество не меняется, то признак **не влияет** на логическую модель (его можно убрать без потери адекватности). Если качество резко падает, то признак **существенен**.

- Сравнение с `Outcome` остаётся **справочным**: кластеризация обучалась без `Outcome`, но при желании можно дополнительно сравнить формулу с `Outcome` как внешнюю проверку.
