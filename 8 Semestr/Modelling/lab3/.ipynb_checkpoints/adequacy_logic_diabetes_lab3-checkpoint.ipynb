# Проверка адекватности: логическая модель (СДНФ) и сравнение с кластеризацией

## Связь с `modeling_process_lab3.ipynb`

В основном ноутбуке кластеризация строится в **пространстве композитных признаков** после сокращения размерности (часто получается **мало измерений**, например \(m=2\)). Это отдельная постановка: геометрия кластеров в сниженной размерности.

**Здесь** — проверка адекватности в духе примеров `lab_3point.py` / `lab_3point2.py`:

- целевой признак \(y\) — **метки k-means** (\(K=2\)) на **исходных (отобранных) числовых признаках** после стандартизации;
- признаки логической модели — **бинаризация** тех же столбцов (порог = медиана **рабочей** выборки);
- на рабочей строится **СДНФ** в виде таблицы «вектор \(\to\) класс» (мажоритарное правило для дубликатов);
- на контрольной считается **эталон `label`** — предсказание полной модели по всем признакам;
- для каждого непустого подмножества признаков проверяется **совпадение** с `label` и с истинными метками k-means на контроле.

`Outcome` в построении \(y\) **не используется** (только справочно при желании).

```python
import numpy as np
import pandas as pd
from itertools import combinations
from collections import defaultdict

np.set_printoptions(suppress=True, precision=4)
pd.set_option("display.max_columns", 20)
pd.set_option("display.width", 120)
```

## 1. Загрузка данных и обозначения признаков

Как в `lab_3point.py`: переименование в `k1..k8`, по желанию **исключаем k7** (DiabetesPedigreeFunction), остаётся **7 признаков** → \(2^7-1 = 127\) непустых подмножеств. Для **8 признаков** установите `USE_8_FEATURES = True` (тогда \(2^8-1 = 255\) вариантов).

```python
CSVPATH = "diabetes.csv"
df = pd.read_csv(CSVPATH)

TARGET_ORIG = "Outcome"
orig_feature_cols = [c for c in df.columns if c != TARGET_ORIG]
rename_map = {col: f"k{i+1}" for i, col in enumerate(orig_feature_cols)}
rename_map[TARGET_ORIG] = "Outcome"
df = df.rename(columns=rename_map)

# True = все 8 признаков; False = 7 признаков без k7 (как в lab_3point.py)
USE_8_FEATURES = False

if USE_8_FEATURES:
    SELECTED = [f"k{i}" for i in range(1, 9)]
else:
    SELECTED = ["k1", "k2", "k3", "k4", "k5", "k6", "k8"]  # без k7

N_FEAT = len(SELECTED)
print("Признаки:", SELECTED)
print("Число признаков для логики:", N_FEAT, "(подмножеств:", 2**N_FEAT - 1, ")")

X_sel = df[SELECTED].to_numpy(dtype=float)
sel_means = X_sel.mean(axis=0)
sel_stds = X_sel.std(axis=0)
sel_stds[sel_stds == 0] = 1.0
X_scaled = (X_sel - sel_means) / sel_stds
```

## 2. K-means \(K=2\) (как в `lab_3point.py`: множественные старты)

Метки кластеров по всему датасету — это **target** \(y\) для логической модели.

```python
def _euclidean(a, b):
    d = a - b
    return float(np.sqrt(np.dot(d, d)))


def _assign(X, centers):
    labels = np.empty(X.shape[0], dtype=int)
    for i in range(X.shape[0]):
        dists = [_euclidean(X[i], centers[k]) for k in range(len(centers))]
        labels[i] = int(np.argmin(dists))
    return labels


def _centers(X, labels, k):
    c = np.zeros((k, X.shape[1]))
    for j in range(k):
        m = labels == j
        if m.any():
            c[j] = X[m].mean(axis=0)
    return c


def _wcss(X, labels, centers):
    return sum(_euclidean(X[i], centers[labels[i]]) ** 2 for i in range(X.shape[0]))


def _kmeans_run(X, k, max_iter, rng):
    idx = rng.choice(X.shape[0], size=k, replace=False)
    centers = X[idx].copy()
    labels = np.full(X.shape[0], -1, dtype=int)
    for _ in range(max_iter):
        new_lbl = _assign(X, centers)
        if np.array_equal(new_lbl, labels):
            break
        labels = new_lbl
        centers = _centers(X, labels, k)
    return labels, centers, _wcss(X, labels, centers)


def kmeans(X, k, n_init=10, max_iter=300, seed=42):
    best_lbl, best_ctr, best_w = None, None, np.inf
    rng_master = np.random.default_rng(seed)
    for _ in range(n_init):
        rng = np.random.default_rng(rng_master.integers(0, 10**9))
        lbl, ctr, w = _kmeans_run(X, k, max_iter, rng)
        if w < best_w:
            best_lbl, best_ctr, best_w = lbl.copy(), ctr.copy(), w
    return best_lbl, best_ctr, best_w


K = 2
y_full, centers_full, wcss_full = kmeans(X_scaled, k=K, n_init=10, seed=42)
counts = np.bincount(y_full, minlength=K)
print(f"WCSS = {wcss_full:.2f}")
for cl in range(K):
    print(f"  Кластер {cl}: {counts[cl]} ({100*counts[cl]/len(y_full):.1f}%)")

y = y_full.astype(int)
```

## 3. Разбивка 50/50: **sorted interleaving** (`lab_3point.py`)

Сортируем индексы по сумме z-оценок, затем чётные позиции → рабочая, нечётные → контрольная (без использования меток кластера при разбиении).

```python
composite = X_scaled.sum(axis=1)
sorted_idx = np.argsort(composite)

work_idx = sorted_idx[::2]
ctrl_idx = sorted_idx[1::2]

print("Рабочая:", len(work_idx), "| Контрольная:", len(ctrl_idx))

y_work = y[work_idx]
y_ctrl = y[ctrl_idx]
print("Кластеры — рабочая:", np.bincount(y_work, minlength=K))
print("Кластеры — контроль:", np.bincount(y_ctrl, minlength=K))
```

## 3b. (опционально) Разбивка как в `lab_3point2.py`: стабильная сортировка по метке кластера

Раскомментируйте блок ниже, если нужен вариант «сначала класс 0, потом 1» и чередование позиций — **стратификация по \(y\)**.

```python
USE_STRATIFIED_CLUSTER_SPLIT = False  # поставьте True для варианта lab_3point2

if USE_STRATIFIED_CLUSTER_SPLIT:
    sorted_idx = np.argsort(y, kind="stable")
    work_idx = sorted_idx[0::2]
    ctrl_idx = sorted_idx[1::2]
    y_work = y[work_idx]
    y_ctrl = y[ctrl_idx]
    print("Используется split по метке кластера (lab_3point2).")
    print("Рабочая:", len(work_idx), "| Контрольная:", len(ctrl_idx))
    print("Кластеры — рабочая:", np.bincount(y_work, minlength=K))
    print("Кластеры — контроль:", np.bincount(y_ctrl, minlength=K))
```

## 4. Бинаризация (как в `lab_3point.py`: \(x_i=1 \Leftrightarrow \) значение **строго больше** медианы рабочей выборки)

Если нужно согласовать с `modeling_process_lab3` (\(x_i=1 \Leftrightarrow \le\) медианы), поменяйте строку на `(X_sel <= medians)`.

```python
work_vals = X_sel[work_idx]
medians = np.median(work_vals, axis=0)

X_bin_all = (X_sel > medians).astype(int)
X_bin_work = X_bin_all[work_idx]
X_bin_ctrl = X_bin_all[ctrl_idx]

print("Медианы (рабочая):", dict(zip(SELECTED, medians)))
```

## 5. СДНФ: lookup-таблица и эталон `label`

```python
def build_sdnf(X_bin, y_loc, feat_mask=None):
    if feat_mask is not None:
        X_bin = X_bin[:, feat_mask]
    groups = defaultdict(list)
    for row, lbl in zip(X_bin, y_loc):
        groups[tuple(row)].append(int(lbl))
    lookup = {vec: (1 if sum(lbls) > len(lbls) / 2 else 0) for vec, lbls in groups.items()}
    default_class = 1 if y_loc.mean() >= 0.5 else 0
    return lookup, default_class


def apply_sdnf(X_bin, lookup, default_class, feat_mask=None):
    if feat_mask is not None:
        X_bin = X_bin[:, feat_mask]
    return np.array([lookup.get(tuple(row), default_class) for row in X_bin])


full_lookup, full_default = build_sdnf(X_bin_work, y_work)
label = apply_sdnf(X_bin_ctrl, full_lookup, full_default)

acc_full_vs_true = (label == y_ctrl).mean()
print("Уникальных комбинаций на рабочей:", len(full_lookup), "из", 2**N_FEAT)
print("Совпадение label с метками k-means на контроле:", f"{acc_full_vs_true:.4f}")
print("Это эталон для сравнения упрощённых моделей (acc_vs_label).")
```

## 6. Перебор всех непустых подмножеств признаков

- **acc_vs_label** — совпадение с предсказаниями полной СДНФ на контроле.
- **acc_vs_true** — совпадение с истинными метками k-means на контроле.

```python
results = []
for r in range(1, N_FEAT + 1):
    for subset in combinations(range(N_FEAT), r):
        mask = list(subset)
        lookup_sub, default_sub = build_sdnf(X_bin_work, y_work, feat_mask=mask)
        preds = apply_sdnf(X_bin_ctrl, lookup_sub, default_sub, feat_mask=mask)
        acc_vs_label = (preds == label).mean()
        acc_vs_true = (preds == y_ctrl).mean()
        X_sub_ctrl = X_bin_ctrl[:, mask]
        coverage = sum(1 for row in X_sub_ctrl if tuple(row) in lookup_sub) / len(X_sub_ctrl)
        feat_str = "+".join(SELECTED[i] for i in subset)
        results.append({
            "n": r,
            "features": feat_str,
            "p": mask,
            "acc_vs_label": acc_vs_label,
            "acc_vs_true": acc_vs_true,
            "lookup_size": len(lookup_sub),
            "coverage": coverage,
        })

results_df = pd.DataFrame(results)
results_df.sort_values(["n", "acc_vs_label"], ascending=[True, False], inplace=True)

THRESHOLD = 0.95
adequate = results_df[results_df["acc_vs_label"] >= THRESHOLD]
print("Всего подмножеств:", len(results_df))
print("Адекватных (acc_vs_label >= {:.2f}):".format(THRESHOLD), len(adequate))
if len(adequate):
    min_n = adequate["n"].min()
    print("Минимальное число признаков среди адекватных:", min_n)
    print(adequate[adequate["n"] == min_n][["features", "acc_vs_label", "acc_vs_true"]].head(10))

results_df.to_csv("tables/adequacy_sdnf_subsets.csv", index=False)
print("Сохранено: tables/adequacy_sdnf_subsets.csv")
```
