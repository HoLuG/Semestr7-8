```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from mpl_toolkits.mplot3d import Axes3D
from grid_refinement import GridRefinement

plt.rcParams['figure.figsize'] = (16, 8)
plt.rcParams['font.size'] = 10
%matplotlib inline
```

```python
def analytical_function(x, y):
    r = np.sqrt(x**2 + y**2)
    return np.sin(r)

n = 10
k = n - 1
x_min, x_max = -3.0, 3.0
y_min, y_max = -3.0, 3.0

print(f"Параметры задачи:")
print(f"  Исходная сетка: {n}×{n} УЗЛОВ (ПСЕВДОРЕГУЛЯРНАЯ сетка)")
print(f"  Ячеек: {k}×{k} = {k*k}")
print(f"  Точки для вычисления (середины ячеек): {k}×{k} = {k*k} точек")
print(f"  Область: x ∈ [{x_min}, {x_max}], y ∈ [{y_min}, {y_max}]")
```

```python
np.random.seed(8)

x_base = np.linspace(x_min, x_max, n)
y_base = np.linspace(y_min, y_max, n)

dx_avg = (x_max - x_min) / (n - 1)
dy_avg = (y_max - y_min) / (n - 1)

x_perturb = np.random.uniform(-0.15 * dx_avg, 0.15 * dx_avg, n)
y_perturb = np.random.uniform(-0.15 * dy_avg, 0.15 * dy_avg, n)

x_perturb[0] = 0.0
x_perturb[-1] = 0.0
y_perturb[0] = 0.0
y_perturb[-1] = 0.0

x_nodes = x_base + x_perturb
y_nodes = y_base + y_perturb

x_nodes = np.sort(x_nodes)
y_nodes = np.sort(y_nodes)

x_nodes[0] = x_min
x_nodes[-1] = x_max
y_nodes[0] = y_min
y_nodes[-1] = y_max

X_grid, Y_grid = np.meshgrid(x_nodes, y_nodes, indexing='ij')

Z_nodes = analytical_function(X_grid, Y_grid)

print(f"Псевдорегулярная сетка:")
print(f"  X узлы: {len(x_nodes)} точек от {x_nodes[0]:.4f} до {x_nodes[-1]:.4f}")
print(f"  Y узлы: {len(y_nodes)} точек от {y_nodes[0]:.4f} до {y_nodes[-1]:.4f}")
print(f"  Интервалы X: min = {np.min(np.diff(x_nodes)):.4f}, max = {np.max(np.diff(x_nodes)):.4f}")
print(f"  Интервалы Y: min = {np.min(np.diff(y_nodes)):.4f}, max = {np.max(np.diff(y_nodes)):.4f}")
print(f"  Матрица высот: {Z_nodes.shape[0]}×{Z_nodes.shape[1]}")
print(f"  Значения Z: min = {Z_nodes.min():.4f}, max = {Z_nodes.max():.4f}")
```

```python
fig = plt.figure(figsize=(16, 6))

ax1 = fig.add_subplot(121, projection='3d')
surf = ax1.plot_surface(X_grid, Y_grid, Z_nodes, cmap=cm.viridis, 
                        alpha=0.9, linewidth=0.3, antialiased=True)
ax1.set_xlabel('X', fontsize=11)
ax1.set_ylabel('Y', fontsize=11)
ax1.set_zlabel('Z = sin(√(x²+y²))', fontsize=11)
ax1.set_title(f'Псевдорегулярная сетка {n}×{n} узлов', fontsize=12, fontweight='bold')
plt.colorbar(surf, ax=ax1, shrink=0.6, aspect=20)

ax2 = fig.add_subplot(122)
contour = ax2.contourf(X_grid, Y_grid, Z_nodes, levels=20, cmap=cm.viridis)
# Показываем узлы сетки
ax2.scatter(X_grid, Y_grid, c='red', s=20, alpha=0.6, marker='o', label='Узлы сетки')
ax2.set_xlabel('X', fontsize=11)
ax2.set_ylabel('Y', fontsize=11)
ax2.set_title(f'Контурная карта псевдорегулярной сетки', fontsize=12, fontweight='bold')
ax2.set_aspect('equal')
ax2.legend()
plt.colorbar(contour, ax=ax2, shrink=0.8)

plt.tight_layout()
plt.show()
```

```python
grid = GridRefinement(x_nodes.copy(), y_nodes.copy(), Z_nodes.copy())

x_midpoints = []
y_midpoints = []

for i in range(k):
    x_mid = (x_nodes[i] + x_nodes[i+1]) / 2.0
    x_midpoints.append(x_mid)

for j in range(k):
    y_mid = (y_nodes[j] + y_nodes[j+1]) / 2.0
    y_midpoints.append(y_mid)

x_midpoints = np.array(x_midpoints)
y_midpoints = np.array(y_midpoints)

X_mid_grid, Y_mid_grid = np.meshgrid(x_midpoints, y_midpoints, indexing='ij')

Z_midpoints_interp = np.zeros((k, k))
for i in range(k):
    for j in range(k):
        Z_midpoints_interp[i, j] = grid.delaunay_interpolation(X_mid_grid[i, j], Y_mid_grid[i, j])

print(f"Вычислены значения в серединах ячеек:")
print(f"  Количество точек: {k}×{k} = {k*k}")
print(f"  Интерполированные значения Z: min = {Z_midpoints_interp.min():.4f}, max = {Z_midpoints_interp.max():.4f}")
```

```python
Z_midpoints_exact = analytical_function(X_mid_grid, Y_mid_grid)

print(f"Точные значения в серединах ячеек:")
print(f"  Точные значения Z: min = {Z_midpoints_exact.min():.4f}, max = {Z_midpoints_exact.max():.4f}")
```

```python
#Z_nodes_exact = analytical_function(X_grid, Y_grid)
#error_nodes = np.abs(Z_nodes - Z_nodes_exact)
#mean_error_nodes = np.mean(error_nodes)
#max_error_nodes = np.max(error_nodes)

#print(f"Погрешность в узловых точках ({n}×{n} = {n*n} точек):")
#print(f"  Средняя абсолютная погрешность: {mean_error_nodes:.10f}")
#print(f"  Максимальная погрешность: {max_error_nodes:.10f}")
```

```python
error_midpoints = np.abs(Z_midpoints_interp - Z_midpoints_exact)
min_error_midpoints = np.min(error_midpoints)
max_error_midpoints = np.max(error_midpoints)
#mean_error_midpoints = np.mean(error_midpoints)
#rms_error_midpoints = np.sqrt(np.mean(error_midpoints**2))
#print(f"Погрешность интерполяции в серединах ячеек:")
#print(f"  Средняя абсолютная погрешность: {mean_error_midpoints:.8f}")
#print(f"  Минимальная абсолютная погрешность: {min_error_midpoints:.8f}")
#print(f"  Максимальная абсолютная погрешность: {max_error_midpoints:.8f}")
#print(f"  RMS погрешность: {rms_error_midpoints:.8f}")

print(f"Погрешность интерполяции в серединах ячеек ({k}×{k} = {k*k} точек):")
print(f"  МИНИМАЛЬНАЯ абсолютная погрешность: {min_error_midpoints:.8f}")
print(f"  МАКСИМАЛЬНАЯ абсолютная погрешность: {max_error_midpoints:.8f}")
```

```python
fig = plt.figure(figsize=(18, 12))

ax1 = fig.add_subplot(2, 3, 1)
im1 = ax1.contourf(X_mid_grid, Y_mid_grid, Z_midpoints_interp, levels=20, cmap=cm.viridis)
ax1.set_xlabel('X', fontsize=10)
ax1.set_ylabel('Y', fontsize=10)
ax1.set_title(f'Интерполированные значения\nв серединах ячеек ({k}×{k})', fontsize=11, fontweight='bold')
ax1.set_aspect('equal')
plt.colorbar(im1, ax=ax1, shrink=0.8)

ax2 = fig.add_subplot(2, 3, 2)
im2 = ax2.contourf(X_mid_grid, Y_mid_grid, Z_midpoints_exact, levels=20, cmap=cm.viridis)
ax2.set_xlabel('X', fontsize=10)
ax2.set_ylabel('Y', fontsize=10)
ax2.set_title(f'Точные значения\nв серединах ячеек ({k}×{k})', fontsize=11, fontweight='bold')
ax2.set_aspect('equal')
plt.colorbar(im2, ax=ax2, shrink=0.8)

ax3 = fig.add_subplot(2, 3, 3)
im3 = ax3.contourf(X_mid_grid, Y_mid_grid, error_midpoints, levels=20, cmap=cm.Reds)
ax3.set_xlabel('X', fontsize=10)
ax3.set_ylabel('Y', fontsize=10)
ax3.set_title(f'Погрешность интерполяции\nMin: {min_error_midpoints:.6f}, Max: {max_error_midpoints:.6f}', fontsize=11, fontweight='bold')
ax3.set_aspect('equal')
plt.colorbar(im3, ax=ax3, shrink=0.8)

ax4 = fig.add_subplot(2, 3, 4, projection='3d')
surf4 = ax4.plot_surface(X_mid_grid, Y_mid_grid, Z_midpoints_interp, cmap=cm.viridis, 
                         alpha=0.9, linewidth=0.3)
ax4.set_xlabel('X', fontsize=9)
ax4.set_ylabel('Y', fontsize=9)
ax4.set_zlabel('Z (интерполированное)', fontsize=9)
ax4.set_title('3D: Интерполированные значения', fontsize=10, fontweight='bold')

ax5 = fig.add_subplot(2, 3, 5, projection='3d')
surf5 = ax5.plot_surface(X_mid_grid, Y_mid_grid, Z_midpoints_exact, cmap=cm.viridis, 
                         alpha=0.9, linewidth=0.3)
ax5.set_xlabel('X', fontsize=9)
ax5.set_ylabel('Y', fontsize=9)
ax5.set_zlabel('Z (точное)', fontsize=9)
ax5.set_title('3D: Точные значения', fontsize=10, fontweight='bold')

ax6 = fig.add_subplot(2, 3, 6)
diff = Z_midpoints_interp - Z_midpoints_exact
im6 = ax6.contourf(X_mid_grid, Y_mid_grid, diff, levels=20, cmap=cm.RdYlBu_r)
ax6.set_xlabel('X', fontsize=10)
ax6.set_ylabel('Y', fontsize=10)
ax6.set_title(f'Разница (интерполированное - точное)\nСредняя: {np.mean(diff):.6f}', 
             fontsize=11, fontweight='bold')
ax6.set_aspect('equal')
plt.colorbar(im6, ax=ax6, shrink=0.8)

plt.tight_layout()
plt.show()
```
